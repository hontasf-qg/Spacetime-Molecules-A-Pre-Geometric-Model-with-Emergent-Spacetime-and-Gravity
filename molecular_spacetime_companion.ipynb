*Interactive companion to the paper of the same title (Physical Review D format; arXiv ancillary file).*

**Abstract.** We construct a background-free model in which spacetime is built from two species of fermionic spinor "atoms": left-handed atoms, written ${/}$, and their right-handed mirror partners, written ${\backslash}$. There is no ambient space; adjacency and geometry are properties of the atoms' states. A pair of atoms forms a bosonic *molecule* ${X}={/}{\backslash}$, and an elementary molecule is a null vector. Two flavors of atoms reconstruct a null tetrad and the Minkowski metric, so the atoms fix the local light cone and leave only a conformal scale free. The dynamics is a Wheeler--DeWitt constraint, written as a balance ${\hat{T}}|\Psi\rangle =-{\hat{V}}|\Psi\rangle$ between the binding ${\hat{V}}$ of neighboring atoms and the transfer ${\hat{T}}$ of excitations between neighboring patches. For fixed $SU(2)$ links we solve the constraint exactly. Its solutions are the molecular states, the zero-point energies of the two species cancel, the molecules are hard-core bosons that obey the oscillator algebra when dilute, and phase-coherent molecular condensates, including a thermofield double, are physical states. In a curved region the condensate's order parameter is the timelike normal of the spatial slice on each patch, while light rays are carried by the bonds. The holonomies of ${\hat{T}}$ reproduce the Cartan structure equations, and the curvature-squared functionals built from them include Starobinsky and MacDowell--Mansouri gravity. If one of these governs the continuum, the model implies two tensor gravitational-wave polarizations and a scalaron range below $10^{-4}$ m, and an $R^2$ inflaton gives $n_s\simeq0.96$--$0.97$ and $r\simeq0.003$--$0.005$. Which action the dynamics selects is open. The algebraic results are machine-checked in Lean 4 and Rocq, and every result is reproduced in a companion notebook.

## I. Introduction

The Wheeler--DeWitt equation [1, 2] expresses the invariance of general relativity under time reparametrization as a constraint on physical states, ${\hat{H}}|\Psi\rangle=0$. In the metric formulation, ${\hat{H}}$ is built from the three-metric and its conjugate momentum, and its kinetic term is governed by the DeWitt supermetric, which is indefinite. In the connection formulation of Ashtekar [3, 4, 5, 6] the same constraint becomes polynomial in a self-dual connection and a densitized triad. Both formulations take a smooth manifold as the arena and quantize the geometry living on it.

A *pre-geometric* program reverses this order: geometry should emerge from more primitive quantum constituents, and the smooth manifold, its light cones, its metric, and its curvature should be low-energy, collective properties. Sakharov's "metrical elasticity" [7], analog gravity [8, 9], the condensed-matter perspective of Volovik [10], and graph-based models such as quantum graphity [11] are well-known examples of the general idea. Spinors have long been suspected to be more primitive than vectors in this respect [12, 13].

The present paper develops a specific and, to our knowledge, new realization in which the primitive objects are two species of fermionic spinor "atoms": a left-handed atom, written ${/}$, and a right-handed atom, written ${\backslash}$. The fundamental composite is the *spacetime molecule* ${X}={/}{\backslash}$, which is a boson. Our guiding physical picture is that (a) a large number of such atoms, bound to their nearest neighbors, pair into molecules that behave like a very large collection of coupled quantum oscillators and can condense, so that at large scales the network behaves like a superfluid whose order parameter is the geometry; (b) the dynamics is a Wheeler--DeWitt constraint whose two parts are always balanced, $${\hat{H}}|\Psi\rangle=\big({\hat{T}}+{\hat{V}}\big)|\Psi\rangle=0
\quad\Longleftrightarrow\quad
{\hat{T}}|\Psi\rangle=-{\hat{V}}|\Psi\rangle,
\tag{1}$$ where ${\hat{V}}$ is the binding of adjacent atoms ("${\hat{V}}\sim{X}{X}$") and ${\hat{T}}$ is *not* a kinetic energy in the usual sense but the flow of information from one configuration of atoms and molecules to another; and (c) gravity, i.e. curvature, arises from the way neighboring patches of the network are connected, which is the job of ${\hat{T}}$.

The aim here is to turn this picture into a mathematically precise model and to test it for *internal consistency*. We are deliberately conservative: every claim is either proved in the text, verified symbolically or numerically in the companion notebook, formally verified in the companion Lean and Rocq files, or explicitly listed as open (Sec. IX). Comparison with the established literature is used only where it supplies a mathematical or observational consistency target, for example the requirement that the continuum limit reproduce an acceptable gravitational action and pass solar-system tests.

**Physical picture.** In one paragraph, the picture developed below is the following. Spacetime is modeled as a network of two species of fermionic chiral spinor atoms, ${/}$ and its mirror ${\backslash}$, with no background space: patches are labels, adjacency is combinatorial, and geometry is a property of the atoms' states. For generic spectra the only physical states of the Wheeler--DeWitt constraint are built from bosonic molecules ${X}={/}{\backslash}$, and each elementary molecule is a light ray (Theorem 2). The atoms alone fix the local light cone, leaving only a conformal scale free (Corollary 1); the zero-point energies of the two species cancel exactly (Proposition 7); and curvature appears as the holonomy of information flow, the transport generated by ${\hat{T}}$, around closed loops of patches [Eq. (33)]. Seen from far away, the network is a *molecular superfluid*. The constraint pairs fermionic atoms into bosonic molecules; dilute molecules obey the oscillator algebra and are coupled through the bonds; their phase-coherent condensates solve the constraint; and the condensate's order parameter carries the geometric data. On a single patch it is a null vector; in a curved region its site-local part is the timelike normal of the spatial slice, so the light rays must be carried by the bonds (Sec. IV E). This is the sequence (fermions, then bosonic pairs, then a superfluid) that is realized in ultracold Fermi gases [14, 15]. It also fits what is observed of spacetime itself, which carries waves [16] and is dragged by rotating masses [17]. Imposing the observed de Sitter asymptotics as a boundary condition places the model in a dS/CFT framework in which the graviton and the Standard-Model fields map to operators of definite weight on the future boundary (Sec. VI). The model then makes concrete predictions: exactly two tensor polarizations of gravitational waves, a scalaron confined by laboratory tests below about $0.1$ mm, and Starobinsky values of $n_s$ and $r$ that next-generation CMB polarization experiments, LiteBIRD among them, can test.

### A. Summary of results

1.  *Kinematics* (Sec. II). A molecule ${X}={/}{\backslash}$ is a vector if and only if the two atoms have opposite handedness (Proposition 2); a single molecule is null (Proposition 1). Two atom flavors at a patch produce four molecules forming a Newman--Penrose null tetrad, and the determinant-like pairing $-\epsilon^{\alpha\beta}\epsilon^{\gamma\delta}
    {X}_{\alpha\gamma}{X}_{\beta\delta}$ reproduces $\eta^{ab}$ times a conformal factor (Proposition 3). For any number of atoms the local light cone is rigid (Corollary 1); all non-conformal geometry must therefore live on the links between patches. Two molecules on a single-flavor patch form the Lorentz scalar that carries the conformal mode; the traceless $(1,1)$ appears only with two flavors, built from the same flavor pairing as the metric (Proposition 4).

2.  *Dynamics* (Sec. III). A constraint built from two positive pieces has no nontrivial solutions (Theorem 1). We construct a mirror-antisymmetric network of fermionic atoms, Eq. (13), that realizes Eq. (1) with ${\hat{V}}$ diagonal (binding) and ${\hat{T}}$ off-diagonal (gauge-covariant information transfer) in the configuration basis.

3.  *Exact solution* (Sec. IV). The constraint is diagonalized exactly; the kernel contains all multi-molecule states and, for generic spectra, coincides with the molecular Fock space (Theorem 2); zero-point energies cancel identically (Proposition 7); ${SU(2)}$ transport enforces Kramers doublets (Proposition 6).

4.  *Molecular superfluid* (Secs. IV C--IV E). The atoms obey Fermi statistics and the molecules are hard-core bosons (Proposition 8), which obey the oscillator algebra exactly in the dilute limit (Proposition 9). Molecular condensates of BCS type solve the constraint and carry a null order parameter (Propositions 10 and 11). The molecular order parameter obeys a linear field equation, the balance ${\hat{T}}=-{\hat{V}}$ at the level of the condensate (Proposition 13). In a curved region its site-local part is forced to be the timelike normal of the spatial slice, so null directions live on the bonds (Corollary 2).

5.  *Geometry* (Sec. V). With frames supplied by the atoms and transporters supplied by ${\hat{T}}$, the first Cartan structure equation is the statement that transported edge vectors close, and the second is the statement that plaquette holonomies of ${\hat{T}}$ equal the curvature 2-form. Curvature is detected gauge-invariantly by the spectrum of ${\hat{T}}$ (Proposition 14).

6.  *Gravitational functionals* (Secs. V G--V H). The constraint has no intrinsic scale (Proposition 18), so quadratic curvature functionals are the natural continuum actions. The metric-weighted holonomy functional gives Stelle-type gravity, $2C^2+R^2/3$ modulo the Euler density (Proposition 15); chirality adds only a Pontryagin term (Proposition 16); the de Sitter-shifted (MacDowell--Mansouri) functional gives exactly Einstein--Hilbert plus $\Lambda$ (Proposition 17), and in spinor form its connection combines the transporters of ${\hat{T}}$ with the tetrad as the ${/}\leftrightarrow{\backslash}$ mixing block. For an $\alpha R+\beta R^2$ sector the scalaron mass is $\alpha/6\beta$, so the linear term cannot vanish at low energies, and the Stelle-type channel on its own always contains a tachyon (Proposition 19).

7.  *Observational consequences* (Sec. VII). Luminal tensor waves, no long-range scalar, a scalaron confined by laboratory tests to sub-millimeter range, exactly two tensor polarizations at gravitational-wave frequencies, Starobinsky values of $n_s$ and $r$, $w=-1$, and the single-field consistency relation for $f_{\rm NL}$.

8.  *de Sitter asymptotics and dS/CFT* (Sec. VI). Imposing the observed de Sitter asymptotics as a boundary condition (Postulate 1) places the model in the dS/CFT framework. The atoms carry the boundary conformal algebra (Proposition 20), the Einstein--Hilbert coefficient is the de Sitter entropy [Eq. (50)], Standard-Model fields map to boundary operators (Table IV), and the $R^2$ channel is Starobinsky's inflaton. A shadow dictionary between the mirror atoms and boundary operators is proposed (Conjecture 1).

9.  *Formal verification* (Sec. VIII). The algebraic heart of the exact solution is checked in Lean 4, and independently in Rocq: the anticommutation relations of the atoms, the bosonic algebra of the molecules, the main theorem that every multi-molecule state and every BCS-type condensate solves the constraint, the molecular field equation, the scale freedom of the constraint, and the sign argument behind Proposition 19, together with the spinor identities of Sec. II, the boundary conformal algebra, and the algebraic identities of Secs. V H and VI.

### B. What “background-free” means here

In this paper a *background* is a pre-existing spacetime in which the constituents live. The model has none. The atoms are not located anywhere: there is no ambient 3+1-dimensional manifold, no coordinate grid, and no metric into which the patches are embedded. A patch is a label $i$ for a set of atomic modes. Adjacency is abstract combinatorial data, namely which pairs of patches share a bond, and not a spatial relation in some larger space. The Minkowski metric, the light cone, Lorentzian signature, and the number of dimensions are derived from the atoms (Sec. II). Coordinates appear only in Sec. V, and only to compare the emergent geometry with known continuum spacetimes; the hypercubic lattices used there are test discretizations of those spacetimes, not part of the model.

Three structures are nevertheless held fixed in the minimal model: the abstract graph $\Gamma$, the Fock inner product (which selects the compact subgroup $SU(2)$; Sec. III E), and, until Sec. V H, the link variables $U_{ij}$. None of them is a spacetime. Making them dynamical belongs to the open program, but is not required for background freedom in the sense used here.

### C. Conventions

We follow the conventions of Carroll [18] throughout. The metric signature is $(-,+,+,+)$, with $\eta_{ab}=\mathrm{diag}(-1,+1,+1,+1)$. Greek indices $\mu,\nu,\ldots$ are coordinate indices; Latin indices $a,b,\ldots$ from the beginning of the alphabet are orthonormal-frame (Lorentz) indices. The vierbein is $e^a{}_\mu$ with inverse $e_a{}^\mu$, so that $g_{\mu\nu}=e^a{}_\mu e^b{}_\nu\eta_{ab}$. The spin connection is $\omega_\mu{}^a{}_b$, with $\omega_{\mu ab}=-\omega_{\mu ba}$. Differential forms follow Carroll's normalization, e.g. $(\mathrm{d} A)_{\mu\nu}=\partial_\mu A_\nu-\partial_\nu A_\mu$. Capital Latin indices $A,B\in\{0,1\}$ are unprimed (left-handed) spinor indices and $A',B'$ primed (right-handed) ones. Greek indices $\alpha,\beta,\ldots\in\{1,2,\ldots\}$ label atom *flavors*, and $i,j$ label patches (vertices of a graph $\Gamma$); $\langle ij\rangle$ denotes a link. The antisymmetric symbols satisfy $\epsilon_{01}=+1$, $\epsilon_{0'1'}=+1$ and $\epsilon^{12}=+1$ for flavor; the frame Levi-Civita symbol has $\epsilon_{0123}=+1$. We set $\hbar=c=1$.

### Notebook setup
Imports, the conventions of Sec. I C, and the check registry. Every verification below calls
`check(reference, language, passed, detail)`; the registry is summarized at the end of Sec. IX.
Set `RUN_HEAVY = False` to skip the three slowest symbolic checks (each has a lighter fallback).

```python
import numpy as np, itertools, math, os, re, shutil, subprocess, hashlib, time
import sympy as sp
from scipy.linalg import expm, null_space
from scipy.stats import unitary_group
from scipy.special import iv as besselI
from scipy.sparse import csr_matrix, identity as sp_identity, kron as sp_kron
import matplotlib.pyplot as plt
plt.rcParams['pdf.fonttype'] = 42; plt.rcParams['ps.fonttype'] = 42     # embed TrueType, not Type 3, fonts in figures
np.set_printoptions(precision=6, suppress=True)
RUN_HEAVY = True          # False skips the slowest symbolic checks (lighter fallbacks are run instead)
try:
    import ipywidgets as W
    HAVE_WIDGETS = True
except Exception:
    HAVE_WIDGETS = False
T_START = time.time()

CHECKS = []
def check(ref, lang, ok, detail=""):
    # lang: 'S' computer algebra, 'N' numerical linear algebra, 'L' Lean 4, 'R' Rocq; ok=None means skipped
    status = None if ok is None else bool(ok)
    CHECKS.append(dict(ref=ref, lang=lang, ok=status, detail=detail))
    tag = {True: "PASS", False: "FAIL", None: "SKIPPED"}[status]
    print(f"[{lang}] {ref}: {tag}" + (f"   ({detail})" if detail else ""))

# Conventions (Sec. I C): eta = diag(-1,+1,+1,+1); eps_{01} = +1; soldering X^{AA'} = X^a sigma_a / sqrt(2)
s0 = np.eye(2, dtype=complex); sx = np.array([[0, 1], [1, 0]], complex)
sy = np.array([[0, -1j], [1j, 0]]); sz = np.diag([1., -1]).astype(complex)
SIG = [s0, sx, sy, sz]; PS = [sx, sy, sz]
eta = np.diag([-1., 1, 1, 1])
EPS = np.array([[0, 1], [-1, 0]], float)
def vec(M): return np.array([np.trace(s @ M) / np.sqrt(2) for s in SIG])
def mat(v): return sum(v[a] * SIG[a] for a in range(4)) / np.sqrt(2)
def dot(u, v): return u @ eta @ v
# the same objects in exact arithmetic
etaS = sp.diag(-1, 1, 1, 1); EPSs = sp.Matrix([[0, 1], [-1, 0]])
SIGs = [sp.eye(2), sp.Matrix([[0, 1], [1, 0]]), sp.Matrix([[0, -sp.I], [sp.I, 0]]), sp.Matrix([[1, 0], [0, -1]])]
def matS(v): return sum((v[a] * SIGs[a] for a in range(4)), sp.zeros(2)) / sp.sqrt(2)
def vecS(M): return [sp.expand((SIGs[a] * M).trace() / sp.sqrt(2)) for a in range(4)]
# Fermionic atoms: Jordan-Wigner annihilation operators of nm modes (mode 0 first), as sparse real matrices,
# so that the creation operator is the transpose.  The vacuum is basis vector 0.
def jw_ops(nm):
    a_ = csr_matrix(np.array([[0., 1.], [0., 0.]])); Z_ = csr_matrix(np.diag([1., -1.])); I_ = sp_identity(2, format='csr')
    ops_ = []
    for k_ in range(nm):
        m_ = None
        for j_ in range(nm):
            f_ = Z_ if j_ < k_ else (a_ if j_ == k_ else I_)
            m_ = f_ if m_ is None else sp_kron(m_, f_, format='csr')
        ops_.append(m_.tocsr())
    return ops_
def anti(A_, B_): return A_ @ B_ + B_ @ A_
def comm(A_, B_): return A_ @ B_ - B_ @ A_
def spnorm(M_): return 0.0 if M_.nnz == 0 else float(abs(M_).max())
print("numpy", np.__version__, "| sympy", sp.__version__, "| widgets:", HAVE_WIDGETS, "| RUN_HEAVY:", RUN_HEAVY)
```

## II. Atoms and molecules: kinematics

### A. Spinor atoms

At each patch $i$ there is a two-dimensional complex space $S_i$ of left-handed atom states, spanned by ${/}^A$, and its complex-conjugate space $\bar S_i$ of right-handed atom states ${\backslash}^{A'}$. The *mirror map* ${/}\leftrightarrow{\backslash}$ is complex conjugation, ${\backslash}^{A'}=\overline{{/}^{A}}$. The only structure assumed on $S_i$ is the antisymmetric pairing $$\omega(\psi,\phi)\equiv\epsilon_{AB}\,\psi^A\phi^B ,
\tag{2}$$ which is preserved by ${SL(2,\mathbb{C})}$, $\psi\mapsto L\psi$, $\det L=1$. Right-handed atoms transform with $\bar L$. No metric, coordinates, or manifold structure is assumed: $\epsilon$ is a symplectic form on a two-dimensional space, not a geometric object on spacetime.

### B. Molecules and the soldering map

A molecule is the composite $${X}^{AA'}={/}^A\,{\backslash}^{A'} .
\tag{3}$$ The set of Hermitian $2\times2$ matrices $\mathcal{M}_i$ at a patch is a four-dimensional real vector space; we identify it with an internal Minkowski space by the soldering map $${X}^{AA'}=\tfrac{1}{\sqrt2}\,{X}^a\,(\sigma_a)^{AA'},\qquad
\sigma_a=({\mathbf{1}},\sigma_x,\sigma_y,\sigma_z),
\tag{4}$$ so that ${X}^a=\tfrac{1}{\sqrt2}\operatorname{tr}(\sigma_a{X})$. A direct computation gives $$\begin{gathered}
\eta_{ab}\,{X}^a\,Y^b=-\,\epsilon_{AB}\,\epsilon_{A'B'}\,{X}^{AA'}Y^{BB'},\\
\det {X}=-\tfrac12\,\eta_{ab}{X}^a{X}^b .
\end{gathered}
\tag{5}$$ The Minkowski inner product is therefore not an input: it is the unique bilinear form on molecules built from the atom pairing (2), and its Lorentzian signature is the signature of the determinant on Hermitian $2\times2$ matrices.

**Verification of Eq. (5).** [N] on random vectors; [S] as a polynomial identity in $X^a, Y^a$.

```python
rng = np.random.default_rng(1)
X = rng.normal(size=4); Y = rng.normal(size=4)
lhs = dot(X, Y); rhs = -np.einsum('ab,cd,ac,bd->', EPS, EPS, mat(X), mat(Y)).real
check("Eq. (5), random vectors", "N",
      abs(lhs - rhs) < 1e-12 and abs(np.linalg.det(mat(X)).real + dot(X, X) / 2) < 1e-12,
      f"eta(X,Y)={lhs:.6f}, -eps eps X Y={rhs:.6f}")
Xs = sp.symbols('X0:4', real=True); Ys = sp.symbols('Y0:4', real=True)
MX, MY = matS(Xs), matS(Ys)
pair = -sum(EPSs[A, B] * EPSs[C, D] * MX[A, C] * MY[B, D] for A in range(2) for B in range(2) for C in range(2) for D in range(2))
etaXY = sum(etaS[a, a] * Xs[a] * Ys[a] for a in range(4))
check("Eq. (5), identity in X^a, Y^a; det X = -eta(X,X)/2", "S",
      sp.expand(pair - etaXY) == 0 and sp.expand(MX.det() + etaXY.subs(dict(zip(Ys, Xs))) / 2) == 0)
```

**Proposition 1 (Single molecules are null).** For any left atom $\xi$ with mirror partner $\bar\xi$, the molecule $\ell^{AA'}=\xi^A\bar\xi^{A'}$ is a real, future-directed null vector, and it is invariant under the phase $\xi\to e^{i\chi}\xi$.

*Proof.* $\ell$ is Hermitian and of rank one, so $\det\ell=0$ and by Eq. (5) $\eta_{ab}\ell^a\ell^b=0$. Its trace $|\xi^0|^2+|\xi^1|^2>0$ gives $\ell^0>0$. The phase cancels between $\xi$ and $\bar\xi$. $\square$

Physically, every elementary (factorized) molecule is a light ray. Timelike and spacelike directions appear only for combinations of molecules: the sum of two non-parallel future null vectors is timelike, and a superposition of molecular states can have a rank-two, non-null order parameter (Corollary 2).

**Verification of Proposition 1.** A molecule $X^{AA'}=\xi^A\bar\xi^{A'}$ is real, null and future pointing; the atom's phase drops out.

```python
rng = np.random.default_rng(2)
xi = rng.normal(size=2) + 1j * rng.normal(size=2)
Xv = vec(np.outer(xi, xi.conj()))
check("Prop. 1: molecule is real, null, future pointing", "N",
      abs(Xv.imag).max() < 1e-14 and abs(dot(Xv, Xv)) < 1e-12 and Xv[0].real > 0,
      f"X^a={np.round(Xv.real, 4)}, eta(X,X)={dot(Xv, Xv).real:.1e}")
check("Prop. 1: phase of the atom is invisible", "N",
      np.allclose(vec(np.outer(np.exp(0.7j) * xi, (np.exp(0.7j) * xi).conj())), Xv))
p_, q_, u_, w_ = sp.symbols('p q u w', real=True)
xs = sp.Matrix([p_ + sp.I * q_, u_ + sp.I * w_])
check("Prop. 1: det(xi xi^dagger) = 0 identically", "S", sp.simplify((xs * xs.H).det()) == 0)
```

### C. Why the atoms must have opposite handedness

**Proposition 2 (Handedness).** A composite of two atoms transforms as a Lorentz vector if and only if the atoms have opposite handedness.

*Proof.* Under ${SL(2,\mathbb{C})}$, $(\tfrac12,0)\otimes(0,\tfrac12)=(\tfrac12,\tfrac12)$, the vector. In contrast, $(\tfrac12,0)\otimes(\tfrac12,0)=(0,0)\oplus(1,0)$: the antisymmetric part is the invariant $\epsilon_{AB}\psi^A\phi^B$ and the symmetric part is a self-dual 2-form. The case $(0,\tfrac12)\otimes(0,\tfrac12)$ is the mirror image. $\square$

The notation already encodes this: ${/}$ and ${\backslash}$ are mirror images, and the requirement that ${X}$ be a vector *forces* the mirror relation to be chirality.

**Verification of Proposition 2.** Under $L\in SL(2,\mathbb C)$ a left atom transforms by $L$ and a right atom by $\bar L$. Opposite handedness gives $X\to LXL^\dagger$, a Lorentz vector; equal handedness gives a singlet plus a symmetric (self-dual) part, and no vector.

```python
rng = np.random.default_rng(3)
def rand_sl2c(r):
    M = r.normal(size=(2, 2)) + 1j * r.normal(size=(2, 2)); return M / np.sqrt(np.linalg.det(M))
L = rand_sl2c(rng)
psi, chi, phi = [rng.normal(size=2) + 1j * rng.normal(size=2) for _ in range(3)]
Xo = np.outer(psi, chi); Xo2 = np.outer(L @ psi, L.conj() @ chi)
check("Prop. 2: opposite handedness -> X -> L X L^dagger, det preserved", "N",
      np.allclose(Xo2, L @ Xo @ L.conj().T) and np.isclose(np.linalg.det(Xo2), np.linalg.det(Xo)))
sing = psi @ EPS @ phi; sing2 = (L @ psi) @ EPS @ (L @ phi)
Ssym = lambda a, b: (np.outer(a, b) + np.outer(b, a)) / 2
# the symmetric part transforms with L (x) L: it is a 3-dimensional (1,0) object, not a 4-vector
check("Prop. 2: equal handedness -> invariant singlet + symmetric (1,0) part", "N",
      np.isclose(sing, sing2) and np.allclose(Ssym(L @ psi, L @ phi), L @ Ssym(psi, phi) @ L.T))
```

### D. The null tetrad and the metric from a dyad of atoms

Suppose a patch carries two flavors of left atoms, $o\equiv{/}_1$ and $\iota\equiv{/}_2$, with mirror partners $\bar o\equiv{\backslash}_1$, $\bar\iota\equiv{\backslash}_2$, and define $$\omega_{12}\equiv\epsilon_{AB}\,o^A\iota^B .
\tag{6}$$ The four molecules $${X}_{\alpha\beta}={/}_\alpha{\backslash}_\beta:\quad
l={X}_{11},\ n={X}_{22},\ m={X}_{12},\ \bar m={X}_{21}
\tag{7}$$ are all null; $l,n$ are real and $m,\bar m$ are complex conjugates. They form a Newman--Penrose null tetrad [19].

**Proposition 3 (Tetrad and metric from atoms).** For $\omega_{12}\neq0$, $$-\,\epsilon^{\alpha\beta}\epsilon^{\gamma\delta}\,{X}^a_{\alpha\gamma}\,{X}^b_{\beta\delta}
=|\omega_{12}|^2\,\eta^{ab}.
\tag{8}$$ For the normalized dyad ($\omega_{12}=1$), $l\cdot n=-1$, $m\cdot\bar m=+1$ and all other inner products vanish, i.e. $\eta^{ab}=-(l^an^b+n^al^b)+m^a\bar m^b+\bar m^a m^b$, and $$E_0=\tfrac{l+n}{\sqrt2},\quad E_3=\tfrac{l-n}{\sqrt2},\quad
E_1=\tfrac{m+\bar m}{\sqrt2},\quad E_2=\tfrac{i(m-\bar m)}{\sqrt2}
\tag{9}$$ is an orthonormal frame, $\eta(E_a,E_b)=\eta_{ab}$.

*Proof.* Expanding the flavor sums, the left-hand side equals $-(l\otimes n+n\otimes l-m\otimes\bar m-\bar m\otimes m)$. A two-dimensional antisymmetric tensor is proportional to $\epsilon$, so $o^A\iota^B-\iota^Ao^B=\omega_{12}\,\epsilon^{AB}$ and $\bar o^{A'}\bar\iota^{B'}-\bar\iota^{A'}\bar o^{B'}=\bar\omega_{12}\,\epsilon^{A'B'}$. In spinor components, $l\otimes n-m\otimes\bar m=o^A\iota^B(\bar o^{A'}\bar\iota^{B'}-\bar\iota^{A'}\bar o^{B'})
=\bar\omega_{12}\,o^A\iota^B\epsilon^{A'B'}$ and $n\otimes l-\bar m\otimes m=-\bar\omega_{12}\,\iota^Ao^B\epsilon^{A'B'}$. Their sum is $|\omega_{12}|^2\epsilon^{AB}\epsilon^{A'B'}$, so the left-hand side is $-|\omega_{12}|^2\epsilon^{AB}\epsilon^{A'B'}$, which by Eq. (5) equals $|\omega_{12}|^2\eta^{ab}$. The remaining statements follow by evaluating inner products with Eq. (5). $\square$

**Verification of Proposition 3.** [N] for a random dyad, including the null-tetrad inner products and the orthonormal frame; [S] the identity $-\epsilon^{\alpha\beta}\epsilon^{\gamma\delta}X^a_{\alpha\gamma}X^b_{\beta\delta}=|\omega_{12}|^2\eta^{ab}$ for a general dyad (eight real parameters).

```python
rng = np.random.default_rng(4)
def molecules(at):
    return {(a, b): vec(np.outer(at[a], at[b].conj())) for a in range(2) for b in range(2)}
def ginv_from(at):
    Xm = molecules(at); G = np.zeros((4, 4), complex)
    for a, b, c, d in itertools.product(range(2), repeat=4):
        G += -EPS[a, b] * EPS[c, d] * np.outer(Xm[(a, c)], Xm[(b, d)])
    return G
o = rng.normal(size=2) + 1j * rng.normal(size=2); io = rng.normal(size=2) + 1j * rng.normal(size=2)
w12 = o @ EPS @ io
G = ginv_from([o, io])
check("Prop. 3: pairing of molecules = |omega_12|^2 eta^{ab}", "N",
      np.abs(G / abs(w12)**2 - np.linalg.inv(eta)).max() < 1e-12, f"|omega_12|^2={abs(w12)**2:.4f}")
Xm = molecules([o, io / w12]); l, n, m, mb = Xm[(0, 0)], Xm[(1, 1)], Xm[(0, 1)], Xm[(1, 0)]
ok_np = all(abs(dot(v, v)) < 1e-12 for v in (l, n, m)) and np.isclose(dot(l, n), -1) and np.isclose(dot(m, mb), 1) \
        and abs(dot(l, m)) < 1e-12 and abs(dot(n, m)) < 1e-12
E = np.array([(l + n) / np.sqrt(2), (m + mb) / np.sqrt(2), 1j * (m - mb) / np.sqrt(2), (l - n) / np.sqrt(2)])
check("Prop. 3: null tetrad (l,n,m,mbar) and orthonormal frame E_a.E_b = eta_ab", "N",
      ok_np and np.allclose((E @ eta @ E.T), eta) and abs(E.imag).max() < 1e-12)
# exact identity for a general dyad
r_ = sp.symbols('o0r o0i o1r o1i i0r i0i i1r i1i', real=True)
oS = sp.Matrix([r_[0] + sp.I * r_[1], r_[2] + sp.I * r_[3]]); iS = sp.Matrix([r_[4] + sp.I * r_[5], r_[6] + sp.I * r_[7]])
atS = [oS, iS]
XmS = {(a, b): vecS(atS[a] * atS[b].H) for a in range(2) for b in range(2)}
GS = sp.zeros(4)
for a, b, c, d in itertools.product(range(2), repeat=4):
    if EPS[a, b] * EPS[c, d] != 0:
        GS += -int(EPS[a, b] * EPS[c, d]) * sp.Matrix(4, 4, lambda i, j: XmS[(a, c)][i] * XmS[(b, d)][j])
wS = (oS.T * EPSs * iS)[0]
check("Prop. 3: exact identity for a general dyad", "S",
      sp.expand(GS - sp.expand(wS * sp.conjugate(wS)) * etaS) == sp.zeros(4))
```

**Corollary 1 (The local light cone is rigid).** For $N$ flavors at a patch, summing Eq. (8) over all pairs gives $\Omega^2\eta^{ab}$ with $\Omega^2=\sum_{\alpha<\beta}|\omega_{\alpha\beta}|^2$, for every configuration of atoms. Consequently the only geometric datum that a single patch can carry is the conformal factor $\Omega^2$, which measures how strongly its atoms pair.

This is a strong constraint and, we think, the central kinematic lesson of the model: the causal structure is built into the atoms; shape, curvature, and gravitational waves cannot be properties of a single patch and must live in the relations *between* patches.

**Verification of Corollary 1.** For $N$ flavors the pairwise sum is $\Omega^2\eta^{ab}$ with $\Omega^2=\sum_{\alpha<\beta}|\omega_{\alpha\beta}|^2$: the light cone is fixed, only the scale varies.

```python
rng = np.random.default_rng(5)
for Nf in (3, 6, 10):
    atoms = [rng.normal(size=2) + 1j * rng.normal(size=2) for _ in range(Nf)]
    Gs = np.zeros((4, 4), complex); Om2 = 0
    for a, b in itertools.combinations(range(Nf), 2):
        Gs += ginv_from([atoms[a], atoms[b]]); Om2 += abs(atoms[a] @ EPS @ atoms[b])**2
    check(f"Cor. 1: N={Nf} flavors give Omega^2 eta^ab", "N",
          np.abs(Gs / Om2 - np.linalg.inv(eta)).max() < 1e-11, f"Omega^2={Om2:.3f}")
```

### E. Local Lorentz transformations are re-pairings of atoms

An ${SL(2,\mathbb{C})}$ re-pairing ${/}_\alpha\to L{/}_\alpha$ preserves $\omega_{12}$ and maps the frame (9) to $E_a\to\Lambda(L)^b{}_aE_b$ with $\Lambda(L)\in SO^+(1,3)$, leaving Eq. (8) invariant. Infinitesimally, for $L={\mathbf{1}}+S$, $\delta{X}=S{X}+{X} S^\dagger$ realizes $\delta{X}^a=\theta^a{}_b{X}^b$ with $$\begin{gathered}
S(\theta)=-\tfrac12\,\theta_{0i}\,\sigma_i+\tfrac{i}{4}\,\epsilon_{ijk}\,\theta_{ij}\,\sigma_k
=\tfrac12 A^i\sigma_i ,\\
A^i=\theta^i{}_0+\tfrac{i}{2}\epsilon_{ijk}\theta_{jk}.
\end{gathered}
\tag{10}$$ The combination $A^i$ is the self-dual part of $\theta_{ab}$: left atoms see only the self-dual half of a Lorentz transformation, right atoms only the anti-self-dual half [12, 20]. Equation (10) is verified at the group level in the companion notebook.

**Verification of Eq. (10).** [N] at the group level, $\Lambda(e^{S(\theta)})=e^{\theta}$, and invariance of $\eta^{ab}$ under re-pairing; [S] the infinitesimal statement $SX+XS^\dagger\leftrightarrow\theta^a{}_bX^b$ exactly.

```python
rng = np.random.default_rng(6)
LC = np.zeros((3, 3, 3))
for (i, j, k), s in {(0, 1, 2): 1, (1, 2, 0): 1, (2, 0, 1): 1, (0, 2, 1): -1, (2, 1, 0): -1, (1, 0, 2): -1}.items(): LC[i, j, k] = s
def S_of(th):   # th = theta_{ab}, lower indices; Eq. (Stheta)
    S = np.zeros((2, 2), complex)
    for i in range(3):
        S += -0.5 * th[0, i + 1] * PS[i]
        for j in range(3):
            for k in range(3): S += 0.25j * LC[i, j, k] * th[i + 1, j + 1] * PS[k]
    return S
def Lam_of_U(U):
    Lm = np.zeros((4, 4))
    for b in range(4):
        e_ = np.zeros(4); e_[b] = 1; Lm[:, b] = vec(U @ mat(e_) @ U.conj().T).real
    return Lm
A = rng.normal(size=(4, 4)); th = A - A.T
check("Eq. (10): Lambda(exp S(theta)) = exp(theta^a_b)", "N",
      np.abs(Lam_of_U(expm(S_of(0.4 * th))) - expm(0.4 * eta @ th)).max() < 1e-12)
Lr = rand_sl2c(rng)
check("Eq. (10): re-pairing the dyad by L leaves eta^{ab} invariant", "N", np.abs(ginv_from([Lr @ o, Lr @ io]) - G).max() < 1e-12)
# exact infinitesimal check
tS = sp.symbols('t01 t02 t03 t12 t13 t23', real=True)
thS = sp.zeros(4); idx = [(0, 1), (0, 2), (0, 3), (1, 2), (1, 3), (2, 3)]
for v_, (a, b) in zip(tS, idx): thS[a, b] = v_; thS[b, a] = -v_
LCs = lambda i, j, k: sp.LeviCivita(i, j, k)
SS = sp.zeros(2)
for i in range(3):
    SS += -sp.Rational(1, 2) * thS[0, i + 1] * SIGs[i + 1]
    for j in range(3):
        for k in range(3): SS += sp.I / 4 * LCs(i, j, k) * thS[i + 1, j + 1] * SIGs[k + 1]
dX = vecS(SS * MX + MX * SS.H)
target = list(etaS * thS * sp.Matrix(Xs))
check("Eq. (10): S X + X S^dagger = theta^a_b X^b (exact)", "S", all(sp.expand(dX[a] - target[a]) == 0 for a in range(4)))
```

### F. Two molecules on a patch: the conformal mode and the $(1,1)$ representation

The atoms are fermions (Sec. III): one quantum of $a_A$ is a spin-$\tfrac12$ state, and for a single flavor $a^\dagger_Aa^\dagger_B$ is antisymmetric in $A$ and $B$.

**Proposition 4 (Two molecules on a patch).** (a) For a single flavor, the two-molecule states $a^\dagger_Aa^\dagger_Bb^\dagger_{A'}b^\dagger_{B'}|0\rangle$ on one patch span a single state, $\epsilon^{AB}\epsilon^{A'B'}a^\dagger_Aa^\dagger_Bb^\dagger_{A'}b^\dagger_{B'}|0\rangle$. It is a Lorentz scalar: the trace part of ${X}\otimes{X}$, which carries the conformal mode. (b) For two flavors $\alpha=1,2$, the 36 two-molecule states decompose as $9\,(0,0)\oplus3\,(1,0)\oplus3\,(0,1)\oplus(1,1)$. The $(1,1)$, the symmetric--traceless rank-two tensor, is spanned by the flavor-paired states $\epsilon^{\alpha\beta}\epsilon^{\gamma\delta}a^\dagger_{\alpha A}a^\dagger_{\beta B}b^\dagger_{\gamma A'}b^\dagger_{\delta B'}|0\rangle$.

*Proof.* (a) Fermi antisymmetry gives $a^\dagger_Aa^\dagger_B=\tfrac12\epsilon_{AB}\,\epsilon^{CD}a^\dagger_Ca^\dagger_D$, a $(0,0)$; likewise for $b^\dagger_{A'}b^\dagger_{B'}$. (b) A pair $a^\dagger_{\alpha A}a^\dagger_{\beta B}$ is antisymmetric under the simultaneous exchange $(\alpha,A)\leftrightarrow(\beta,B)$. It is therefore either symmetric in flavor and antisymmetric in spin (three $(0,0)$ states) or antisymmetric in flavor, $\propto\epsilon^{\alpha\beta}$, and symmetric in spin (one $(1,0)$). The same holds for $b$-pairs with $(0,1)$, and the tensor product gives the decomposition. Its $(1,1)$ part comes from the flavor-antisymmetric pairs in both sectors. $\square$

The notebook verifies both statements with the $SU(2)_L$ and $SU(2)_R$ Casimirs. The result fits Corollary 1. A single-flavor patch can hold only the conformal factor, while the traceless, graviton-like $(1,1)$ content requires the same flavor pairing $-\epsilon^{\alpha\beta}\epsilon^{\gamma\delta}$ that produced $\eta^{ab}$ in Proposition 3. Bosonic atoms would exchange these roles; we return to this choice in Sec. IX.

**Verification of Proposition 4.** Fermionic atoms (Jordan–Wigner operators). (a) One flavor: the two-molecule states $a^\dagger_Aa^\dagger_Bb^\dagger_{A'}b^\dagger_{B'}\ket0$ span a single state with $SU(2)_L$ and $SU(2)_R$ Casimirs both $0$. (b) Two flavors: the 36 two-molecule states decompose as $9(0,0)\oplus3(1,0)\oplus3(0,1)\oplus(1,1)$, and the $(1,1)$ is exactly the span of the flavor-paired states $\epsilon^{\alpha\beta}\epsilon^{\gamma\delta}a^\dagger_{\alpha A}a^\dagger_{\beta B}b^\dagger_{\gamma A'}b^\dagger_{\delta B'}\ket0$.

```python
def Jgen(ops_):      # SU(2) generators (1/2) sum a_A^dag sigma_AB a_B, summed over flavors; ops_[f][A]
    return [0.5 * sum(P[A, B] * (o_[A].T @ o_[B]) for o_ in ops_ for A in range(2) for B in range(2)) for P in PS]
def casimir(ops_): return sum(j_ @ j_ for j_ in Jgen(ops_))
# (a) one flavor
f4 = jw_ops(4); aF = [f4[0], f4[1]]; bF = [f4[2], f4[3]]
vac4 = np.zeros(16); vac4[0] = 1
St1 = np.array([aF[A].T @ (aF[B].T @ (bF[Ap].T @ (bF[Bp].T @ vac4))) for A, B, Ap, Bp in itertools.product(range(2), repeat=4)]).T
r1 = np.linalg.matrix_rank(St1); q1 = np.linalg.svd(St1)[0][:, 0]
cL1 = q1 @ (casimir([aF]) @ q1); cR1 = q1 @ (casimir([bF]) @ q1)
check("Prop. 4(a): one flavor, two molecules = one Lorentz-scalar state (the trace)", "N",
      r1 == 1 and abs(cL1) < 1e-12 and abs(cR1) < 1e-12, f"{r1} state, J_L^2 = {cL1:.0f}, J_R^2 = {cR1:.0f}")
# (b) two flavors: a_{alpha A} -> mode 2 alpha + A, b_{alpha A'} -> 4 + 2 alpha + A'
f8 = jw_ops(8); aF2 = [[f8[0], f8[1]], [f8[2], f8[3]]]; bF2 = [[f8[4], f8[5]], [f8[6], f8[7]]]
vac8 = np.zeros(256); vac8[0] = 1
idxs = list(itertools.product(range(2), repeat=2))
St2 = np.array([aF2[al][A].T @ (aF2[be][B].T @ (bF2[ga][Ap].T @ (bF2[de][Bp].T @ vac8)))
                for (al, A), (be, B), (ga, Ap), (de, Bp) in itertools.product(idxs, repeat=4)]).T
r2 = np.linalg.matrix_rank(St2); Q2 = np.linalg.svd(St2, full_matrices=False)[0][:, :r2]
CL2 = Q2.T @ casimir(aF2) @ Q2; CR2 = Q2.T @ casimir(bF2) @ Q2
assert np.allclose(CL2 @ CR2, CR2 @ CL2)
evc = np.linalg.eigh(CL2 + 10 * CR2)
labels_ = np.round(evc[0]).astype(int)                                   # J_L^2 + 10 J_R^2
counts_ = {(int(v_ % 10), int(v_ // 10)): int(np.sum(labels_ == v_)) for v_ in np.unique(labels_)}
EPSf = np.array([[0, 1], [-1, 0]])
St11 = np.array([sum(EPSf[al, be] * EPSf[ga, de] * (aF2[al][A].T @ (aF2[be][B].T @ (bF2[ga][Ap].T @ (bF2[de][Bp].T @ vac8))))
                     for al in range(2) for be in range(2) for ga in range(2) for de in range(2))
                 for A, B, Ap, Bp in itertools.product(range(2), repeat=4)]).T
P11 = evc[1][:, labels_ == 22]; P11 = Q2 @ P11
same_span = np.linalg.matrix_rank(np.hstack([P11, St11])) == 9 == np.linalg.matrix_rank(St11)
check("Prop. 4(b): two flavors, 36 states = 9(0,0) + 3(1,0) + 3(0,1) + (1,1); (1,1) = flavor-paired states", "N",
      r2 == 36 and counts_ == {(0, 0): 9, (2, 0): 9, (0, 2): 9, (2, 2): 9} and same_span,
      f"(J_L^2, J_R^2) multiplicities {counts_}")
```

## III. The Wheeler–DeWitt constraint

### A. A no-go theorem

**Theorem 1 (No balance between positive parts).** Let ${\hat{T}},{\hat{V}}$ be self-adjoint and non-negative on a Hilbert space. If $({\hat{T}}+{\hat{V}})|\Psi\rangle =0$, then ${\hat{T}}|\Psi\rangle ={\hat{V}}|\Psi\rangle =0$.

*Proof.* $0=\langle \Psi|({\hat{T}}+{\hat{V}})|\Psi\rangle =\|{\hat{T}}^{1/2}\Psi\|^2+\|{\hat{V}}^{1/2}\Psi\|^2$, so ${\hat{T}}^{1/2}\Psi={\hat{V}}^{1/2}\Psi=0$, hence ${\hat{T}}\Psi={\hat{V}}\Psi=0$. $\square$

A literal network of ordinary harmonic oscillators, with positive kinetic and positive potential energy, therefore admits no state in which the balance (1) is nontrivial. The two pieces cannot both be non-negative; the realization below makes both of them indefinite. In geometrodynamics the indefiniteness resides in the DeWitt supermetric [1]; indefinite-metric oscillators also appear in relativistic oscillator models [21]; and in exactly soluble minisuperspace models the Wheeler--DeWitt operator is a difference of oscillators [22, 23]. Here the indefiniteness will be supplied by the mirror structure of the atoms.

**Verification of Theorem 1.** [N] For random positive $\hat T,\hat V$ without a common kernel, $\hat T+\hat V$ is invertible, so the constraint has no nonzero solution; an indefinite split does admit solutions. [L] The expectation-value form is the Lean theorem `no_go` (Sec. VIII).

```python
rng = np.random.default_rng(7)
def randpsd(n, r):
    B = rng.normal(size=(n, r)) + 1j * rng.normal(size=(n, r)); return B @ B.conj().T
Tp, Vp = randpsd(6, 3), randpsd(6, 3)
lmin = np.linalg.eigvalsh(Tp + Vp).min()
check("Thm. 1: T, V >= 0 -> T+V has no kernel", "N", lmin > 1e-8, f"min eig(T+V) = {lmin:.3f}")
Vi = np.diag([1., 2, 3, -1, -2, -3]); Ti = -Vi.copy()
check("Thm. 1: an indefinite split admits T psi = -V psi with V psi != 0", "N",
      null_space(Ti + Vi).shape[1] > 0)
```

### B. Hilbert space

Each patch $i$ carries fermionic modes $a_{iA}$ (${/}$ atoms) and $b_{iA'}$ (${\backslash}$ atoms) with $$\{a_{iA},a^\dagger_{jB}\}=\delta_{ij}\delta_{AB},\quad
\{b_{iA'},b^\dagger_{jB'}\}=\delta_{ij}\delta_{A'B'},
\tag{11}$$ all other anticommutators vanishing [24]. Spin-$\tfrac12$ atoms are fermions, as the connection between spin and statistics [25] requires once Lorentz invariance emerges; their composites, the molecules, are then bosons (Proposition 8). The flavor label of Sec. II is a spectator in the minimal dynamics (it multiplies every degeneracy below by the number of flavors) and is suppressed. The local atom-number imbalance is $\hat D_i=\hat N_{{/} i}-\hat N_{{\backslash} i}$, with $\hat N_{{/} i}=\sum_Aa^\dagger_{iA}a_{iA}$.

### C. Bonds, binding, and information flow

Each link $\langle ij\rangle$ carries a transporter $U_{ij}$ that identifies the spinor space at $j$ with that at $i$, with $U_{ji}=U_{ij}^{-1}$. We take $U_{ij}\in{SU(2)}$ in this section; see Sec. III E. The *bond operator* of the left atoms is the gauge-covariant harmonic stretching energy $$\hat B^{{/}}_{ij}=\big(a_i-U_{ij}a_j\big)^\dagger\big(a_i-U_{ij}a_j\big)\ \ge\ 0 ,
\tag{12}$$ which vanishes only when the atom at $i$ agrees with the atom transported from $j$. For right atoms, $\hat B^{{\backslash}}_{ij}$ is defined in the same way with $a\to b$ and $U_{ij}\to\bar U_{ij}$. We define the Wheeler--DeWitt constraint $$\boxed{\;
{\hat{H}}=\epsilon_0\sum_i\hat D_i\;+\;J\sum_{\langle ij\rangle}\Big(\hat B^{{/}}_{ij}-\hat B^{{\backslash}}_{ij}\Big)
\;}
\tag{13}$$ with constants $\epsilon_0>0$ (on-site binding) and $J>0$ (bond stiffness). Expanding the bonds splits ${\hat{H}}$ into the two pieces of Eq. (1): 

\begin{align}
{\hat{V}}&=\sum_i\big(\epsilon_0+Jz_i\big)\,\hat D_i,
\tag{14}\\
{\hat{T}}&=-J\sum_{\langle ij\rangle}\Big[a_i^\dagger U_{ij}a_j+a_j^\dagger U_{ij}^\dagger a_i
-\big(b_i^\dagger\bar U_{ij}b_j+b_j^\dagger\bar U_{ij}^\dagger b_i\big)\Big],
\tag{15}
\end{align}

 where $z_i$ is the coordination number (number of bonds) of patch $i$. The split is canonical: ${\hat{V}}$ is the *diagonal* part of ${\hat{H}}$ in the configuration (occupation-number) basis, the energy with which each atom is bound by its bonds, while ${\hat{T}}$ is the *off-diagonal* part, which moves excitations from patch to patch and thereby changes the configuration. This is the precise sense in which ${\hat{T}}$ is "information flow" rather than kinetic energy. Both pieces come from the same bonds: *the bond that binds adjacent molecules is also the channel along which information flows*.

Three structural properties follow by inspection.

*(i) Bonds as springs.* Equation (13) is quadratic and conserves $\hat N_{/}$ and $\hat N_{\backslash}$ separately. In each sector the single-atom operator is the discrete covariant Laplacian of Eq. (16), the operator of a harmonic lattice with springs $(q_i-q_j)^2$, and the atoms hop along the bonds with excitation-conserving ("beam-splitter") amplitudes. Because the atoms are fermions, each mode holds at most one atom; the harmonic-oscillator degrees of freedom are the molecules (Sec. IV C).

*(ii) Mirror antisymmetry.* The mirror map $\mathcal{M}$, exchanging $a\leftrightarrow b$ and $U\leftrightarrow\bar U$, satisfies $\mathcal{M}{\hat{H}}\mathcal{M}^{-1}=-{\hat{H}}$. Each of ${\hat{T}}$ and ${\hat{V}}$ is a difference of two mirror-image halves and is indefinite, which is what Theorem 1 requires.

*(iii) Local gauge covariance.* Under $a_i\to\Lambda_ia_i$, $b_i\to\bar\Lambda_ib_i$, $U_{ij}\to\Lambda_iU_{ij}\Lambda_j^\dagger$ with $\Lambda_i\in{SU(2)}$, ${\hat{H}}$, ${\hat{T}}$ and ${\hat{V}}$ are invariant. Re-pairing the atoms at a patch is a gauge transformation, as in Sec. II.

### D. First-quantized form and stability

Collecting the modes into vectors $a=(a_{iA})$ and $b=(b_{iA'})$ of length $2N$ for $N$ patches, $${\hat{H}}=a^\dagger h\,a-b^\dagger\bar h\,b,\qquad
h=\epsilon_0{\mathbf{1}}+J\,\Delta_U ,
\tag{16}$$ where $\Delta_U$ is the *connection Laplacian* of the graph: $(\Delta_U)_{ii}=z_i{\mathbf{1}}_2$, $(\Delta_U)_{ij}=-U_{ij}$, $(\Delta_U)_{ji}=-U^\dagger_{ij}$. It is the discrete covariant Laplacian $-\nabla^2$ acting on spinors.

**Proposition 5 (Stability of each sector).** $\Delta_U\ge0$, since $a^\dagger\Delta_Ua=\sum_{\langle ij\rangle}\hat B^{{/}}_{ij}$. Hence $h\ge\epsilon_0>0$, and each mirror sector separately is stable: $a^\dagger ha\ge0$ on every state.

All indefiniteness of ${\hat{H}}$ thus resides in the relative sign between the mirror sectors. The individual sectors are as well behaved as an ordinary crystal.

**Verification of Eqs. (13)–(16) and Proposition 5.** The graph used throughout is a triangle $0\!-\!1\!-\!2$ with a pendant patch $3$ attached to $0$ ($z=(3,2,2,1)$), with random $SU(2)$ links drawn from a fixed seed. The graph is abstract: no positions are assigned to the patches. Checks: $a^\dagger\Delta_Ua=\sum_{\langle ij\rangle}|a_i-U_{ij}a_j|^2\ge0$; gauge covariance; $\hat V$ diagonal and $\hat T$ off-diagonal in the configuration basis.

```python
rngG = np.random.default_rng(2026)
def su2(r=None):
    U = unitary_group.rvs(2, random_state=(r if r is not None else rngG)); return U / np.sqrt(np.linalg.det(U))
NS = 4; LINKS = [(0, 1), (1, 2), (2, 0), (0, 3)]
z = np.zeros(NS, int)
for i, j in LINKS: z[i] += 1; z[j] += 1
eps0, Jc = 1.0, 0.3
def conn_laplacian(U, links=LINKS, N=NS):
    zz = np.zeros(N, int)
    for i, j in links: zz[i] += 1; zz[j] += 1
    D = np.zeros((2 * N, 2 * N), complex)
    for i in range(N): D[2*i:2*i+2, 2*i:2*i+2] = zz[i] * np.eye(2)
    for (i, j), u in U.items():
        D[2*i:2*i+2, 2*j:2*j+2] += -u; D[2*j:2*j+2, 2*i:2*i+2] += -u.conj().T
    return D
Ucurv = {l_: su2() for l_ in LINKS}; Uflat = {l_: np.eye(2) for l_ in LINKS}
h = eps0 * np.eye(2 * NS) + Jc * conn_laplacian(Ucurv)
av = rngG.normal(size=2 * NS) + 1j * rngG.normal(size=2 * NS)
bond = sum(np.linalg.norm(av[2*i:2*i+2] - Ucurv[(i, j)] @ av[2*j:2*j+2])**2 for i, j in LINKS)
check("Prop. 5: a^dag Delta_U a = sum of bond terms >= 0", "N",
      np.isclose((av.conj() @ conn_laplacian(Ucurv) @ av).real, bond) and np.linalg.eigvalsh(conn_laplacian(Ucurv)).min() > -1e-12)
g_ = [su2() for _ in range(NS)]; Gg = np.zeros((2 * NS, 2 * NS), complex)
for i in range(NS): Gg[2*i:2*i+2, 2*i:2*i+2] = g_[i]
Ugauge = {(i, j): g_[i] @ Ucurv[(i, j)] @ g_[j].conj().T for (i, j) in LINKS}
check("Eq. (16): gauge covariance h(U^g) = G h(U) G^dagger", "N",
      np.allclose(conn_laplacian(Ugauge), Gg @ conn_laplacian(Ucurv) @ Gg.conj().T))
I8 = np.eye(2 * NS)
H11 = np.kron(h, I8) - np.kron(I8, h.conj())                    # one "/" and one "\" atom
vdiag = np.kron(np.diag(eps0 + Jc * z), np.eye(2)).astype(complex)
V11 = np.kron(vdiag, I8) - np.kron(I8, vdiag); T11 = H11 - V11
check("Eqs. (14)-(15): V diagonal, T off-diagonal in the configuration basis", "N",
      np.allclose(V11, np.diag(np.diag(V11))) and np.allclose(np.diag(T11), 0))
print("coordination z =", z, "  tr U_triangle =", round(np.trace(Ucurv[(0, 1)] @ Ucurv[(1, 2)] @ Ucurv[(2, 0)]).real, 4))
```

### E. Reality conditions

The kinematics of Sec. II is ${SL(2,\mathbb{C})}$ covariant, but Eq. (15) is Hermitian with $U_{ji}=U_{ij}^\dagger$, whereas gauge covariance requires $U_{ji}=U_{ij}^{-1}$. These agree only for $U\in{SU(2)}$. The Fock inner product therefore selects the compact subgroup, exactly as in the real Ashtekar--Barbero formulation [4, 26]. The boost part of ${SL(2,\mathbb{C})}$ acts through complexified canonical transformations, $a\to La$, $a^\dagger\to a^\dagger L^{-1}$, which preserve Eq. (11) but not the adjoint relation. This is the analog of the reality conditions of the self-dual Ashtekar variables [3, 27]. In Sec. V the classical geometry is reconstructed with general ${SL(2,\mathbb{C})}$ transporters; the quantum constraint is solved here for ${SU(2)}$. Implementing the reality conditions in the quantum theory is listed as open.

## IV. Exact solution of the constraint

### A. Normal modes

Let $hu_k=\epsilon_ku_k$ ($k=1,\ldots,2N$) with orthonormal eigenvectors $u_k$. Since $h$ is Hermitian, $\bar h\,\bar u_k=\epsilon_k\bar u_k$: the mirror sector has the same spectrum. Define $$c_k=u_k^\dagger a,\qquad d_k=u_k^{T}b ,
\tag{17}$$ which obey the anticommutation relations (11), since a unitary change of single-atom basis preserves them. Then $${\hat{H}}=\sum_{k=1}^{2N}\epsilon_k\big(c_k^\dagger c_k-d_k^\dagger d_k\big).
\tag{18}$$ Indeed $b^\dagger\bar hb=\sum_{kl}d_k^\dagger(u_k^T\bar h\bar u_l)d_l$ and $u_k^T\bar u_l=\overline{u_k^\dagger u_l}=\delta_{kl}$.

**Proposition 6 (Kramers degeneracy and mirror isospectrality).** Let $\mathcal{E}={\mathbf{1}}_N\otimes i\sigma_y$ and $\Theta=\mathcal{E}K$, with $K$ complex conjugation. For ${SU(2)}$ transporters, $[\Theta,h]=0$ and $\Theta^2=-{\mathbf{1}}$, so every eigenvalue of $h$ is (at least) doubly degenerate. Equivalently, $\bar h=\mathcal{E}^{-1}h\,\mathcal{E}$: the two mirror sectors are unitarily equivalent through the atom pairing $\epsilon=i\sigma_y$.

*Proof.* For $U\in{SU(2)}$, $\sigma_y\bar U\sigma_y=U$, so $\Theta$ maps every block of $h$ to itself; the diagonal blocks are real multiples of ${\mathbf{1}}_2$. Kramers' theorem gives the degeneracy. Writing $\Theta h=h\Theta$ as $\mathcal{E}\bar hK=h\mathcal{E}K$ gives $\bar h=\mathcal{E}^{-1}h\,\mathcal{E}$. $\square$

Kramers' theorem here is a consequence of the spinor pairing (2); no physical time-reversal symmetry is assumed. Each energy level $\lambda$ hosts a doublet $c_{\lambda s}$, $s=\pm$, and a mirror doublet $d_{\lambda s'}$.

**Verification of Proposition 6.** $\Theta=(\mathbf 1\otimes i\sigma_y)K$ commutes with $h$, $\Theta^2=-1$, and $\bar h=\mathcal E^{-1}h\,\mathcal E$ with $\mathcal E=\mathbf 1\otimes i\sigma_y$: every level is a Kramers doublet and the mirror sector is isospectral.

```python
Th = np.kron(np.eye(NS), 1j * sy)
ev = np.linalg.eigvalsh(h); evb = np.linalg.eigvalsh(h.conj())
check("Prop. 6: [h, Theta] = 0 and Theta^2 = -1", "N",
      np.abs(Th @ h.conj() - h @ Th).max() < 1e-13 and np.allclose(Th @ Th.conj(), -np.eye(2 * NS)))
check("Prop. 6: hbar = E^-1 h E (E = 1 x i sigma_y); Kramers doublets; isospectral mirror", "N",
      np.abs(np.linalg.inv(Th) @ h @ Th - h.conj()).max() < 1e-13 and np.allclose(ev[0::2], ev[1::2]) and np.allclose(ev, evb))
```

**Proposition 7 (Zero-point cancellation).** With antisymmetric (Weyl) ordering of the fermionic modes, ${\hat{H}}_{\rm Weyl}={\hat{H}}-\tfrac12\operatorname{tr} h+\tfrac12\operatorname{tr}\bar h={\hat{H}}$ for every connection $U$.

*Proof.* $\tfrac12(c_k^\dagger c_k-c_kc_k^\dagger)=c_k^\dagger c_k-\tfrac12$ and likewise for $d_k$; the constants cancel mode by mode because $\operatorname{tr}\bar h=\overline{\operatorname{tr} h}=\operatorname{tr} h$. $\square$

Each left mode carries the fermionic zero-point energy $-\epsilon_k/2$ (the Dirac-sea value), and its mirror partner, which enters ${\hat{H}}$ with the opposite sign, carries $+\epsilon_k/2$. The $2N$ zero-point energies of the left sector are therefore cancelled exactly by those of the right sector, for any graph and any connection. The mechanism resembles the boson--fermion cancellation of supersymmetry, but here it acts between two fermionic sectors of opposite constraint sign. The atoms therefore contribute no static vacuum energy. Whether the cancellation extends to the full quantum effective action once the links are dynamical, which would rule out a Sakharov-type induced action [7], is open; we return to this point in Sec. V H.

**Verification of Proposition 7.** The fermionic zero-point constant of the antisymmetrically ordered constraint, $-\tfrac12\operatorname{tr}h+\tfrac12\operatorname{tr}\bar h$, vanishes for every graph and connection. [L] and [R] theorem `zero_point_cancellation`.

```python
zp = [0.5 * (np.trace(hh) - np.trace(hh.conj())).real
      for hh in [h] + [eps0 * np.eye(2 * NS) + Jc * conn_laplacian({l_: su2() for l_ in LINKS}) for _ in range(20)]]
check("Prop. 7: zero-point constant vanishes (21 connections)", "N", max(abs(v) for v in zp) < 1e-13)
```

### B. Physical states

Let $|m,n\rangle$ denote the normal-mode Fock state with occupations $m_k\in\{0,1\}$ of the $c$ modes and $n_k\in\{0,1\}$ of the $d$ modes, and define the *molecule operators* $${X}^\dagger_{kl}=c_k^\dagger d_l^\dagger ,\qquad {X}^\dagger_k\equiv{X}^\dagger_{kk}.
\tag{19}$$ They satisfy $[{\hat{H}},{X}^\dagger_{kl}]=(\epsilon_k-\epsilon_l){X}^\dagger_{kl}$.

**Theorem 2 (Solutions of the Wheeler--DeWitt equation).** Let $E(m,n)=\sum_k\epsilon_k(m_k-n_k)$. Then:

1.  $\ker{\hat{H}}$ is the closed span of the states $|m,n\rangle$ with $E(m,n)=0$.

2.  Every multi-molecule state with distinct modes, $$|\Psi\rangle ={X}^\dagger_{k_1}\cdots{X}^\dagger_{k_r}|0\rangle  ,$$ satisfies ${\hat{H}}|\Psi\rangle =0$ and is the basis vector $|\mathrm{occ},\mathrm{occ}\rangle$, where $\mathrm{occ}$ is the indicator of $\{k_1,\ldots,k_r\}$; a repeated mode gives zero. The $2^{2N}$ molecular states are linearly independent, so $\dim\ker{\hat{H}}\ge2^{2N}$, which grows without bound with the number of patches. More generally, ${X}^\dagger_{kl}$ preserves $\ker{\hat{H}}$ whenever $\epsilon_k=\epsilon_l$.

3.  If the distinct levels $\{\lambda\}$ of $h$, with degeneracies $g_\lambda$, are rationally independent, then $\ker{\hat{H}}$ is exactly the *molecular Fock space*: the span of states obtained from $|0\rangle$ by the operators ${X}^\dagger_{\lambda ss'}=c^\dagger_{\lambda s}d^\dagger_{\lambda s'}$, of dimension $\prod_\lambda\binom{2g_\lambda}{g_\lambda}$ ($6^N$ for $N$ Kramers doublets).

*Proof.* (a) follows from Eq. (18), which is diagonal in $|m,n\rangle$ with eigenvalue $E(m,n)$. (b) Applying ${X}^\dagger_k$ raises $m_k$ and $n_k$ from $0$ to $1$ and leaves $E$ unchanged, and $E(0,0)=0$; $({X}^\dagger_k)^2=0$ by the Pauli principle. In the mode-interleaved ordering $\prod_k(c_k^\dagger)^{m_k}(d_k^\dagger)^{n_k}|0\rangle$ of the basis, the two fermionic signs of ${X}^\dagger_k=c_k^\dagger d_k^\dagger$ cancel, so the state is $|\mathrm{occ},\mathrm{occ}\rangle$ with coefficient $1$. (c) Let $M_\lambda$ and $N_\lambda$ be the total occupations of level $\lambda$ in the two sectors. Then $E=\sum_\lambda\lambda(M_\lambda-N_\lambda)$ with integer coefficients, so rational independence forces $M_\lambda=N_\lambda$ for every $\lambda$. Within a level, any monomial $c^\dagger\cdots c^\dagger d^\dagger\cdots d^\dagger$ with equal numbers of $c^\dagger$ and $d^\dagger$ factors is, up to sign, a product of molecule operators. Choosing $p$ of the $g_\lambda$ modes in each sector gives $\sum_p\binom{g_\lambda}{p}^2=\binom{2g_\lambda}{g_\lambda}$ states per level. $\square$

Theorem 2 makes precise the statement that molecules are what the constraint allows: for generic geometry, atoms are confined into ${/}{\backslash}$ pairs, and unpaired atoms are unphysical. When the spectrum is commensurate, additional resonant solutions appear, with equal energy but unequal atom numbers in the two sectors.

**Molecular multiplets.** By Proposition 6, each level carries four molecules $c^\dagger_{\lambda s}d^\dagger_{\lambda s'}$. For a flat connection, where $U$ is a pure gauge, the doublet index is the global spinor index and these four states form exactly the vector $(\tfrac12,\tfrac12)$ of Sec. II; the doubly occupied level $c^\dagger_{\lambda+}c^\dagger_{\lambda-}d^\dagger_{\lambda+}d^\dagger_{\lambda-}|0\rangle$ is the scalar of Proposition 4(a). For a curved connection Kramers' theorem preserves the dimension of the multiplet.

**The balance is nontrivial.** On a graph with nonuniform coordination, ${\hat{V}}$ does not commute with ${\hat{H}}$, and physical states satisfy Eq. (1) with ${\hat{V}}|\Psi\rangle \neq0$. Table I reports an example computed in the notebook.

**Table I.** Exact solution on the four-patch graph "triangle $0\!-\!1\!-\!2$ plus pendant $3$" ($z=(3,2,2,1)$), with fermionic atoms, random ${SU(2)}$ links, $\epsilon_0=1$, $J=0.3$. The holonomy around the triangle has $\operatorname{tr} U_\triangle=-0.996$, far from the flat value $2$. Kernel dimensions are computed in the full Fock space of $2^{16}$ states, built with Jordan--Wigner operators in the site basis. Reproduced in the companion notebook.

| Quantity                                                |          Value          |
|:--------------------------------------------------------|:-----------------------:|
| spectrum of $h$ (doubly degenerate)                     |   $1.0900,\ 1.3000,$    |
|                                                         |    $1.6720,\ 2.3380$    |
| spectrum of $\bar h$                                    |        identical        |
| $\lambda_{\min}(\Delta_U)$: flat, pure gauge            |        $0$, $0$         |
| $\lambda_{\min}(\Delta_U)$: curved                      |        $0.2999$         |

**Verification of Theorem 2 and generation of Table I.** The full fermionic Fock space of the four-patch graph ($8$ left and $8$ right atomic modes, $2^{16}=65\,536$ states) is built with Jordan–Wigner operators in the site basis, and $\hat H=a^\dagger ha-b^\dagger\bar hb$ is assembled from them. Checks: (i) the $p$-atom spectrum of $a^\dagger ha$ consists of sums of $p$ *distinct* single-atom energies (Fermi statistics); (ii) every product of normal-mode molecules on distinct modes solves the constraint, and a repeated mode gives zero; (iii) direct null spaces of the one- and two-molecule sectors; (iv) the kernel dimension in every sector $(N_/,N_\backslash)$ against the generic count $\prod_\lambda\binom{4}{2}=6^4$. [L], [R] theorems `wdw_molecular_solution`, `molecules_at_occ`, `distinct_solutions`, `molecules_repeat`.

```python
NM = 2 * NS                                     # 8 atomic modes per species
FF = jw_ops(2 * NM); aS, bS = FF[:NM], FF[NM:]  # site-basis fermions: a_0..a_7 then b_0..b_7
def onebody(hh, ops_):
    return sum(hh[i, j] * (ops_[i].T @ ops_[j]) for i in range(NM) for j in range(NM) if abs(hh[i, j]) > 1e-15)
HA, HB = onebody(h, aS), onebody(h.conj(), bS)
Hf = (HA - HB).tocsr()
dimF = 2**(2 * NM)
bits = (np.arange(dimF)[:, None] >> np.arange(2 * NM)[::-1]) & 1
nA, nB_ = bits[:, :NM].sum(1), bits[:, NM:].sum(1)
w, Uv = np.linalg.eigh(h)
# (i) Fermi statistics of the many-atom spectrum
fermi_ok = True
for p_ in range(NM + 1):
    sel = np.where((nA == p_) & (nB_ == 0))[0]
    specA = np.sort(np.linalg.eigvalsh(HA[sel][:, sel].toarray()))
    slater = np.sort([sum(c_) for c_ in itertools.combinations(w, p_)]) if p_ else np.array([0.])
    fermi_ok &= np.allclose(specA, slater, atol=1e-10)
check("Thm. 2: p-atom spectrum = sums of p distinct single-atom energies (Fermi statistics)", "N", fermi_ok)
# (ii) molecular states
vacF = np.zeros(dimF, complex); vacF[0] = 1
cdagN = [sum(Uv[i, k] * aS[i].T for i in range(NM)) for k in range(NM)]           # c_k^dag = sum_i u_k(i) a_i^dag
ddagN = [sum(Uv[j, k].conj() * bS[j].T for j in range(NM)) for k in range(NM)]    # d_k^dag = sum_j conj(u_k(j)) b_j^dag
XdagN = [(cdagN[k] @ ddagN[k]).tocsr() for k in range(NM)]
rngK = np.random.default_rng(31)
res_mol = 0.0
for _ in range(12):
    ks_ = rngK.choice(NM, size=rngK.integers(1, NM + 1), replace=False)
    v_ = vacF.copy()
    for k_ in ks_: v_ = XdagN[k_] @ v_
    res_mol = max(res_mol, np.linalg.norm(Hf @ v_) / np.linalg.norm(v_))
rep_ = np.linalg.norm(XdagN[3] @ (XdagN[3] @ (XdagN[5] @ vacF)))
check("Thm. 2(b): products of molecules on distinct modes solve H Psi = 0; a repeated mode gives 0", "N",
      res_mol < 1e-12 and rep_ < 1e-12, f"max residual {res_mol:.1e}, repeated mode {rep_:.0e}")
# (iii) direct null spaces
def sector(p_, q_):
    sel = np.where((nA == p_) & (nB_ == q_))[0]; return Hf[sel][:, sel].toarray()
dim1 = null_space(sector(1, 1)).shape[1]
dim2 = int(np.sum(np.abs(np.linalg.eigvalsh(sector(2, 2))) < 1e-9))
# (iv) every sector, from the exact tensor structure H_pq = A_p x 1 - 1 x B_q
specs = {}
for p_ in range(NM + 1):
    selA = np.where((nA == p_) & (nB_ == 0))[0]; selB = np.where((nA == 0) & (nB_ == p_))[0]
    specs[p_] = (np.linalg.eigvalsh(HA[selA][:, selA].toarray()), np.linalg.eigvalsh(HB[selB][:, selB].toarray()))
kerdim = {(p_, q_): int(np.sum(np.abs(specs[p_][0][:, None] - specs[q_][1][None, :]) < 1e-9)) for p_ in range(NM + 1) for q_ in range(NM + 1)}
nlev = len(np.unique(np.round(w, 9)))
gen = np.poly1d([1.])
for _ in range(nlev): gen = gen * np.poly1d([1., 4., 1.])
pred = {p_: int(round(c_)) for p_, c_ in enumerate(gen.coeffs[::-1])}
total = sum(kerdim.values())
check("Thm. 2(c): kernel only in sectors N_/ = N_\\ with the molecular counts (1+4x+x^2)^4", "N",
      all(kerdim[(p_, q_)] == (pred[p_] if p_ == q_ else 0) for p_ in range(NM + 1) for q_ in range(NM + 1))
      and dim1 == pred[1] and dim2 == pred[2] and total == 6**nlev,
      f"sector dims {[kerdim[(p_, p_)] for p_ in range(NM + 1)]}, total {total} = 6^{nlev}")
Psi = np.kron(Uv[:, 0], Uv[:, 0].conj())
nV, nT, nB = np.linalg.norm(V11 @ Psi), np.linalg.norm(T11 @ Psi), np.linalg.norm((T11 + V11) @ Psi)
check("Eq. (1): T Psi = -V Psi with V Psi != 0 (nonuniform graph)", "N", nB < 1e-12 and nV > 0.1 and np.isclose(nV, nT))
trU = np.trace(Ucurv[(0, 1)] @ Ucurv[(1, 2)] @ Ucurv[(2, 0)]).real
lev = ev[0::2]
TABLE1 = {"spectrum of h (doubly degenerate)": ", ".join(f"{v:.4f}" for v in lev),
          "spectrum of hbar": "identical" if np.allclose(ev, evb) else "DIFFERENT",
          "lambda_min(Delta_U): flat, pure gauge": f"{np.linalg.eigvalsh(conn_laplacian(Uflat)).min():.0f}, {abs(np.linalg.eigvalsh(conn_laplacian(Ugauge)).min()):.0f}",
          "lambda_min(Delta_U): curved": f"{np.linalg.eigvalsh(conn_laplacian(Ucurv)).min():.4f}",
          "dim ker H, one-molecule sector (64)": f"{dim1} (pred. {pred[1]})",
          "dim ker H, two-molecule sector (784)": f"{dim2} (pred. {pred[2]})",
          "dim ker H, all sectors (65536)": f"{total} (pred. 6^{nlev} = {6**nlev})",
          "||V Psi|| = ||T Psi||, lowest molecule": f"{nV:.4f}",
          "||(T+V) Psi||": f"{nB:.0e}",
          "(1/2)(tr h - tr hbar)": f"{zp[0]:.0f} (exact)",
          "tr U_triangle; rotation angle": f"{trU:.4f}; {2*np.arccos(trU/2):.2f} rad"}
print("Table I (generated):")
for k_, v_ in TABLE1.items(): print(f"  {k_:42s} {v_}")
```

### C. Molecules are bosons: hard-core algebra and the oscillator limit

The atoms obey Eq. (11); the molecules they form do not. Define the molecule annihilation operator ${X}_k=d_kc_k$, the adjoint of ${X}^\dagger_k=c^\dagger_kd^\dagger_k$, and the occupations $n^c_k=c^\dagger_kc_k$, $n^d_k=d^\dagger_kd_k$.

**Proposition 8 (Atoms are fermions, molecules are hard-core bosons).** (a) $\{c_k,c^\dagger_l\}=\{d_k,d^\dagger_l\}=\delta_{kl}$, and all creation operators of atoms anticommute; in particular $(c^\dagger_k)^2=(d^\dagger_k)^2=0$. (b) Molecule operators commute with each other, $[{X}^\dagger_k,{X}^\dagger_l]=0$, and with the atom operators of every other mode, $[{X}^\dagger_k,c^\dagger_l]=[{X}^\dagger_k,d^\dagger_l]=0$ for $l\neq k$. (c) They obey the hard-core boson algebra $$=\delta_{kl}\big(1-n^c_k-n^d_k\big),\qquad({X}^\dagger_k)^2=0 .
\tag{20}$$

*Proof.* (a) is Eq. (11) in the normal-mode basis. (b) Moving a product of two fermion operators past a third operator costs two signs. (c) For $k=l$, ${X}_k{X}^\dagger_k=(1-n^c_k)(1-n^d_k)$ and ${X}^\dagger_k{X}_k=n^c_kn^d_k$; for $k\neq l$ all four operators anticommute pairwise. $\square$

A molecule is therefore an integer-spin $(\tfrac12,\tfrac12)$ object with Bose statistics, built from two half-integer-spin fermions, as spin--statistics requires of a composite. The only departure from an ideal boson is the hard core: one molecule per mode. This is the algebra of Anderson's pseudospin description of Cooper pairs [28].

**Verification of Proposition 8.** Four normal modes per species ($256$ states). Atoms anticommute; molecules $X_k^\dagger=c_k^\dagger d_k^\dagger$ commute with each other and with the atoms of other modes; $[X_k,X_l^\dagger]=\delta_{kl}(1-n^c_k-n^d_k)$ and $(X_k^\dagger)^2=0$. [L], [R] theorems `cdag_anticomm`, `cdag_ddag_anticomm`, `molecules_commute`, `molecule_atom_comm_c`, `molecule_hardcore`, `molecule_cross`, `molecule_pauli`.

```python
Om4 = 4; F8 = jw_ops(2 * Om4); cO, dO = F8[:Om4], F8[Om4:]
Id8 = sp_identity(2**(2 * Om4), format='csr')
XO = [(dO[k] @ cO[k]).tocsr() for k in range(Om4)]; XOd = [x_.T.tocsr() for x_ in XO]
car_ok = all(spnorm(anti(cO[k], cO[l].T) - (Id8 if k == l else 0 * Id8)) < 1e-12 and spnorm(anti(cO[k].T, dO[l].T)) < 1e-12
             and spnorm(anti(dO[k], dO[l].T) - (Id8 if k == l else 0 * Id8)) < 1e-12 for k in range(Om4) for l in range(Om4))
check("Prop. 8: atoms obey the canonical anticommutation relations", "N", car_ok)
bos_ok = all(spnorm(comm(XOd[k], XOd[l])) < 1e-12 for k in range(Om4) for l in range(Om4)) and \
         all(spnorm(comm(XOd[k], cO[l].T)) < 1e-12 and spnorm(comm(XOd[k], dO[l])) < 1e-12 for k in range(Om4) for l in range(Om4) if l != k)
check("Prop. 8: molecules commute with molecules and with atoms of other modes (bosons)", "N", bos_ok)
hc_ok = all(spnorm(comm(XO[k], XOd[l]) - ((Id8 - cO[k].T @ cO[k] - dO[k].T @ dO[k]) if k == l else 0 * Id8)) < 1e-12
            for k in range(Om4) for l in range(Om4)) and all(spnorm(XOd[k] @ XOd[k]) < 1e-12 for k in range(Om4))
check("Prop. 8: [X_k, X_l^dag] = delta_kl (1 - n^c_k - n^d_k), (X_k^dag)^2 = 0 (hard core)", "N", hc_ok)
```

The hard core matters only when molecules are dense. Collect $\Omega$ modes into a set $S$ (for example, the modes of a band or of a coarse-grained region) and define the collective molecule operators $$S^+=\sum_{k\in S}{X}^\dagger_k,\qquad S^-=(S^+)^\dagger,\qquad B^\dagger=\Omega^{-1/2}S^+ .
\tag{21}$$

**Proposition 9 (Dilute molecules obey the oscillator algebra).** (a) $[S^-,S^+]=\Omega-N^S_c-N^S_d$, with $N^S_c=\sum_{k\in S}n^c_k$ and $N^S_d=\sum_{k\in S}n^d_k$. On the molecular states of $S$ (those with $n^c_k=n^d_k$), $S^\pm$ and $S^z=\hat n-\Omega/2$, with $\hat n$ the number of molecules, generate $\mathfrak{su}(2)$. The molecular states form $(\mathbb C^2)^{\otimes\Omega}$, a reducible representation; the symmetric states $(S^+)^n|0\rangle$ carry its spin-$\Omega/2$ component. (b) The normalized states $\ket n\propto(S^+)^n|0\rangle$ satisfy $$\langle n+1|B^\dagger|n\rangle=\sqrt{n+1}\,\sqrt{1-n/\Omega},\qquad [B,B^\dagger]=1-2\hat n/\Omega
\tag{22}$$ on molecular states. For $n\ll\Omega$, $B$ is an ordinary harmonic-oscillator ladder operator; the correction of relative order $n/\Omega$ is the Pauli (hard-core) repulsion between molecules. (c) The condensate $\prod_{k\in S}(1+z{X}^\dagger_k)|0\rangle$ has a binomial distribution of molecule numbers with parameter $|z|^2/(1+|z|^2)$, which tends to the Poisson distribution of an oscillator coherent state as $\Omega\to\infty$ at fixed mean.

*Proof.* (a) follows from Eq. (20) summed over $S$; on molecular states $N^S_c=N^S_d=\hat n$. (b) For spin $j=\Omega/2$ and $m=n-\Omega/2$, $\langle m+1|S^+|m\rangle^2=(j-m)(j+m+1)=(n+1)(\Omega-n)$. (c) The factors commute (Proposition 8), so each mode is independently occupied with probability $|z|^2/(1+|z|^2)$. $\square$

The oscillator algebra of the molecules is thus derived, not assumed. Dilute molecules are independent bosonic modes, and their coherent states are the $\Omega\to\infty$ limit of the molecular condensates below. The Pauli principle of the constituent atoms supplies the leading departure from the oscillator algebra. This is oscillator kinematics. The molecule operators commute with ${\hat{H}}$, so within the constraint they carry zero frequency, and oscillator dynamics with nonzero frequencies requires a relational time and dynamical links (Sec. IV E). The notebook checks Eq. (22) for $\Omega=6$ and the convergence of the binomial to the Poisson distribution.

**Verification of Proposition 9.** [N] For $\Omega=6$ modes ($4096$ states): $[S^-,S^+]=\Omega-N_c-N_d$, and on the molecular states $\ket n\propto(S^+)^n\ket0$ the ladder matrix elements are $\sqrt{(n+1)(\Omega-n)}$, i.e. those of an oscillator $\sqrt\Omega\,\sqrt{n+1}$ times $\sqrt{1-n/\Omega}$. [N] The molecule-number distribution of the condensate $\prod_k(1+zX_k^\dagger)\ket0$ is binomial and tends to the Poisson distribution of an oscillator coherent state as $\Omega\to\infty$ at fixed mean. [S] The su(2) identity behind the matrix elements.

```python
Om6 = 6; F12 = jw_ops(2 * Om6); c6, d6 = F12[:Om6], F12[Om6:]
X6 = [(d6[k] @ c6[k]).tocsr() for k in range(Om6)]
Sm = sum(X6).tocsr(); Sp = Sm.T.tocsr()
Ntot = sum(c6[k].T @ c6[k] + d6[k].T @ d6[k] for k in range(Om6))
Id12 = sp_identity(2**(2 * Om6), format='csr')
check("Prop. 9: [S^-, S^+] = Omega - N_c - N_d", "N", spnorm(comm(Sm, Sp) - (Om6 * Id12 - Ntot)) < 1e-12)
v_ = np.zeros(2**(2 * Om6)); v_[0] = 1; kets = []
for n_ in range(Om6 + 1):
    kets.append(v_ / np.linalg.norm(v_)); v_ = Sp @ v_
mel = np.array([kets[n_ + 1] @ (Sp @ kets[n_]) for n_ in range(Om6)])
check("Prop. 9: <n+1|S^+|n> = sqrt((n+1)(Omega-n)) = sqrt(Omega) sqrt(n+1) sqrt(1-n/Omega)", "N",
      np.allclose(mel, np.sqrt((np.arange(Om6) + 1) * (Om6 - np.arange(Om6)))), f"{np.round(mel, 4)}")
zc_ = 0.4
v_ = np.zeros(2**(2 * Om6)); v_[0] = 1
for k in range(Om6): v_ = v_ + zc_ * (X6[k].T @ v_)
v_ /= np.linalg.norm(v_)
Nmol = (Ntot @ v_) / 2
pn = np.array([np.sum(np.abs(v_[np.isclose(np.round(Ntot.diagonal() / 2), n_)])**2) for n_ in range(Om6 + 1)])
pz = zc_**2 / (1 + zc_**2)
binom_ = np.array([math.comb(Om6, n_) * pz**n_ * (1 - pz)**(Om6 - n_) for n_ in range(Om6 + 1)])
from scipy.stats import binom as binom_dist, poisson as poisson_dist
def tv_to_poisson(Om_, mean_=1.0):
    nn = np.arange(0, 60); return 0.5 * np.sum(np.abs(binom_dist.pmf(nn, Om_, mean_ / Om_) - poisson_dist.pmf(nn, mean_)))
tvs = {Om_: tv_to_poisson(Om_) for Om_ in (6, 60, 600, 6000)}
check("Prop. 9: condensate molecule number is binomial and -> Poisson (oscillator coherent state)", "N",
      np.allclose(pn, binom_) and all(tvs[a_] > tvs[b_] for a_, b_ in [(6, 60), (60, 600), (600, 6000)]) and tvs[6000] < 1e-3,
      "total-variation distance to Poisson at mean 1: " + ", ".join(f"Omega={k_}: {v2:.1e}" for k_, v2 in tvs.items()))
jS, mS = sp.symbols('j m')
check("Prop. 9: (j - m)(j + m + 1) = (n + 1)(Omega - n) for j = Omega/2, m = n - Omega/2 (exact)", "S",
      sp.expand(((jS - mS) * (jS + mS + 1)).subs({jS: sp.Symbol('Om') / 2, mS: sp.Symbol('n') - sp.Symbol('Om') / 2})
                - (sp.Symbol('n') + 1) * (sp.Symbol('Om') - sp.Symbol('n'))) == 0)
```

### D. Molecular condensates and the molecular field equation

Theorem 2 has an important corollary for the reconstruction of geometry: the constraint admits phase-coherent condensates of molecules, and their order parameter carries the light cone. Consider first a single patch, ${\hat{H}}=\epsilon_0(\hat N_{/}-\hat N_{\backslash})$, and for a normalized left atom $\hat\xi$ let $\hat\ell^{AA'}=\hat\xi^A\bar{\hat\xi}^{A'}$, a null vector (Proposition 1).

**Proposition 10 (Single-patch condensates carry a null direction).** For any complex $u,v$, the state $$|\Psi_\ell\rangle=\big(u+v\,{X}^\dagger_{\hat\ell}\big)|0\rangle ,
\qquad {X}^\dagger_{\hat\ell}=\hat\ell^{AA'}a^\dagger_Ab^\dagger_{A'}=(\hat\xi\cdot a^\dagger)(\bar{\hat\xi}\cdot b^\dagger) ,
\tag{23}$$ satisfies (i) ${\hat{H}}|\Psi_\ell\rangle=0$; (ii) $({X}^\dagger_{\hat\ell})^2=0$ and $\langle a_A\rangle=\langle b_{A'}\rangle=0$; and (iii) $\langle b_{A'}a_A\rangle=\Delta\,\hat\ell^{AA'}$ with $\Delta=\bar uv/(|u|^2+|v|^2)$, so the molecular order parameter is a null vector, of length $|\Delta|\le\tfrac12$.

*Proof.* (i) Every term has $N_{/}=N_{\backslash}$. (ii) $(\hat\xi\cdot a^\dagger)^2=0$ by Eq. (11); the two terms of $|\Psi_\ell\rangle$ differ by two atoms, so single-atom operators have zero expectation value. (iii) $b_{A'}a_A{X}^\dagger_{\hat\ell}|0\rangle =\hat\xi^A\bar{\hat\xi}^{A'}|0\rangle$, and $|\bar uv|\le\tfrac12(|u|^2+|v|^2)$. $\square$

Atoms are not separately observable in this state, and the atomic phase drops out, but the molecular order parameter $\langle b_{A'}a_A\rangle$ is the phase-independent null vector of Proposition 1. With two flavors, condensing the four pairs $a_\alpha b_\beta$ along $\xi_\alpha\bar\xi_\beta$ gives the null tetrad (7), so the patch tetrad of Sec. II is the order-parameter data of physical states. The Pauli bound $|\Delta|\le\tfrac12$ has a physical meaning: one flavor holds at most one molecule, and the order parameter of a patch grows only with the number of molecules it contains ($N_f$ flavors condensed along the same $\hat\xi$ give $N_f\Delta\,\hat\ell$; checked in the notebook).

**Verification of Proposition 10.** Single patch, one flavor, fermionic atoms ($16$ states): $\Psi=(u+vX^\dagger_{\hat\xi})\ket0$ satisfies $\hat H\Psi=0$; $(X^\dagger_{\hat\xi})^2=0$; single-atom amplitudes vanish; and $\langle b_{A'}a_A\rangle=\bar uv\,\hat\xi^A\bar{\hat\xi}^{A'}/(|u|^2+|v|^2)$ is a null vector whose length is at most $\tfrac12$.

```python
xi = np.array([0.55 + 0.2j, -0.3 + 0.4j]); xih = xi / np.linalg.norm(xi)
ell = np.outer(xi, xi.conj()); ellh = np.outer(xih, xih.conj())
Xxi = (sum(xih[A] * aF[A].T for A in range(2)) @ sum(xih[Ap].conj() * bF[Ap].T for Ap in range(2))).tocsr()
uC, vC = 0.8 + 0.1j, 0.5 - 0.3j
PsiC = uC * vac4 + vC * (Xxi @ vac4); nrm = np.vdot(PsiC, PsiC).real
NL = aF[0].T @ aF[0] + aF[1].T @ aF[1]; NR = bF[0].T @ bF[0] + bF[1].T @ bF[1]
ordp = np.array([[np.vdot(PsiC, bF[Ap] @ (aF[A] @ PsiC)) / nrm for Ap in range(2)] for A in range(2)])   # <b_A' a_A>, rows A
Delta = np.conj(uC) * vC / (abs(uC)**2 + abs(vC)**2)
check("Prop. 10: H Psi = 0, (X^dag)^2 = 0, <a_A> = <b_A'> = 0", "N",
      np.linalg.norm((NL - NR) @ PsiC) < 1e-13 and spnorm(Xxi @ Xxi) < 1e-13
      and max(abs(np.vdot(PsiC, o_ @ PsiC)) for o_ in aF + bF) < 1e-13)
check("Prop. 10: <b_A' a_A> = Delta xi^A conj(xi^A') with |Delta| <= 1/2; a null vector", "N",
      np.allclose(ordp, Delta * ellh) and abs(dot(vec(ordp), vec(ordp))) < 1e-13 and abs(Delta) <= 0.5,
      f"Delta = {Delta:.4f}, |Delta| = {abs(Delta):.4f}")
```

On a lattice the same construction works mode by mode.

**Proposition 11 (Molecular condensates solve the constraint).** For any complex $u_k,v_k$, the BCS-type state $$|\Psi_{\rm BCS}\rangle=\prod_k\big(u_k+v_k\,{X}^\dagger_k\big)|0\rangle 
\tag{24}$$ satisfies ${\hat{H}}|\Psi_{\rm BCS}\rangle=0$, with order parameter $\langle d_lc_k\rangle=\delta_{kl}\Delta_k$, $\Delta_k=\bar u_kv_k/(|u_k|^2+|v_k|^2)$. For a uniform condensate ($u_k=u$, $v_k=v$), the site-basis order parameter $\pi_{lm}\equiv\langle b_la_m\rangle$ is $\Delta\,\delta_{lm}$: on every patch it is $\Delta$ times the identity of spinor space.

*Proof.* The factors commute (Proposition 8) and each commutes with ${\hat{H}}$ (Theorem 2); the single-mode computation is that of Proposition 10. Inverting Eq. (17), $a_m=\sum_ku_k(m)c_k$ and $b_l=\sum_k\bar u_k(l)d_k$, so $\pi_{lm}=\Delta\sum_k\bar u_k(l)u_k(m)=\Delta\,\delta_{lm}$ by completeness. $\square$

**The thermofield double.** Choosing $v_k/u_k=e^{-\beta\epsilon_k/2}$ makes Eq. (24) the thermofield double of the left sector. Tracing out the right atoms leaves the Gibbs state $e^{-\beta a^\dagger ha}/Z$, with Fermi--Dirac occupations $(e^{\beta\epsilon_k}+1)^{-1}$; the notebook checks this in the full Fock space. The mirror sector plays the role of the "tilde" copy of thermofield dynamics [29], and ${\hat{H}}=a^\dagger ha-b^\dagger\bar hb$ is the difference of the two Hamiltonians that annihilates the thermofield vacuum, the structure Israel found for the two exteriors of an eternal black hole [30]. The constraint does not fix $\beta$: every temperature, and every non-thermal choice of $u_k,v_k$, gives a physical state.

**Proposition 12 (A condensate compatible with every connection).** Let $Q^\dagger=\sum_{i,A}a^\dagger_{iA}b^\dagger_{iA}$. For every connection $U$, $Q^\dagger=\sum_k{X}^\dagger_k$ in the normal modes of $h(U)$, and $[{\hat{H}}(U),Q^\dagger]=0$. Hence $e^{zQ^\dagger}|0\rangle =\prod_k(1+z{X}^\dagger_k)|0\rangle$ solves ${\hat{H}}(U)|\Psi\rangle =0$ for all $U$ at once.

*Proof.* By completeness, $\sum_kc^\dagger_kd^\dagger_k=\sum_{ij}\big(\sum_ku_k(i)\bar u_k(j)\big)a^\dagger_ib^\dagger_j=\sum_ia^\dagger_ib^\dagger_i$. Directly, $[a^\dagger ha-b^\dagger\bar hb,Q^\dagger]=\sum_{mi}(h_{mi}-\bar h_{im})a^\dagger_mb^\dagger_i=0$ because $h$ is Hermitian. The exponential factorizes because the ${X}^\dagger_k$ commute and square to zero. $\square$

This is the uniform condensate of Proposition 11, and at $z=1$ it is the infinite-temperature thermofield double. It explains why its order parameter is $\Delta\,{\mathbf{1}}$ on every patch for every connection. It also sets a limit: the uniform condensate is compatible with flat and curved connections alike and cannot by itself select a geometry. Only nonuniform condensates, such as the finite-temperature thermofield double, whose amplitudes depend on the spectrum of $h(U)$, distinguish between connections. How a state comes to favor a particular geometry is a dynamical question (Table VIII).

The state (24) is a superposition of different molecule numbers with a definite relative phase. The constraint conserves $\hat N_{/}$ and $\hat N_{\backslash}$ separately, so it is invariant under the global $U(1)$ generated by the molecule number $\hat N_{/}+\hat N_{\backslash}$, which multiplies $a$ and $b$ by the same phase. Unlike the relative phase, this $U(1)$ cannot be absorbed into the links: $a^\dagger_iU_{ij}a_j$ and $b^\dagger_i\bar U_{ij}b_j$ would require conjugate compensations. It is therefore a global symmetry, and the condensates break it: every phase of $v/u$ gives a solution, so the phase of the order parameter is a flat direction of the constraint. This is the defining structure of a superfluid.

The order parameter is constrained by the dynamics. With $\Pi_{lm}=b_la_m$, Eq. (16) gives the exact operator identity $$=\bar h\,\Pi-\Pi\,h^{T} .
\tag{25}$$ The molecular bilinears thus form a closed linear system under ${\hat{H}}$, with the bond couplings of $h$. Under the flow generated by ${\hat{H}}$ they obey the linear equations of a network of coupled oscillators, whether the atoms are bosons or fermions.

**Proposition 13 (Molecular field equation).** In every normalizable physical state, the order parameter $\pi=\langle\Pi\rangle$ commutes with the one-atom operator: $$\bar h\,\pi=\pi\,\bar h,\qquad\text{equivalently}\qquad [\bar t,\pi]=-[\bar v,\pi],
\tag{26}$$ where $\bar h=\bar v+\bar t$ splits into the binding $\bar v=\mathrm{diag}(\epsilon_0+Jz_i)\otimes{\mathbf{1}}_2$ and the transfer $\bar t=-J\bar A_U$, with $(\bar A_U)_{ij}=\bar U_{ij}$ on links. In normal modes, $(\epsilon_l-\epsilon_k)\langle d_lc_k\rangle=0$: the condensate pairs only modes of equal energy.

*Proof.* For ${\hat{H}}|\Psi\rangle =0$, $\langle\Psi|[{\hat{H}},\Pi]|\Psi\rangle=0$; use Eq. (25) and $h^T=\bar h$. In normal modes, $[{\hat{H}},d_lc_k]=(\epsilon_l-\epsilon_k)d_lc_k$. $\square$

Equation (26) is the balance (1) written for the order parameter: information flow along the bonds cancels the binding imbalance. Its consequence for site-local order parameters is sharp.

**Corollary 2 (The site-local order parameter is the timelike normal).** Let $\pi$ be site local, $\pi_{(i,\cdot),(j,\cdot)}=\delta_{ij}\ell_i$, in a physical state. Then $\ell_i=\bar U_{ij}\ell_j\bar U_{ij}^{-1}$ on every link; the order parameter is covariantly constant. Its soldered vector, ${X}_i=\ell_i^T$, commutes with the holonomy of every cycle through $i$. Consequently: (a) for a flat connection, ${X}_i$ is any covariantly constant vector; (b) if the holonomies through $i$ are rotations about a common axis $\hat n$, then ${X}_i\in\mathrm{span}\{{\mathbf{1}},\hat n\cdot\vec\sigma\}$, and its only null directions are the two light rays $(1,\pm\hat n)$ along that axis; (c) if two holonomies through $i$ do not commute, then ${X}_i\propto{\mathbf{1}}$, i.e. ${X}^a_i\propto(1,0,0,0)$: the order parameter is the unit timelike normal of the spatial slice.

*Proof.* For block-diagonal $\pi$, the binding $\bar v$ commutes with $\pi$, and the off-diagonal blocks of Eq. (26) read $\bar U_{ij}\ell_j=\ell_i\bar U_{ij}$. Transporting around a cycle shows that $\ell_i$ commutes with its holonomy. The commutant of a non-scalar element of ${SU(2)}$ is the span of ${\mathbf{1}}$ and that element, and the commutant of two non-commuting elements is the scalars. Under $a\to\Lambda a$, $b\to\bar\Lambda b$, $\pi\to\bar\Lambda\pi\Lambda^T$, so $\pi^T$ transforms as ${X}\to\Lambda{X}\Lambda^\dagger$. $\square$

The uniform condensate of Proposition 11 realizes case (c) exactly: its order parameter is $\Delta\,{\mathbf{1}}$ on every patch. The corollary is the quantum counterpart of Corollary 1. In a curved region, a patch can carry only the timelike direction singled out by the condensate (the rest frame of the molecular superfluid, which is the frame whose rotation group ${SU(2)}$ is selected by the Fock inner product, Sec. III E). The null directions, and with them shape, curvature, and gravitational waves, must be carried by bond molecules $\langle b_la_m\rangle$ with $l\neq m$. The notebook finds site-local solution spaces of dimension $4$, $2$, and $1$ for a flat connection, the four-patch graph of Table I (one curved cycle), and the same graph with a second, non-commuting cycle.

**Verification of Proposition 11.** The uniform condensate $\prod_k(u+vX_k^\dagger)\ket0$ over all eight normal modes of the four-patch graph is built in the $65\,536$-dimensional site-basis Fock space of Theorem 2. It solves $\hat H\Psi=0$, has molecular order parameter $\langle d_lc_k\rangle=\delta_{kl}\Delta$, and in the site basis $\pi_{lm}=\langle b_la_m\rangle=\Delta\,\delta_{lm}$: on every patch the order parameter is the identity of spinor space. A nonuniform condensate also solves the constraint, and the thermofield double ($v_k/u_k=e^{-\beta\epsilon_k/2}$) reduces to the Gibbs state of the left sector, and (Proposition 12) the uniform condensate $e^{zQ^\dagger}\ket0$ solves the constraint for every connection. [L], [R] theorem `bcs_solution`.

```python
def condensate(us, vs):
    v_ = vacF.copy()
    for k_ in range(NM): v_ = us[k_] * v_ + vs[k_] * (XdagN[k_] @ v_)
    return v_ / np.linalg.norm(v_)
def orderparam(v_):          # pi_{lm} = <b_l a_m>, site basis (8 x 8)
    return np.array([[np.vdot(v_, bS[l_] @ (aS[m_] @ v_)) for m_ in range(NM)] for l_ in range(NM)])
uU, vU = 0.8, 0.6 * np.exp(0.9j)
PsiU = condensate([uU] * NM, [vU] * NM)
DeltaU = np.conj(uU) * vU / (abs(uU)**2 + abs(vU)**2)
piU = orderparam(PsiU)
cN = [ddagN[k].T for k in range(NM)]
dlc = np.array([[np.vdot(PsiU, (ddagN[l_].T @ (cdagN[k_].T @ PsiU))) for k_ in range(NM)] for l_ in range(NM)])
check("Prop. 11: uniform condensate solves H Psi = 0; <d_l c_k> = Delta delta_kl", "N",
      np.linalg.norm(Hf @ PsiU) < 1e-11 and np.allclose(dlc, DeltaU * np.eye(NM)), f"Delta = {DeltaU:.4f}")
check("Prop. 11: site-basis order parameter pi = Delta x identity on every patch", "N",
      np.allclose(piU, DeltaU * np.eye(NM)))
rngB = np.random.default_rng(77)
usR = rngB.normal(size=NM) + 1j * rngB.normal(size=NM); vsR = rngB.normal(size=NM) + 1j * rngB.normal(size=NM)
PsiR = condensate(usR, vsR); piR = orderparam(PsiR)
check("Prop. 11: nonuniform condensate solves H Psi = 0 (order parameter not site-local)", "N",
      np.linalg.norm(Hf @ PsiR) < 1e-11 and np.abs(piR - np.diag(np.diag(piR))).max() > 1e-3)
# thermofield double: v_k/u_k = exp(-beta eps_k/2); tracing out the right atoms leaves exp(-beta a^dag h a)/Z
betaT = 1.3
PsiT = condensate(np.ones(NM), np.exp(-betaT * w / 2))
MT = PsiT.reshape(2**NM, 2**NM)                       # a-modes are the leading Jordan-Wigner factors
rhoL = MT @ MT.conj().T
F8L = jw_ops(NM); HAL = sum(h[i, j] * (F8L[i].T @ F8L[j]) for i in range(NM) for j in range(NM)).toarray()
gibbs = expm(-betaT * HAL); gibbs /= np.trace(gibbs).real
occT = np.array([np.trace(rhoL @ (F8L[0].T @ F8L[0]).toarray()).real])
# Proposition (blind): Q^dag = sum_i a_i^dag b_i^dag commutes with H(U) for every connection U
Qd = sum(aS[i].T @ bS[i].T for i in range(NM)).tocsr()
blind = [spnorm(comm(Hf, Qd))]
for Utest in (Uflat, {l_: su2(np.random.default_rng(91 + n_)) for n_, l_ in enumerate(LINKS)}):
    ht = eps0 * np.eye(2 * NS) + Jc * conn_laplacian(Utest)
    Ht = (onebody(ht, aS) - onebody(ht.conj(), bS)).tocsr()
    blind.append(spnorm(comm(Ht, Qd)))
zq = 0.7
vq = vacF.copy(); term_ = vacF.copy()
for n_ in range(1, NM + 1):
    term_ = zq * (Qd @ term_) / n_; vq = vq + term_
check("Prop. 12: [H(U), Q^dag] = 0 for curved, flat and a second curved U; exp(z Q^dag)|0> = prod (1 + z X_k^dag)|0>", "N",
      max(blind) < 1e-12 and np.allclose(vq / np.linalg.norm(vq), condensate(np.ones(NM), zq * np.ones(NM))),
      f"max |[H(U), Q^dag]| = {max(blind):.1e}")
check("Prop. 11: thermofield double solves H Psi = 0; Tr_R |Psi><Psi| = exp(-beta a^dag h a)/Z", "N",
      np.linalg.norm(Hf @ PsiT) < 1e-11 and np.abs(rhoL - gibbs).max() < 1e-12,
      f"beta = {betaT}, max |rho_L - Gibbs| = {np.abs(rhoL - gibbs).max():.1e}")
```

**Verification of Proposition 13 and Corollary 2.** [N] The operator identity $[\hat H,b_la_m]=(\bar h\,ba-ba\,h^{T})_{lm}$ on the $65\,536$-dimensional Fock space; $[\bar h,\pi]=0$ for the uniform condensate, the nonuniform condensate, and a molecular state; the balance form $[\bar t,\pi]=-[\bar v,\pi]$. [N] Site-local solutions of $[\bar h,\pi]=0$: their dimension is $4$ for a flat connection, $2$ (spanned by $\mathbf 1$ and the holonomy axis, containing two null directions) for the four-patch graph with one curved cycle, and $1$ (the identity: the timelike normal) once a second, non-commuting cycle is added. [L], [R] theorem `pair_shift`.

```python
ba = lambda l_, m_: (bS[l_] @ aS[m_]).tocsr()
hb = h.conj()
fe_op = max(spnorm(comm(Hf, ba(l_, m_)) - (sum(hb[l_, n_] * ba(n_, m_) for n_ in range(NM)) - sum(h[m_, n_] * ba(l_, n_) for n_ in range(NM))))
            for l_, m_ in [(0, 0), (1, 4), (6, 3)])
check("Prop. 13: [H, b_l a_m] = (hbar b a - b a h^T)_lm (operator identity)", "N", fe_op < 1e-12, f"max deviation {fe_op:.1e}")
v1 = XdagN[2] @ (XdagN[5] @ vacF); v1 = v1 + 0.3 * (XdagN[0] @ vacF); v1 = v1 / np.linalg.norm(v1)
fe_states = [np.abs(hb @ p_ - p_ @ hb).max() for p_ in (piU, piR, orderparam(v1))]
vbar = np.kron(np.diag(eps0 + Jc * z), np.eye(2)).astype(complex); tbar = hb - vbar
check("Prop. 13: [hbar, pi] = 0 in physical states; [tbar, pi] = -[vbar, pi]", "N",
      max(fe_states) < 1e-11 and np.abs((tbar @ piR - piR @ tbar) + (vbar @ piR - piR @ vbar)).max() < 1e-11
      and np.abs(vbar @ piR - piR @ vbar).max() > 1e-3, f"max |[hbar, pi]| = {max(fe_states):.1e}")
def sitelocal_dim(Ulinks, links, N_):
    hh = eps0 * np.eye(2 * N_) + Jc * conn_laplacian(Ulinks, links, N_); hbb = hh.conj()
    cols = []
    for i_ in range(N_):
        for e_ in range(4):
            E_ = np.zeros((2 * N_, 2 * N_), complex); E_[2*i_ + e_ // 2, 2*i_ + e_ % 2] = 1
            cols.append((hbb @ E_ - E_ @ hbb).ravel())
    ns_ = null_space(np.array(cols).T)
    return ns_.shape[1], ns_
d_flat, _ = sitelocal_dim(Uflat, LINKS, NS)
d_one, ns_one = sitelocal_dim(Ucurv, LINKS, NS)
LINKS2 = LINKS + [(1, 3)]; Ucurv2 = dict(Ucurv); Ucurv2[(1, 3)] = su2(np.random.default_rng(5))
d_two, ns_two = sitelocal_dim(Ucurv2, LINKS2, NS)
idmat = ns_two[:, 0].reshape(NS, 4)
is_identity = all(np.allclose(idmat[i_].reshape(2, 2), idmat[0, 0] * np.eye(2)) for i_ in range(NS))
check("Cor. 2: site-local solutions: flat 4, one curved cycle 2, two non-commuting cycles 1 (= identity)", "N",
      (d_flat, d_one, d_two) == (4, 2, 1) and is_identity, f"dimensions {d_flat}, {d_one}, {d_two}")
# one cycle: the patch-0 blocks span {1, W}, W the holonomy acting on pi; its eigenprojectors are real null vectors
blocks = np.array([ns_one[:, k].reshape(NS, 4)[0] for k in range(2)]).T          # 4 x 2
Ub = {k_: v_.conj() for k_, v_ in Ucurv.items()}
found = None
for Wm in (Ub[(0, 1)] @ Ub[(1, 2)] @ Ub[(2, 0)], Ub[(2, 0)] @ Ub[(1, 2)] @ Ub[(0, 1)],
           Ub[(0, 1)].conj().T @ Ub[(2, 0)].conj().T @ Ub[(1, 2)].conj().T, Ub[(2, 0)].T @ Ub[(1, 2)].T @ Ub[(0, 1)].T):
    evW, vecW = np.linalg.eig(Wm)
    projs = [np.outer(vecW[:, k_], vecW[:, k_].conj()) / np.vdot(vecW[:, k_], vecW[:, k_]).real for k_ in range(2)]
    if all(np.linalg.matrix_rank(np.column_stack([blocks, P_.ravel()]), tol=1e-9) == 2 for P_ in projs):
        found = projs; break
nulls_ok = found is not None and all(abs(dot(vec(P_.T), vec(P_.T))) < 1e-10 and np.allclose(P_, P_.conj().T) for P_ in found)
check("Cor. 2: with one cycle the solutions are spanned by 1 and the holonomy; its two eigenprojectors are real null vectors", "N",
      nulls_ok)
```

### E. Physical picture: spacetime as a molecular superfluid

The results of this section assemble into one physical picture. The constituents are fermions (Proposition 8). The constraint confines them into bosonic molecules; each elementary molecule is a light ray (Theorem 2, Proposition 1). Dilute molecules obey the oscillator algebra, and the molecular field is coupled through the same bonds that bind the atoms (Propositions 9 and 13), and their phase-coherent condensates are physical states (Proposition 11). This is the structure of a fermionic superfluid. The same sequence, fermionic atoms binding into bosonic molecules that condense, is observed in ultracold $^{40}$K and $^6$Li gases. There molecules formed from fermionic atoms Bose-condense [14], and across the BCS--BEC crossover [31, 32] the pair condensate is a superfluid with quantized vortices [15]. The two regimes of that crossover correspond to the two regimes of Proposition 9. Dilute pairs ($n\ll\Omega$) are oscillators, as in a molecular Bose--Einstein condensate, while dense pairs ($n\sim\Omega$) are dominated by the Pauli principle, as in a BCS state.

In this superfluid the geometry is the order parameter, as in Volovik's picture of emergent gravity in superfluid helium [10]. Three properties of the order parameter carry three geometric meanings.

-   *Shape: the light cone.* The order parameter of a single patch is a null vector (Proposition 10), and the four molecules of a two-flavor patch form a null tetrad (Proposition 3).

-   *Length: the conformal factor.* The length of the order parameter counts molecules (the Pauli bound), so the conformal factor $\Omega^2$ of Corollary 1 measures the condensate density.

-   *Phase: the superfluid phase.* The phase of the order parameter is the superfluid phase, the flat direction of the constraint identified above.

In a curved region the condensate defines a rest frame (Corollary 2), as every superfluid does, and the light rays live on the bonds between patches.

In a laboratory superfluid, sound is an oscillation of the order parameter, flow is a gradient of its phase, and the phonons propagate on an effective Lorentzian metric set by the condensate [8, 9, 10]. Spacetime is observed to behave like such a medium: it carries waves [16] and is dragged by rotating masses [17]. In the present model the corresponding statements are targets, not results. Gravitational waves should be small oscillations of the bond order parameter, in the harmonic regime of Proposition 9, and gravitational fields should be its spatial variations. Deriving these collective modes and their spectrum requires dynamical links and is the main open problem (Sec. V H). Two further qualifications keep the picture honest. First, the pairing is kinematic: the constraint selects paired states but contains no attractive interaction and no binding energy, as befits a constraint, which has no energy to minimize. Whether dynamical links produce spatially localized pairs is open. Second, phase coherence is not yet superfluidity. Phase stiffness, collective modes, and the response to phase gradients remain to be established (Table VIII). What the minimal model establishes are the ingredients:

-   fermionic constituents;

-   bosonic pairs;

-   harmonic molecular dynamics in the dilute limit;

-   phase-coherent condensates that solve the constraint;

-   an order parameter that carries the light cone.

## V. Geometry from the links: Cartan structure equations

### A. Discrete data and the continuum dictionary

The pre-geometry supplies three kinds of data:

1.  an internal Minkowski space $\mathcal{M}_i$ with frame $E_{a,i}$ at each patch, built from atoms (Sec. II, Sec. IV D);

2.  a transporter on each link, the $U_{ij}$ of ${\hat{T}}$, acting on frame components as $\Lambda_{ij}=\Lambda(U_{ij})$;

3.  an edge vector $e^a(ij)$: the frame components, in the frame of patch $i$, of the internal vector assigned to the link $\langle ij\rangle$. In the minimal realization this is a combination of *bond molecules* ${/}_i(\bar U_{ij}{\backslash}_j)$ straddling the link; a single bond molecule is null, so at least two are needed for a timelike or spacelike edge. Corollary 2 shows that in curved regions this is forced: a site-local molecular order parameter is the timelike normal, so the null directions must be carried by the bonds.

For a patch at coordinate position $x$ and its neighbor at $x+h\hat\mu$, the continuum dictionary is $$\begin{aligned}
\Lambda_{x,x+h\hat\mu}&=\exp\!\big[h\,\omega_\mu(x+\tfrac h2\hat\mu)\big],\\
e^a(x,\mu)&=h\,e^a{}_\mu\big(x+\tfrac h2\hat\mu\big),
\end{aligned}
\tag{27}$$ where $\omega_\mu$ is the matrix $\omega_\mu{}^a{}_b$. The sign follows from parallel transport, $\nabla_\mu V^a=\partial_\mu V^a+\omega_\mu{}^a{}_bV^b=0$: a vector at $x+\mathrm{d} x$ is carried back to $x$ by ${\mathbf{1}}+\omega_\mu\mathrm{d} x^\mu$. In the spinor representation, $U_{x,x+h\hat\mu}=\exp[hS(\omega_\mu)]$ with $S$ from Eq. (10). The left atoms are transported by the self-dual part of the spin connection and the right atoms by its anti-self-dual part.

### B. First structure equation: torsion is a closure defect

Carroll's first structure equation [18] reads $$T^a=\mathrm{d} e^a+\omega^a{}_b\wedge e^b ,
\tag{28}$$ or in components $T^a{}_{\mu\nu}=\partial_\mu e^a{}_\nu-\partial_\nu e^a{}_\mu+\omega_\mu{}^a{}_be^b{}_\nu
-\omega_\nu{}^a{}_be^b{}_\mu$. Its discrete form compares the two ways of going around an elementary plaquette, transporting each second leg back to the base patch: $$\begin{aligned}
\mathcal{T}^a(x;\mu\nu)={}&\big[e(x,\mu)+\Lambda_{x,x+h\hat\mu}\,e(x+h\hat\mu,\nu)\big]^a\\
&-\big[e(x,\nu)+\Lambda_{x,x+h\hat\nu}\,e(x+h\hat\nu,\mu)\big]^a .
\end{aligned}
\tag{29}$$ Expanding with Eq. (27) gives $\mathcal{T}^a(x;\mu\nu)=h^2\,T^a{}_{\mu\nu}(x)+O(h^3)$. Torsion is the failure of the plaquette to close after transport. Requiring closure ($T^a=0$) determines the transporters from the edge data. The unique solution is Carroll's $$\omega_\mu{}^a{}_b=e^a{}_\nu\,e_b{}^\lambda\,\Gamma^\nu{}_{\mu\lambda}-e_b{}^\lambda\,\partial_\mu e^a{}_\lambda ,
\tag{30}$$ with $\Gamma$ the Christoffel symbols. In the language of the model, *a torsion-free ${\hat{T}}$ has no independent information*: the transport rule is fixed by how the atom frames and edge molecules vary from patch to patch. Admitting $T^a\neq0$ (Einstein--Cartan [33, 34, 35]) is a physical choice that the model leaves open. The spinorial atoms are natural torsion sources.

### C. Second structure equation: curvature is plaquette holonomy

Carroll's second structure equation is $$R^a{}_b=\mathrm{d}\omega^a{}_b+\omega^a{}_c\wedge\omega^c{}_b ,
\tag{31}$$ i.e. $R^a{}_{b\mu\nu}=\partial_\mu\omega_\nu{}^a{}_b-\partial_\nu\omega_\mu{}^a{}_b
+\omega_\mu{}^a{}_c\omega_\nu{}^c{}_b-\omega_\nu{}^a{}_c\omega_\mu{}^c{}_b$, with $R^\rho{}_{\sigma\mu\nu}=e_a{}^\rho e^b{}_\sigma R^a{}_{b\mu\nu}$. Its discrete form is the holonomy of the transporters of ${\hat{T}}$ around a plaquette, $$\begin{aligned}
\Lambda_p(x;\mu\nu)={}&\Lambda_{x,x+h\hat\mu}\,\Lambda_{x+h\hat\mu,\,x+h\hat\mu+h\hat\nu}\\
&\times\Lambda_{x+h\hat\nu,\,x+h\hat\mu+h\hat\nu}^{-1}\,\Lambda_{x,x+h\hat\nu}^{-1}.
\end{aligned}
\tag{32}$$ Writing each factor as an exponential and applying the Baker--Campbell--Hausdorff formula to second order, the linear terms combine into $h^2(\partial_\mu\omega_\nu-\partial_\nu\omega_\mu)$, and the six pairwise commutators of $(\omega_\mu,\omega_\nu,-\omega_\mu,-\omega_\nu)$ sum to $2[\omega_\mu,\omega_\nu]$ with coefficient $\tfrac12$. Hence $$\Lambda_p(x;\mu\nu)={\mathbf{1}}+h^2R_{\mu\nu}(x)+O(h^3),
\tag{33}$$ with $(R_{\mu\nu})^a{}_b=R^a{}_{b\mu\nu}$ as in Eq. (31). *Spacetime curvature is the net Lorentz rotation of an atom's frame when information flows around a closed loop of patches.* The Bianchi identities $\mathrm{d} T^a+\omega^a{}_b\wedge T^b=R^a{}_b\wedge e^b$ and $\mathrm{d} R^a{}_b+\omega^a{}_c\wedge R^c{}_b
-R^a{}_c\wedge\omega^c{}_b=0$ follow from Eqs. (28) and (31). On the lattice they express consistency of transport around the faces of an elementary cube.

### D. Worked examples and numerical verification

Table II lists the connection and frame curvature components obtained from Eqs. (30) and (31) for two standard geometries. The notebook confirms symbolically that $T^a=0$ identically and that $e_a{}^\rho e^b{}_\sigma R^a{}_{b\mu\nu}$ coincides with the Riemann tensor built from Christoffel symbols, for both metrics.

**Table II.** Spin connection and frame components $R^a{}_{bcd}$ from Cartan's equations (Carroll conventions). FLRW (flat): $e^0=\mathrm{d} t$, $e^i=a(t)\,\mathrm{d} x^i$. Schwarzschild: $e^0=f^{1/2}\mathrm{d} t$, $e^1=f^{-1/2}\mathrm{d} r$, $e^2=r\,\mathrm{d}\theta$, $e^3=r\sin\theta\,\mathrm{d}\phi$, $f=1-2M/r$. Reproduced in the companion notebook.

| FLRW                                   | Schwarzschild                                     |
|:---------------------------------------|:--------------------------------------------------|
| $\omega^0{}_i=\dot a\,\mathrm{d} x^i$  | $\omega^0{}_1=(M/r^2)\,\mathrm{d} t$              |
|                                        | $\omega^1{}_2=-f^{1/2}\mathrm{d}\theta$           |
|                                        | $\omega^1{}_3=-f^{1/2}\sin\theta\,\mathrm{d}\phi$ |
|                                        | $\omega^2{}_3=-\cos\theta\,\mathrm{d}\phi$        |

|                                        | $R^2{}_{323}=2M/r^3$                              |

**Verification of Eqs. (29)–(33) in the continuum and generation of Table II.** For FLRW and Schwarzschild: torsion vanishes, the Cartan curvature agrees with the Riemann tensor built from Christoffel symbols, and the frame components agree with Table II. The helpers `geometry` and `invariants` (frame curvature, $\mathrm{Riem}^2$, $\mathrm{Ric}^2$, $R$, $C^2$, $E$) are used again in Secs. V E–V H.

```python
t, r, thv, phv, x, y, zc, Mm = sp.symbols('t r theta phi x y z M', positive=True)
afun = sp.Function('a', positive=True)(t)
def christoffel(co, g, gi):
    return [[[sp.simplify(sum(gi[nu, s] * (sp.diff(g[s, m_], co[l_]) + sp.diff(g[s, l_], co[m_]) - sp.diff(g[m_, l_], co[s])) for s in range(4)) / 2)
              for l_ in range(4)] for m_ in range(4)] for nu in range(4)]
def geometry(co, e):
    S_ = sp.simplify
    e = sp.Matrix(e); g = S_(e.T * etaS * e); gi = S_(g.inv()); Ei = S_(e.inv())
    Gam = christoffel(co, g, gi)
    om = [sp.Matrix(4, 4, lambda A, B: S_(sum(e[A, nu] * Ei[lam, B] * Gam[nu][mu][lam] for nu in range(4) for lam in range(4))
                                         - sum(Ei[lam, B] * sp.diff(e[A, lam], co[mu]) for lam in range(4)))) for mu in range(4)]
    Rmn = {(mu, nu): (S_(sp.diff(om[nu], co[mu]) - sp.diff(om[mu], co[nu]) + om[mu] * om[nu] - om[nu] * om[mu]) if mu != nu else sp.zeros(4))
           for mu in range(4) for nu in range(4)}
    Rf = sp.MutableDenseNDimArray.zeros(4, 4, 4, 4)
    for a_, b_, c_, d_ in itertools.product(range(4), repeat=4):
        Rf[a_, b_, c_, d_] = S_(sum(Rmn[(mu, nu)][a_, b_] * Ei[mu, c_] * Ei[nu, d_] for mu in range(4) for nu in range(4)))
    return dict(co=co, e=e, g=g, gi=gi, Ei=Ei, Gam=Gam, om=om, Rmn=Rmn, Rf=Rf, sqrtg=S_(sp.sqrt(-g.det())))
def invariants(Gd):
    Rf = Gd['Rf']; up = lambda i: etaS[i, i]
    Rl = sp.MutableDenseNDimArray.zeros(4, 4, 4, 4)
    for a_, b_, c_, d_ in itertools.product(range(4), repeat=4): Rl[a_, b_, c_, d_] = etaS[a_, a_] * Rf[a_, b_, c_, d_]
    Riem2 = sp.simplify(sum(Rl[a_, b_, c_, d_]**2 * up(a_) * up(b_) * up(c_) * up(d_) for a_, b_, c_, d_ in itertools.product(range(4), repeat=4)))
    Ric = sp.Matrix(4, 4, lambda b_, d_: sum(Rf[a_, b_, a_, d_] for a_ in range(4)))
    Rs = sp.simplify(sum(Ric[b_, b_] * up(b_) for b_ in range(4)))
    Ric2 = sp.simplify(sum(Ric[a_, b_]**2 * up(a_) * up(b_) for a_ in range(4) for b_ in range(4)))
    Cw = lambda a_, b_, c_, d_: (Rl[a_, b_, c_, d_] - sp.Rational(1, 2) * (etaS[a_, c_] * Ric[b_, d_] - etaS[a_, d_] * Ric[b_, c_] - etaS[b_, c_] * Ric[a_, d_] + etaS[b_, d_] * Ric[a_, c_])
                                 + Rs / 6 * (etaS[a_, c_] * etaS[b_, d_] - etaS[a_, d_] * etaS[b_, c_]))
    C2 = sp.simplify(sum(Cw(a_, b_, c_, d_)**2 * up(a_) * up(b_) * up(c_) * up(d_) for a_, b_, c_, d_ in itertools.product(range(4), repeat=4)))
    return dict(Riem2=Riem2, Ric2=Ric2, R=Rs, C2=C2, E=sp.simplify(Riem2 - 4 * Ric2 + Rs**2))
def cartan_report(Gd):
    co, e, om, Gam, Ei = Gd['co'], Gd['e'], Gd['om'], Gd['Gam'], Gd['Ei']
    Tmax = max(abs(sp.simplify(sp.diff(e[A, nu], co[mu]) - sp.diff(e[A, mu], co[nu]) + sum(om[mu][A, B] * e[B, nu] - om[nu][A, B] * e[B, mu] for B in range(4))))
               for A in range(4) for mu in range(4) for nu in range(4))
    Riem = lambda rho, sig, mu, nu: sp.diff(Gam[rho][nu][sig], co[mu]) - sp.diff(Gam[rho][mu][sig], co[nu]) + sum(Gam[rho][mu][l_] * Gam[l_][nu][sig] - Gam[rho][nu][l_] * Gam[l_][mu][sig] for l_ in range(4))
    dev = max(abs(sp.simplify(sum(Ei[rho, A] * e[B, sig] * Gd['Rmn'][(mu, nu)][A, B] for A in range(4) for B in range(4)) - Riem(rho, sig, mu, nu)))
              for mu, nu in itertools.combinations(range(4), 2) for rho, sig in itertools.product(range(4), repeat=2))
    return Tmax, dev
f_ = 1 - 2 * Mm / r
TABLE2 = {"FLRW": ([t, x, y, zc], sp.diag(1, afun, afun, afun),
                   {**{(0, i, 0, i): sp.diff(afun, t, 2) / afun for i in (1, 2, 3)},
                    **{(i, j, i, j): sp.diff(afun, t)**2 / afun**2 for i, j in ((1, 2), (1, 3), (2, 3))}}),
          "Schwarzschild": ([t, r, thv, phv], sp.diag(sp.sqrt(f_), 1 / sp.sqrt(f_), r, r * sp.sin(thv)),
                   {(0, 1, 0, 1): 2 * Mm / r**3, (0, 2, 0, 2): -Mm / r**3, (0, 3, 0, 3): -Mm / r**3,
                    (1, 2, 1, 2): -Mm / r**3, (1, 3, 1, 3): -Mm / r**3, (2, 3, 2, 3): 2 * Mm / r**3})}
GEO = {}
for name, (co, e, expect) in TABLE2.items():
    Gd = geometry(co, e); GEO[name] = Gd
    Tmax, dev = cartan_report(Gd)
    ok_tab = all(sp.simplify(Gd['Rf'][k] - v_) == 0 for k, v_ in expect.items())
    nonzero = {(a_, b_, c_, d_) for a_, b_, c_, d_ in itertools.product(range(4), repeat=4)
               if a_ < b_ and c_ < d_ and sp.simplify(Gd['Rf'][a_, b_, c_, d_]) != 0}
    check(f"Table II, {name}: T^a = 0; R(Cartan) = R(Christoffel); frame components as tabulated", "S",
          Tmax == 0 and dev == 0 and ok_tab and nonzero <= set(expect) | {(k[2], k[3], k[0], k[1]) for k in expect})
    for (A, B) in itertools.combinations(range(4), 2):
        comps = [(co[mu], sp.simplify(Gd['om'][mu][A, B])) for mu in range(4) if sp.simplify(Gd['om'][mu][A, B]) != 0]
        if comps: print(f"    omega^{A}_{B} = " + " + ".join(f"({c_}) d{s_}" for s_, c_ in comps))
```

To test Eqs. (29) and (33) on an actual lattice, we take FLRW with $a(t)=1+0.3t^2+0.1t^3$, build link transporters and edge vectors from Eq. (27), and compare plaquette holonomies and closure defects with the continuum curvature and torsion as the spacing $h$ decreases (Fig. 1). The holonomy converges to the curvature, and the closure defect is $O(h^3)$ for the torsion-free connection. Adding a contorsion $K_x{}^0{}_1=K_x{}^1{}_0=\tfrac12$, the normalized closure defect converges to the predicted torsion $T^a{}_{tx}=K_t{}^a{}_be^b{}_x-K_x{}^a{}_be^b{}_t$.

(b) Closure defect (29) divided by $h^2$: it vanishes as $O(h)$ for the torsion-free connection, and with contorsion it converges to the predicted torsion. Reproduced in the companion notebook.

(27)–(33) on a lattice and generation of Fig. 1.** FLRW with $a(t)=1+0.3t^2+0.1t^3$ at $x_0=(0.4,0.1,-0.2,0.3)$; link transporters $\Lambda=\exp[h\,\omega_\mu(\text{midpoint})]$ and edge molecules $e^a(x,\mu)=h\,e^a{}_\mu(\text{midpoint})$. With an added contorsion $K_x{}^0{}_1=K_x{}^1{}_0=\tfrac12$ the closure defect converges to the predicted torsion.

```python
coN = [t, x, y, zc]; aN = 1 + sp.Rational(3, 10) * t**2 + sp.Rational(1, 10) * t**3; eN = sp.diag(1, aN, aN, aN)
GN = geometry(coN, eN)
omf = [sp.lambdify(coN, GN['om'][mu], 'numpy') for mu in range(4)]
Rfun = {(0, 1): sp.lambdify(coN, GN['Rmn'][(0, 1)], 'numpy'), (1, 2): sp.lambdify(coN, GN['Rmn'][(1, 2)], 'numpy')}
ef = sp.lambdify(coN, eN, 'numpy')
x0 = np.array([0.4, 0.1, -0.2, 0.3]); I4 = np.eye(4)
def link(xa, mu, hh, Kc_=None):
    w_ = np.array(omf[mu](*(xa + 0.5 * hh * I4[mu])), float)
    if Kc_ is not None: w_ = w_ + Kc_[mu]
    return expm(hh * w_)
def plaq(xa, mu, nu, hh):
    return link(xa, mu, hh) @ link(xa + hh * I4[mu], nu, hh) @ np.linalg.inv(link(xa + hh * I4[nu], mu, hh)) @ np.linalg.inv(link(xa, nu, hh))
def edge(xa, mu, hh): return hh * np.array(ef(*(xa + 0.5 * hh * I4[mu])), float)[:, mu]
def closure(xa, mu, nu, hh, Kc_=None):
    return (edge(xa, mu, hh) + link(xa, mu, hh, Kc_) @ edge(xa + hh * I4[mu], nu, hh)) - (edge(xa, nu, hh) + link(xa, nu, hh, Kc_) @ edge(xa + hh * I4[nu], mu, hh))
Kc = np.zeros((4, 4, 4)); Kc[1, 0, 1] = 0.5; Kc[1, 1, 0] = 0.5
ev0 = np.array(ef(*x0), float); Tpred = Kc[0] @ ev0[:, 1] - Kc[1] @ ev0[:, 0]
hs = np.array([0.2, 0.1, 0.05, 0.025, 0.0125]); rows = []
for hh in hs:
    c01 = x0 + 0.5 * hh * (I4[0] + I4[1]); c12 = x0 + 0.5 * hh * (I4[1] + I4[2])
    rows.append((hh, np.abs((plaq(x0, 0, 1, hh) - I4) / hh**2 - np.array(Rfun[(0, 1)](*c01), float)).max(),
                 np.abs((plaq(x0, 1, 2, hh) - I4) / hh**2 - np.array(Rfun[(1, 2)](*c12), float)).max(),
                 np.abs(closure(x0, 0, 1, hh)).max() / hh**2, np.abs(closure(x0, 0, 1, hh, Kc) / hh**2 - Tpred).max()))
rows = np.array(rows)
print("   h     |(P-1)/h^2-R_tx|  |(P-1)/h^2-R_xy|  |closure|/h^2 (T=0)  |closure/h^2-T| (T!=0)")
for rw in rows: print("%7.4f   %13.3e   %13.3e   %15.3e   %17.3e" % tuple(rw))
slopes = [np.polyfit(np.log(rows[-3:, 0]), np.log(rows[-3:, k]), 1)[0] for k in (1, 2, 3, 4)]
check("Eq. (33) and Fig. 1(a): holonomy -> curvature, error O(h^2) for (t,x), O(h) for (x,y)", "N",
      1.8 < slopes[0] < 2.3 and 0.8 < slopes[1] < 1.3, f"slopes {slopes[0]:.2f}, {slopes[1]:.2f}")
check("Eq. (29): closure defect -> torsion (0 and contorsion cases), error O(h)", "N", all(0.8 < s_ < 1.3 for s_ in slopes[2:]), f"slopes {slopes[2]:.2f}, {slopes[3]:.2f}")
from matplotlib.ticker import NullFormatter
fig, ax = plt.subplots(1, 2, figsize=(9, 3.4))
ax[0].loglog(rows[:, 0], rows[:, 1], 'o-', label=r'$(t,x)$ plaquette'); ax[0].loglog(rows[:, 0], rows[:, 2], 's-', label=r'$(x,y)$ plaquette')
ax[0].loglog(rows[:, 0], 0.5 * rows[:, 0], 'k--', lw=0.8, label=r'$\propto h$')
ax[0].set_xlabel(r'lattice spacing $h$'); ax[0].set_ylabel(r'$\max|(\Lambda_p-1)/h^2-R_{\mu\nu}|$'); ax[0].legend(fontsize=8); ax[0].set_title(r'(a) holonomy $\to$ curvature', fontsize=10)
ax[1].loglog(rows[:, 0], rows[:, 3], 'o-', label=r'torsion-free $\omega$'); ax[1].loglog(rows[:, 0], rows[:, 4], 's-', label=r'$\omega+K$: deviation from $T^a$')
ax[1].loglog(rows[:, 0], 0.4 * rows[:, 0], 'k--', lw=0.8, label=r'$\propto h$')
ax[1].set_xlabel(r'lattice spacing $h$'); ax[1].set_ylabel('closure defect / $h^2$'); ax[1].legend(fontsize=8); ax[1].set_title(r'(b) closure $\to$ torsion', fontsize=10)
for a_ in ax: a_.set_xticks(hs); a_.set_xticklabels([f'{v:g}' for v in hs]); a_.xaxis.set_minor_formatter(NullFormatter())
plt.tight_layout(); plt.savefig('fig_cartan_lattice.pdf'); plt.savefig('fig_cartan_lattice.png', dpi=150); plt.show()
```

### E. The Einstein–Cartan functional

The first gravitational functional we need is linear in curvature. We fix the normalization of the first-order action in Carroll's conventions. The notebook verifies, for FLRW with arbitrary $a(t)$, that $\tfrac12\tilde\epsilon^{\mu\nu\rho\sigma}\epsilon_{abcd}\,e^a{}_\mu e^b{}_\nu R^{cd}{}_{\rho\sigma}=2\sqrt{-g}\,R$ (with $\tilde\epsilon^{0123}=+1$ and $\epsilon_{0123}=+1$), so that $$S_{\rm EC}=\frac{1}{32\pi G}\int\epsilon_{abcd}\,e^a\wedge e^b\wedge R^{cd}
=\frac{1}{16\pi G}\int\mathrm{d}^4x\,\sqrt{-g}\,R .
\tag{34}$$ Using Eqs. (27) and (33), the lattice transcription is $$\begin{aligned}
S_\Lambda=\frac{1}{64\pi G}\sum_x{}&\tilde\epsilon^{\mu\nu\rho\sigma}\,\epsilon_{abcd}\,
e^a(x,\mu)\,e^b(x,\nu)\\
&\times\big[\Lambda_p(x;\rho\sigma)-{\mathbf{1}}\big]^{cd},
\end{aligned}
\tag{35}$$ which converges to $S_{\rm EC}$ (compare the lattice tetrad formulations of Ref. [36] and the simplicial formulation of Ref. [37]). The notebook confirms that the lattice density converges linearly in $h$ to $2\sqrt{-g}\,R$. In this functional the holonomy is traced against the *dual* of the plaquette's area bivector, which is built from edge molecules; this is why it is linear in $\Lambda_p-{\mathbf{1}}$. It is the first of the functionals compared in Secs. V G and V H.

**Verification of Eqs. (34)–(35).** [S] $\tfrac12\tilde\epsilon^{\mu\nu\rho\sigma}\epsilon_{abcd}e^a{}_\mu e^b{}_\nu R^{cd}{}_{\rho\sigma}=2\sqrt{-g}R$ (FLRW, general $a(t)$). [N] The lattice density built from plaquette holonomies and edge molecules converges linearly in $h$. The helper `lattice_functionals` also returns the functionals of Eqs. (38) and (42), checked below.

```python
from itertools import permutations
G_F = geometry([t, x, y, zc], sp.diag(1, afun, afun, afun))
eF, RmnF = G_F['e'], G_F['Rmn']
dens = 0
for A, B, Cc, D in permutations(range(4)):
    for mu, nu, rho, sg in permutations(range(4)):
        dens += sp.Rational(1, 2) * sp.LeviCivita(mu, nu, rho, sg) * sp.LeviCivita(A, B, Cc, D) * eF[A, mu] * eF[B, nu] * sum(RmnF[(rho, sg)][Cc, E_] * etaS[E_, D] for E_ in range(4))
Rscalar = 6 * (afun * sp.diff(afun, t, 2) + sp.diff(afun, t)**2) / afun**2
check("Eq. (34): (1/2) eps eps e e R = 2 sqrt(-g) R (exact, FLRW)", "S", sp.simplify(dens - 2 * G_F['sqrtg'] * Rscalar) == 0)
def lcsign(p):
    p = list(p); s_ = 1
    for i in range(4):
        for j in range(i + 1, 4):
            if p[i] > p[j]: s_ = -s_
            elif p[i] == p[j]: return 0
    return s_
LC4 = np.zeros((4, 4, 4, 4))
for p in permutations(range(4)): LC4[p] = lcsign(p)
IN = invariants(GN)
Riem2f, Rsf, Euf, sqgf = [sp.lambdify(coN, q) for q in (IN['Riem2'], IN['R'], IN['E'], GN['sqrtg'])]
LamMM = 0.7
def lattice_functionals(hh):
    Emat = np.array([edge(x0, mu, hh) / hh for mu in range(4)]).T          # e^a_mu from edge molecules
    Einv = np.linalg.inv(Emat)
    Rl = np.zeros((4, 4, 4, 4))
    for mu in range(4):
        for nu in range(mu + 1, 4):
            Pm = (plaq(x0, mu, nu, hh) - I4) / hh**2; Rl[:, :, mu, nu] = Pm; Rl[:, :, nu, mu] = -Pm
    Rframe = np.einsum('abmn,mc,nd->abcd', Rl, Einv, Einv)
    Rlow = np.einsum('ae,ebcd->abcd', eta, Rframe); Rup = np.einsum('abcd,aA,bB,cC,dD->ABCD', Rlow, eta, eta, eta, eta)
    RupAB = np.einsum('abmn,bc->acmn', Rl, eta)
    EE = np.einsum('am,bn->abmn', Emat, Emat); EE = EE - EE.transpose(0, 1, 3, 2)
    Fm = RupAB - LamMM / 3 * EE
    return (np.einsum('abcd,abcd->', Rlow, Rup),                                   # metric-weighted holonomy functional
            0.5 * np.einsum('mnrs,abcd,am,bn,cdrs->', LC4, LC4, Emat, Emat, RupAB),  # Einstein-Cartan density
            0.25 * np.einsum('mnrs,abcd,abmn,cdrs->', LC4, LC4, Fm, Fm))             # MacDowell-Mansouri density
CONT = (Riem2f(*x0), 2 * sqgf(*x0) * Rsf(*x0), sqgf(*x0) * Euf(*x0) - 4 * LamMM / 3 * sqgf(*x0) * (Rsf(*x0) - 2 * LamMM))
HL = [0.1, 0.05, 0.025, 0.0125]; LF = np.array([lattice_functionals(hh) for hh in HL])
ERR = np.abs(LF - np.array(CONT))
print("continuum: Riem2 = %.6f   2 sqrt(-g) R = %.6f   MM density = %.6f" % CONT)
for hh, vals, er in zip(HL, LF, ERR): print("h=%7.4f:  %.6f (err %.2e)   %.6f (err %.2e)   %.6f (err %.2e)" % (hh, vals[0], er[0], vals[1], er[1], vals[2], er[2]))
def conv_order(k): return np.polyfit(np.log(HL), np.log(ERR[:, k]), 1)[0]
check("Eq. (35): lattice Einstein-Cartan density -> 2 sqrt(-g) R, error O(h)", "N", 0.8 < conv_order(1) < 1.3, f"order {conv_order(1):.2f}")
```

### F. Curvature in the spectrum of ${\hat{T}}$

The transporters enter the constraint only through ${\hat{T}}$. Curvature is therefore visible in the spectral data that determine the physical states (Theorem 2).

**Proposition 14 (Spectral detection of curvature).** Let $\Gamma$ be connected, with girth (shortest cycle length) $g$, and let $U_{ij}\in{SU(2)}$.

1.  $\Delta_U$ has a zero mode if and only if the connection is flat, i.e. every holonomy is trivial.

2.  $\operatorname{tr} h^n$ is independent of $U$ for $n<g$, and $$\operatorname{tr} h^g(U)-\operatorname{tr} h^g({\mathbf{1}})=2g\,(-J)^g\sum_{c}\big(\operatorname{tr} U_c-2\big),
    \tag{36}$$ summed over all cycles $c$ of length $g$, with $U_c$ the holonomy around $c$. In particular, the spectrum is gauge invariant.

*Proof.* (a) $a^\dagger\Delta_Ua=\sum_{\langle ij\rangle}|a_i-U_{ij}a_j|^2=0$ requires $a_i=U_{ij}a_j$ on every link, i.e. a covariantly constant spinor, which must be fixed by every holonomy. A nontrivial ${SU(2)}$ element has eigenvalues $e^{\pm i\phi/2}\neq1$ and fixes no nonzero vector. (b) $\operatorname{tr} h^n$ is a sum over closed walks of length $n$ of products of matrix elements. A walk acquires a nontrivial $U$-dependence only if it encircles a cycle without retracing, which needs at least $g$ hops. At $n=g$ there are $2g$ such walks per cycle ($g$ starting points, two orientations), each contributing $(-J)^g\operatorname{tr} U_c$ (the ${SU(2)}$ trace is real), compared with $(-J)^g\operatorname{tr}{\mathbf{1}}_2$ for $U={\mathbf{1}}$. $\square$

For a small rotation angle $\phi_c$, $\operatorname{tr} U_c-2\simeq-\phi_c^2/4$, so the right side of Eq. (36) is proportional to Wilson's plaquette action [38, 39]; more generally, spectral invariants of covariant Laplacians encode curvature [40]. As it stands, however, this functional weights every plaquette equally. In the continuum it is Yang--Mills theory for the Lorentz group on the fixed coordinate geometry of the lattice, which is background dependent and is not yet a gravitational action. Section V G shows how weighting by molecular edge data turns it into one. Curvature also lifts the bottom of the spectrum of ${\hat{T}}$. For a four-cycle with $U_p=e^{-i\theta\sigma_z/2}$, $\lambda_{\min}(\Delta_U)=2-2\cos(\theta/8)$ (Fig. 2). Equation (36) is checked in the notebook for a square plaquette with random ${SU(2)}$ links, where it reads $8J^4(\operatorname{tr} U_p-2)$ and agrees to machine precision.

**Figure 2.** Spectrum of the connection Laplacian $\Delta_U$ on a four-cycle as a function of the holonomy angle $\theta$ ($U_p=e^{-i\theta\sigma_z/2}$). Every level is a Kramers doublet (Proposition 6). The zero mode exists only for the flat connection, and curvature lifts it as $2-2\cos(\theta/8)$ (Proposition 14). Reproduced in the companion notebook.

**Verification of Proposition 14 and Eq. (36), and generation of Fig. 2.** On the square (girth 4) and on the triangle-plus-pendant graph of Sec. III (girth 3): $\operatorname{tr}h^n$ is independent of $U$ for $n<g$, and $\operatorname{tr}h^g(U)-\operatorname{tr}h^g(\mathbf 1)=2g(-J)^g(\operatorname{tr}U_c-2)$. The lowest eigenvalue of $\Delta_U$ on the square is $2-2\cos(\theta/8)$.

```python
LSQ = [(0, 1), (1, 2), (2, 3), (3, 0)]
def h_sq(Ulist): return eps0 * np.eye(8) + Jc * conn_laplacian(dict(zip(LSQ, Ulist)), LSQ, 4)
rngS = np.random.default_rng(9)
Us = [su2(rngS) for _ in range(4)]; Up = Us[0] @ Us[1] @ Us[2] @ Us[3]
dn = [np.trace(np.linalg.matrix_power(h_sq(Us), n_) - np.linalg.matrix_power(h_sq([np.eye(2)] * 4), n_)).real for n_ in range(1, 5)]
check("Eq. (36), square (g=4)", "N", max(abs(v) for v in dn[:3]) < 1e-12 and np.isclose(dn[3], 2 * 4 * Jc**4 * (np.trace(Up).real - 2)),
      f"tr h^4 shift {dn[3]:.6e} vs {8 * Jc**4 * (np.trace(Up).real - 2):.6e}")
UtrS = Ucurv[(0, 1)] @ Ucurv[(1, 2)] @ Ucurv[(2, 0)]
dt = [np.trace(np.linalg.matrix_power(h, n_) - np.linalg.matrix_power(eps0 * np.eye(2 * NS) + Jc * conn_laplacian(Uflat), n_)).real for n_ in range(1, 4)]
check("Eq. (36), triangle + pendant (g=3)", "N", max(abs(v) for v in dt[:2]) < 1e-12 and np.isclose(dt[2], 2 * 3 * (-Jc)**3 * (np.trace(UtrS).real - 2)))
Lg = [su2(rngS) for _ in range(4)]
Ug = [Lg[i] @ Us[k] @ Lg[j].conj().T for k, (i, j) in enumerate(LSQ)]
check("Prop. 14: spectrum gauge invariant", "N", np.abs(np.linalg.eigvalsh(h_sq(Ug)) - np.linalg.eigvalsh(h_sq(Us))).max() < 1e-12)
thetas = np.linspace(0, 2 * np.pi, 181)
specs = np.array([np.linalg.eigvalsh(conn_laplacian(dict(zip(LSQ, [expm(-0.5j * th_ * sz), np.eye(2), np.eye(2), np.eye(2)])), LSQ, 4)) for th_ in thetas])
check("Prop. 14: zero mode iff flat; lambda_min = 2 - 2 cos(theta/8)", "N",
      abs(specs[0, 0]) < 1e-12 and specs[1:, 0].min() > 0 and np.allclose(specs[:, 0], 2 - 2 * np.cos(thetas / 8)))
def fig2(theta_mark=None):
    plt.figure(figsize=(4.6, 3.3))
    for k in range(8): plt.plot(thetas, specs[:, k], 'k-', lw=1)
    if theta_mark is not None:
        sp_ = np.linalg.eigvalsh(conn_laplacian(dict(zip(LSQ, [expm(-0.5j * theta_mark * sz)] + [np.eye(2)] * 3)), LSQ, 4))
        plt.axvline(theta_mark, color='tab:red', lw=0.8); plt.plot([theta_mark] * 8, sp_, 'o', color='tab:red', ms=4)
    plt.xlabel(r'holonomy angle $\theta$  ($U_p=e^{-i\theta\sigma_z/2}$)'); plt.ylabel(r'eigenvalues of $\Delta_U$')
    plt.title('Kramers-doubled spectrum; curvature lifts the zero mode', fontsize=9)
    plt.xticks([0, np.pi, 2 * np.pi], ['0', r'$\pi$', r'$2\pi$']); plt.tight_layout()
fig2(); plt.savefig('fig_spectrum_holonomy.pdf'); plt.savefig('fig_spectrum_holonomy.png', dpi=150); plt.show()
if HAVE_WIDGETS:
    W.interact(lambda theta: (fig2(theta), plt.show()), theta=W.FloatSlider(min=0, max=2 * np.pi, step=0.05, value=1.0, description='theta'))
```

### G. Quadratic functionals built from ${\hat{T}}$ and the molecules

The pre-geometry supplies two ingredients for gravitational functionals: the plaquette holonomies $\Lambda_p$ of the transporters in ${\hat{T}}$, and the edge vectors $e^a(x,\mu)$ carried by bond molecules. The Einstein--Cartan functional (35) is linear in $\Lambda_p-{\mathbf{1}}$. Traces of holonomies, which is what the spectrum of ${\hat{T}}$ produces (Proposition 14), are instead quadratic.

**Lemma 1 (Holonomy traces are quadratic in curvature).** For every plaquette, with no sum over $\mu,\nu$, $$\begin{aligned}
\operatorname{tr}\Lambda_p(x;\mu\nu)-4&=\tfrac{h^4}{2}\operatorname{tr}\!\big(R_{\mu\nu}R_{\mu\nu}\big)+O(h^5)\\
&=-\tfrac{h^4}{2}\,R^{ab}{}_{\mu\nu}R_{ab\,\mu\nu}+O(h^5).
\end{aligned}
\tag{37}$$

*Proof.* By the Baker--Campbell--Hausdorff formula, $\Lambda_p=\exp Z$ with $Z=h^2R_{\mu\nu}+O(h^3)\in\mathfrak{so}(1,3)$. Every element of $\mathfrak{so}(1,3)$ is traceless, so $\operatorname{tr} e^Z-4=\tfrac12\operatorname{tr} Z^2+O(Z^3)$. $\square$

The linear term is absent, so no trace of a holonomy alone can produce the Einstein--Hilbert density. Weighting the plaquettes by molecular edge data makes the quadratic functional geometric.

**Verification of Lemma 1 and Eq. (37).** $(\operatorname{tr}\Lambda_p-4)\big/\big[\tfrac{h^4}{2}\operatorname{tr}(R_{\mu\nu}R_{\mu\nu})\big]\to1$; the linear term is absent because $\mathfrak{so}(1,3)$ generators are traceless.

```python
rat = []
for hh in [0.1, 0.05, 0.025]:
    Rm = np.array(Rfun[(0, 1)](*(x0 + 0.5 * hh * (I4[0] + I4[1]))), float)
    rat.append((np.trace(plaq(x0, 0, 1, hh)) - 4) / (hh**4 / 2 * np.trace(Rm @ Rm)))
print("ratios:", np.round(rat, 6))
check("Lemma 1: tr Lambda_p - 4 = (h^4/2) tr(R R) + O(h^5)", "N", abs(rat[-1] - 1) < 0.05 and abs(rat[-1] - 1) < abs(rat[0] - 1))
```

**Proposition 15 (Metric-weighted holonomy functional).** Let $\hat e_a{}^\mu(x)$ be the inverse of the matrix $e^a(x,\mu)/h$, and define the lattice curvature $\hat R^a{}_{bcd}(x)=\sum_{\mu\nu}\big[(\Lambda_p(x;\mu\nu)-{\mathbf{1}})/h^2\big]^a{}_b\,\hat e_c{}^\mu\hat e_d{}^\nu$. Then $$\begin{aligned}
S_W&\equiv\sum_xh^4\det\!\big[e^a(x,\mu)/h\big]\,\hat R_{abcd}\hat R^{abcd}\\
&\;\longrightarrow\;\int\mathrm{d}^4x\sqrt{-g}\,R_{abcd}R^{abcd},
\end{aligned}
\tag{38}$$ and in four dimensions $$R_{abcd}R^{abcd}=2\,C_{abcd}C^{abcd}+\tfrac13R^2-E,
\tag{39}$$ with $C_{abcd}$ the Weyl tensor and $E=R_{abcd}R^{abcd}-4R_{ab}R^{ab}+R^2$ the Euler (Gauss--Bonnet) density.

*Proof.* Convergence follows from Eqs. (27) and (33). In four dimensions $C_{abcd}C^{abcd}=R_{abcd}R^{abcd}-2R_{ab}R^{ab}+\tfrac13R^2$; eliminating $R_{ab}R^{ab}$ between this and the definition of $E$ gives Eq. (39). $\square$

The notebook verifies Eq. (39) symbolically for a general static, spherically symmetric metric with two free functions (where $C^2\neq0$ and $R\neq0$) and for FLRW. It also verifies that the lattice functional converges linearly in $h$. Since $\int\sqrt{-g}\,E$ is topological in four dimensions, the metric-weighted holonomy functional is Stelle's quadratic gravity [41, 42], with the Weyl-squared and $R^2$ coefficients in the fixed ratio $6:1$.

**Verification of Proposition 15 and Eq. (39).** [S] $\mathrm{Riem}^2=2C^2+\tfrac13R^2-E$ on a static spherically symmetric metric with two free functions, where $C^2\neq0$ and $R\neq0$ (heavy; fallback: Schwarzschild, where $R=0$, and FLRW, where $C^2=0$). [N] The lattice functional converges linearly in $h$.

```python
Af = sp.Function('A', positive=True)(r); Bf = sp.Function('B', positive=True)(r)
cases = ([("static spherical A(r), B(r)", [t, r, thv, phv], sp.diag(sp.sqrt(Af), sp.sqrt(Bf), r, r * sp.sin(thv)))] if RUN_HEAVY else
         [("Schwarzschild", None, None)]) + [("FLRW a(t)", None, None)]
INV = {}
for name, co_, e_ in cases:
    Gd = geometry(co_, e_) if co_ is not None else (GEO["Schwarzschild"] if name == "Schwarzschild" else G_F)
    Iv = invariants(Gd); INV[name] = Iv
    check(f"Eq. (39), {name}", "S", sp.simplify(Iv['Riem2'] - (2 * Iv['C2'] + Iv['R']**2 / 3 - Iv['E'])) == 0,
          f"C2 = 0: {sp.simplify(Iv['C2']) == 0}, R = 0: {sp.simplify(Iv['R']) == 0}")
check("Prop. 15: lattice holonomy functional -> Riem^2, error O(h)", "N", 0.8 < conv_order(0) < 1.3, f"order {conv_order(0):.2f}")
```

**Proposition 16 (Chiral holonomy).** In the left-atom representation (10), $$\operatorname{tr} S(\theta)S(\theta')=-\tfrac14\,\theta_{ab}\theta'^{ab}+\tfrac{i}{8}\,\epsilon_{abcd}\,\theta^{ab}\theta'^{cd}.
\tag{40}$$ Consequently the metric-weighted density built from the spin-$\tfrac12$ holonomies of the left atoms, $\operatorname{tr}[S(R_{\mu\nu})S(R^{\mu\nu})]$, has real part $-\tfrac14R_{abcd}R^{abcd}$ and imaginary part proportional to $\epsilon_{abcd}R^{ab}{}_{ef}R^{cdef}$, the Chern--Pontryagin density, which is a total derivative.

*Proof.* $S=\tfrac12A^i\sigma_i$ with $A^i$ the self-dual combination of Eq. (10), so $\operatorname{tr} SS'=\tfrac12A^iA'^i$; the real part of $A^iA'^i$ is $-\tfrac12\theta_{ab}\theta'^{ab}$ and the imaginary part is the mixed term $\theta_{i0}\epsilon_{ijk}\theta'_{jk}+(\theta\leftrightarrow\theta')$. The coefficients are confirmed by an exact fit in the companion notebook. $\square$

Chirality therefore does not remove the Weyl-squared term; it adds only a topological imaginary part.

**Verification of Proposition 16 and Eq. (40).** [N] least-squares fit over random generators; [S] exact identity in the six components of each generator.

```python
rng = np.random.default_rng(10)
rowsF, rhsF = [], []
for _ in range(20):
    A_ = rng.normal(size=(4, 4)); th1 = A_ - A_.T; B_ = rng.normal(size=(4, 4)); th2 = B_ - B_.T
    up1, up2 = eta @ th1 @ eta, eta @ th2 @ eta
    rowsF.append([np.sum(th1 * up2), 1j * np.einsum('abcd,ab,cd->', LC4, up1, up2)]); rhsF.append(np.trace(S_of(th1) @ S_of(th2)))
solF = np.linalg.lstsq(np.array(rowsF), np.array(rhsF), rcond=None)[0]
check("Eq. (40): coefficients -1/4 and 1/8", "N", np.allclose(solF, [-0.25, 0.125]), f"fit a={solF[0].real:.6f}, b={solF[1].real:.6f}")
def S_symb(th_):
    S_ = sp.zeros(2)
    for i in range(3):
        S_ += -sp.Rational(1, 2) * th_[0, i + 1] * SIGs[i + 1]
        for j in range(3):
            for k in range(3): S_ += sp.I / 4 * sp.LeviCivita(i, j, k) * th_[i + 1, j + 1] * SIGs[k + 1]
    return S_
uS = sp.symbols('u01 u02 u03 u12 u13 u23', real=True); thS2 = sp.zeros(4)
for v_, (a, b) in zip(uS, idx): thS2[a, b] = v_; thS2[b, a] = -v_
up1S, up2S = etaS * thS * etaS, etaS * thS2 * etaS
rhsS = -sp.Rational(1, 4) * sum(thS[a, b] * up2S[a, b] for a in range(4) for b in range(4)) \
       + sp.I / 8 * sum(sp.LeviCivita(a, b, c_, d_) * up1S[a, b] * up2S[c_, d_] for a, b, c_, d_ in permutations(range(4)))
check("Eq. (40): exact identity", "S", sp.expand((S_symb(thS) * S_symb(thS2)).trace() - rhsS) == 0)
```

**The de Sitter-shifted functional.** The holonomy can also be combined with the molecular area bivector before squaring. Define $$F^{ab}=R^{ab}-\tfrac{\Lambda}{3}\,e^a\wedge e^b ,
\tag{41}$$ whose lattice form is $F^{ab}_p=[\Lambda_p-{\mathbf{1}}]^{ab}/h^2-\tfrac\Lambda3[e(x,\mu)\wedge e(x,\nu)]^{ab}/h^2$.

**Proposition 17 (MacDowell--Mansouri functional).** With $\tilde\epsilon^{0123}=\epsilon_{0123}=+1$, $$\tfrac14\tilde\epsilon^{\mu\nu\rho\sigma}\epsilon_{abcd}\,F^{ab}{}_{\mu\nu}F^{cd}{}_{\rho\sigma}
=\sqrt{-g}\,E-\tfrac{4\Lambda}{3}\sqrt{-g}\,\big(R-2\Lambda\big).
\tag{42}$$ Hence $$\begin{aligned}
S_{\rm MM}&\equiv-\frac{3}{64\pi G\Lambda}\int\epsilon_{abcd}\,F^{ab}\wedge F^{cd}\\
&=-\frac{3}{64\pi G\Lambda}\int\mathrm{d}^4x\sqrt{-g}\,E\\
&\quad+\frac{1}{16\pi G}\int\mathrm{d}^4x\sqrt{-g}\,(R-2\Lambda),
\end{aligned}
\tag{43}$$ and $F^{ab}$ vanishes identically on de Sitter space with cosmological constant $\Lambda$.

*Proof.* Expand $F\wedge F$. The $R\wedge R$ term gives $\sqrt{-g}\,E$; the cross term reduces, by the normalization identity of Eq. (34), to $-\tfrac{4\Lambda}{3}\sqrt{-g}\,R$; and the quartic term uses $\tilde\epsilon^{\mu\nu\rho\sigma}\epsilon_{abcd}e^a{}_\mu e^b{}_\nu e^c{}_\rho e^d{}_\sigma=24\sqrt{-g}$. On de Sitter space $R^{ab}=\tfrac\Lambda3e^a\wedge e^b$. The notebook verifies Eq. (42) symbolically for FLRW with arbitrary $a(t)$ and for Schwarzschild--de Sitter, verifies $F\equiv0$ on de Sitter, and verifies convergence of the lattice density. $\square$

This is the construction of MacDowell and Mansouri [43, 44]: a functional quadratic in a curvature whose only local content is general relativity with a cosmological constant, with no Weyl-squared term and no scalaron.

**Verification of Proposition 17 and Eqs. (41)–(43).** [S] $\tfrac14\tilde\epsilon\epsilon FF=\sqrt{-g}E-\tfrac{4\Lambda}{3}\sqrt{-g}(R-2\Lambda)$ for FLRW with general $a(t)$ and for Schwarzschild–de Sitter (heavy), and $F\equiv0$ on de Sitter. [N] The lattice MacDowell–Mansouri density converges linearly in $h$.

```python
Lsym, Hs, Ms = sp.symbols('Lambda H M', positive=True)
def mm_check(Gd, Lam):
    I_ = invariants(Gd); e_ = Gd['e']; Rmn_ = Gd['Rmn']
    Rup = lambda mu, nu, a_, b_: sum(Rmn_[(mu, nu)][a_, c_] * etaS[c_, b_] for c_ in range(4))
    ee = lambda mu, nu, a_, b_: e_[a_, mu] * e_[b_, nu] - e_[a_, nu] * e_[b_, mu]
    Ff = lambda mu, nu, a_, b_: Rup(mu, nu, a_, b_) - Lam / 3 * ee(mu, nu, a_, b_)
    DMM = 0; DE = 0
    for mu, nu, rho, sg in permutations(range(4)):
        l1 = sp.LeviCivita(mu, nu, rho, sg)
        for a_, b_, c_, d_ in permutations(range(4)):
            l2 = l1 * sp.LeviCivita(a_, b_, c_, d_)
            DMM += sp.Rational(1, 4) * l2 * Ff(mu, nu, a_, b_) * Ff(rho, sg, c_, d_); DE += sp.Rational(1, 4) * l2 * Rup(mu, nu, a_, b_) * Rup(rho, sg, c_, d_)
    res = sp.simplify(sp.simplify(DMM) - Gd['sqrtg'] * I_['E'] + sp.Rational(4, 3) * Lam * Gd['sqrtg'] * (I_['R'] - 2 * Lam))
    Fzero = all(sp.simplify(Ff(mu, nu, a_, b_)) == 0 for mu in range(4) for nu in range(4) for a_ in range(4) for b_ in range(4))
    return res, Fzero
res1, _ = mm_check(G_F, Lsym)
check("Eq. (42), FLRW a(t)", "S", res1 == 0)
if RUN_HEAVY:
    fS = 1 - 2 * Ms / r - Lsym * r**2 / 3
    res2, _ = mm_check(geometry([t, r, thv, phv], sp.diag(sp.sqrt(fS), 1 / sp.sqrt(fS), r, r * sp.sin(thv))), Lsym)
    check("Eq. (42), Schwarzschild-de Sitter (0<theta<pi)", "S", sp.simplify(res2.subs(sp.Abs(sp.sin(thv)), sp.sin(thv))) == 0)
else:
    check("Eq. (42), Schwarzschild-de Sitter", "S", None, "RUN_HEAVY = False")
res3, F0 = mm_check(geometry([t, x, y, zc], sp.diag(1, sp.exp(Hs * t), sp.exp(Hs * t), sp.exp(Hs * t))), 3 * Hs**2)
check("Eq. (41): F = 0 identically on de Sitter with Lambda = 3H^2", "S", F0 and res3 == 0)
check("Prop. 17: lattice MacDowell-Mansouri density converges, error O(h)", "N", 0.8 < conv_order(2) < 1.3, f"order {conv_order(2):.2f}")
lP2 = 6.674e-11 * 1.0546e-34 / 2.998e8**3; Lam_obs = 1.1e-52
MMcoef = 3 / (64 * np.pi * lP2 * Lam_obs)
print("dimensionless coefficient 3/(64 pi G Lambda) = %.1e" % MMcoef)
```

**Spinor form: the atoms as a de Sitter spinor.** In the Dirac representation in a Weyl basis, $\gamma^a=\left(\begin{smallmatrix}0&\sigma^a\\ \bar\sigma^a&0\end{smallmatrix}\right)$ with $\sigma^a=({\mathbf{1}},\vec\sigma)$ and $\bar\sigma^a=(-{\mathbf{1}},\vec\sigma)$, the Lorentz generators $S_D(\theta)=\tfrac14\theta_{ab}\gamma^a\gamma^b$ are block diagonal. The lower block is exactly $S(\theta)$ of Eq. (10), acting on left atoms, and the upper block is $-S(\theta)^\dagger$, the dual of the right-atom representation. A pair $({\backslash},{/})$ is therefore a Dirac spinor. Adding $P_a=\tfrac12\gamma_a\gamma_5$, the ten generators close into $\mathfrak{so}(4,1)$; the trace form has six negative directions, the compact $\mathfrak{so}(4)$. The $P_a$ are off diagonal: they convert ${/}$ into ${\backslash}$ and back. The connection $$\mathcal{A}=S_D(\omega)+\ell^{-1}\,e^aP_a,\qquad \Lambda=3/\ell^2 ,
\tag{44}$$ is an $\mathfrak{so}(4,1)$ connection whose curvature vanishes identically on de Sitter space. The notebook checks this directly, and checks that the $\mathfrak{so}(3,2)$ choice $P_a=\tfrac12\gamma_a$ does not vanish in the companion notebook. Its Lorentz part is the spinor image of Eq. (41), and its translational part is the torsion [44]. In the language of this paper, the diagonal blocks of $\mathcal{A}$ are the transporters of ${\hat{T}}$ (self-dual on ${/}$, anti-self-dual on ${\backslash}$), and the off-diagonal block is the tetrad: the bond molecules that connect ${/}$ to ${\backslash}$. The MacDowell--Mansouri functional is the quadratic form of the curvature of this single connection. A caveat applies. The constraint (13) conserves $\hat N_{/}$ and $\hat N_{\backslash}$ separately, whereas a quantum implementation of the off-diagonal block requires chirality-mixing terms such as $a_i^\dagger(e\cdot\sigma)b_j$. These are mirror even and do not commute with $\sum_i\hat D_i$, so they would break the exact solution of Sec. IV. How to incorporate them is open.

**Verification of Eq. (44).** In the Weyl basis: $S_D(\theta)=\tfrac14\theta_{ab}\gamma^a\gamma^b$ is block diagonal, with the left-atom block equal to Eq. (10); $P_a=\tfrac12\gamma_a\gamma_5$ is off-diagonal (it mixes $/$ and $\backslash$) and closes $\mathfrak{so}(4,1)$; $A=S_D(\omega)+\ell^{-1}e^aP_a$ is flat on de Sitter, whereas $P_a=\tfrac12\gamma_a$ gives $\mathfrak{so}(3,2)$ and is not flat there.

```python
Z2 = np.zeros((2, 2)); sigB = [-s0, sx, sy, sz]
gam = [np.block([[Z2, SIG[a_]], [sigB[a_], Z2]]) for a_ in range(4)]
g5 = np.block([[np.eye(2), Z2], [Z2, -np.eye(2)]])
gl = [sum(eta[a_, b_] * gam[b_] for b_ in range(4)) for a_ in range(4)]
SDn = lambda th_: 0.25 * sum(th_[a_, b_] * gam[a_] @ gam[b_] for a_ in range(4) for b_ in range(4))
rng = np.random.default_rng(11); A_ = rng.normal(size=(4, 4)); thr = A_ - A_.T
check("Eq. (44): Clifford algebra; S_D generates Lorentz; lower block = S(theta) (left atoms)", "N",
      all(np.allclose(gam[a_] @ gam[b_] + gam[b_] @ gam[a_], 2 * eta[a_, b_] * np.eye(4)) for a_ in range(4) for b_ in range(4))
      and all(np.allclose(gam[c_] @ SDn(thr) - SDn(thr) @ gam[c_], sum((eta @ thr)[c_, d_] * gam[d_] for d_ in range(4))) for c_ in range(4))
      and np.allclose(SDn(thr)[2:, 2:], S_of(thr)) and np.allclose(SDn(thr)[:2, :2], -S_of(thr).conj().T))
def Jg(a_, b_):
    th_ = np.zeros((4, 4)); th_[a_, b_] = 1; th_[b_, a_] = -1; return SDn(th_)
sigs = {}
for name, P in [("dS", [0.5 * g_ @ g5 for g_ in gl]), ("AdS", [0.5 * g_ for g_ in gl])]:
    gens = [Jg(a_, b_) for a_ in range(4) for b_ in range(a_ + 1, 4)] + P
    evK = np.linalg.eigvalsh(np.array([[np.trace(X_ @ Y_).real for Y_ in gens] for X_ in gens]))
    sigs[name] = (int(np.sum(evK > 1e-9)), int(np.sum(evK < -1e-9)), all(np.allclose(p_[:2, :2], 0) and np.allclose(p_[2:, 2:], 0) for p_ in P))
check("Eq. (44): P_a = gamma_a gamma_5/2 is chirality-mixing; algebra so(4,1)", "N", sigs["dS"] == (4, 6, True), f"trace-form signature {sigs['dS'][:2]}")
gamS = [sp.Matrix(g_) for g_ in gam]; glS = [sp.Matrix(g_) for g_ in gl]; g5S = sp.Matrix(g5)
SDs = lambda thl: sp.Rational(1, 4) * sum((thl[a_, b_] * gamS[a_] * gamS[b_] for a_ in range(4) for b_ in range(4)), sp.zeros(4))
GdS = geometry([t, x, y, zc], sp.diag(1, sp.exp(Hs * t), sp.exp(Hs * t), sp.exp(Hs * t)))
worst = {}
for name, P in [("AdS", [g_ / 2 for g_ in glS]), ("dS", [g_ * g5S / 2 for g_ in glS])]:
    Acon = [SDs(etaS * GdS['om'][mu]) + Hs * sum((GdS['e'][a_, mu] * P[a_] for a_ in range(4)), sp.zeros(4)) for mu in range(4)]
    wv = 0
    for mu in range(4):
        for nu in range(mu + 1, 4):
            Fc = sp.simplify(sp.diff(Acon[nu], [t, x, y, zc][mu]) - sp.diff(Acon[mu], [t, x, y, zc][nu]) + Acon[mu] * Acon[nu] - Acon[nu] * Acon[mu])
            wv = max(wv, max(abs(complex(sp.N(v_.subs({t: 0.3, Hs: 1.0})))) for v_ in Fc))
    worst[name] = wv
check("Eq. (44): A = S_D(omega) + H e^a P_a is flat on de Sitter (and not for so(3,2))", "S",
      worst["dS"] < 1e-12 and worst["AdS"] > 0.1, f"max|F| dS = {worst['dS']:.1e}, AdS = {worst['AdS']:.2f}")
```

### H. Which gravitational dynamics? Quadratic gravity as the natural target

Everything above holds for any choice of transporters. In the minimal model of Sec. III the $U_{ij}$ are background data; the constraint is then solved exactly, and the kinematic reconstruction of tetrad, connection, torsion, and curvature goes through. Gravity requires more: the transporters must become dynamical, and the Wheeler--DeWitt constraint must then select a gravitational action in the continuum. Four structural results constrain which action that can be.

**Proposition 18 (The constraint has no intrinsic scale).** For every $\lambda\neq0$, $\ker(\lambda{\hat{H}})=\ker{\hat{H}}$. Writing ${\hat{H}}=\epsilon_0\big[\sum_i\hat D_i+(J/\epsilon_0)\sum_{\langle ij\rangle}(\hat B^{/}_{ij}-\hat B^{\backslash}_{ij})\big]$, the physical states depend only on the dimensionless ratio $J/\epsilon_0$ and on the transporters $U_{ij}$.

The proof is immediate. The consequence is not. The pre-geometry contains no dimensionful constant, so any continuum action derived from its physical states can contain dimensionful couplings only through a scale that is generated dynamically. In four dimensions $\int\sqrt{-g}\,R^2$ and $\int\sqrt{-g}\,C^2$ are dimensionless, whereas the Einstein--Hilbert and cosmological terms require the dimensionful coefficients $1/G$ and $\Lambda/G$. *Quadratic gravity is therefore the natural continuum target for this pre-geometry*: it is what a scale-free microscopic theory can produce with pure-number coefficients, while the linear term must be induced. Scale invariance is likewise a standard motivation for quadratic gravity [45].

**Verification of Proposition 18.** [N] $\ker(\lambda\hat H)=\ker\hat H$ as subspaces (one- and two-molecule sectors of the fermionic model). [L], [R] theorem `kernel_scale`.

```python
S11, S22 = sector(1, 1), sector(2, 2)
proj = lambda M_: (lambda K2: K2 @ K2.conj().T)(null_space(M_))
ok_scale = all(np.allclose(proj(S11), proj(lam_ * S11)) and np.allclose(proj(S22), proj(lam_ * S22)) for lam_ in (0.37, 5.0, -2.0))
check("Prop. 18: ker(lambda H) = ker H", "N", ok_scale)
```

Second, Proposition 7 suggests, but does not establish, that the Sakharov route [7] is closed. The static vacuum energy of the atoms is independent of $U$. Once the links are dynamical, however, fluctuations of the two mirror sectors could still generate derivative-dependent terms in an effective action for $U$ and $e$, and deciding this requires the fluctuation determinants of the dynamical model. If the cancellation does extend to the full effective action, any gravitational action must come from the *molecular* content of physical states, for example from condensates such as those of Proposition 11.

Third, the functionals of Secs. V E and V G define three distinct channels, compared in Table III.

**Table III.** Gravitational functionals that can be assembled from plaquette holonomies of ${\hat{T}}$ and molecular edge vectors. $E$ is the Euler density and $P$ the Pontryagin density; both are topological.

| Channel                   | Molecular origin                                           | Continuum content                  | Propagating modes                                                      | Status                                                                    |
|:--------------------------|:-----------------------------------------------------------|:-----------------------------------|:-----------------------------------------------------------------------|:--------------------------------------------------------------------------|
| (a) Stelle type         | metric-weighted holonomy (Props. 15 and 16)                | $2C^2+R^2/3$ (up to $E$ and $iP$)  | with an Einstein term: graviton, massive spin-2 ghost, scalaron [42] | renormalizable [41]; alone, one massive mode is tachyonic (Prop. 19)    |
| (b) Starobinsky type    | square of the local density in Eq. (35)                    | $\alpha R+\beta R^2$               | graviton and scalaron with $m_f^2=\alpha/6\beta$; no ghost [45]      | inflaton of Sec. VI F [46, 47]; $\alpha\neq0$ required                  |
| (c) MacDowell--Mansouri | $\epsilon$-weighted de Sitter-shifted curvature (Prop. 17) | $(R-2\Lambda)/16\pi G$ (up to $E$) | graviton only                                                          | GR with $\Lambda$; coefficient $3/(64\pi G\Lambda)\approx5\times10^{119}$ |

Fourth, the linear term cannot vanish at low energies. For channel (b), take $S=(16\pi G_0)^{-1}\int\sqrt{-g}\,f(R)$ with $f(R)=\alpha R+\beta R^2$ and expand about a background of constant curvature $\bar R$. The tensor kinetic term is multiplied by $f'(\bar R)=\alpha+2\beta\bar R$, so the effective coupling is $G_{\rm eff}=G_0/f'(\bar R)$. The scalaron mass is $$m_f^2=\frac13\left(\frac{f'(\bar R)}{f''(\bar R)}-\bar R\right)=\frac{\alpha}{6\beta},
\tag{45}$$ independent of the background. In the scale-invariant limit $\alpha=0$, the tensor mode propagates only because $\bar R\neq0$, in agreement with Ref. [45]: a curved molecular condensate induces the Einstein term, with $G_{\rm eff}=G_0/(2\beta\bar R)$. But Eq. (45) then gives a *massless* scalaron. Metric $f(R)$ gravity is equivalent to Brans--Dicke theory with $\omega=0$ [48], and a massless Brans--Dicke scalar gives $\gamma_{\rm PPN}=(1+\omega)/(2+\omega)=\tfrac12$, whereas the Cassini time-delay measurement gives $\gamma_{\rm PPN}-1=(2.1\pm2.3)\times10^{-5}$ [49]. The Einstein-frame potential of pure $R^2$ is constant, $(f'R-f)/f'^2=1/4\beta$, so no chameleon-type screening is available. Pure $R^2$ is therefore excluded as the low-energy theory, and some linear channel, either Eq. (35) or channel (c), must survive. Quadratic gravity remains the natural form at the fundamental level; the linear term is the scale generated by the condensate.

**Verification of Eq. (45) and the pure-$R^2$ limit.** For $f=\alpha R+\beta R^2$: $m_f^2=\tfrac13(f'/f''-\bar R)=\alpha/6\beta$; $\alpha=0$ gives a massless scalaron, a constant Einstein-frame potential $1/4\beta$, and $\gamma_{\rm PPN}=\tfrac12$ (Brans–Dicke $\omega=0$).

```python
al, be, Rb, R_ = sp.symbols('alpha beta Rbar R', positive=True)
f = al * R_ + be * R_**2; fp = sp.diff(f, R_); fpp = sp.diff(f, R_, 2)
mf2 = sp.simplify((fp / fpp).subs(R_, Rb) / 3 - Rb / 3)
check("Eq. (45): m_f^2 = alpha/(6 beta)", "S", sp.simplify(mf2 - al / (6 * be)) == 0, f"m_f^2 = {mf2}")
V_E = sp.simplify(((fp * R_ - f) / fp**2).subs(al, 0))
check("pure R^2: massless scalaron, constant Einstein-frame potential 1/(4 beta)", "S", mf2.subs(al, 0) == 0 and sp.simplify(V_E - 1 / (4 * be)) == 0)
omBD = 0; gamma_pureR2 = sp.Rational(1 + omBD, 2 + omBD)
check("pure R^2: gamma_PPN = (1+omega)/(2+omega) = 1/2", "S", gamma_pureR2 == sp.Rational(1, 2))
xi_, Phib = sp.symbols('xi Phibar', positive=True)
fFL = R_ * (1 - xi_ * Phib**2) + be * R_**2
mFL = sp.simplify((sp.diff(fFL, R_) / sp.diff(fFL, R_, 2)).subs(R_, Rb) / 3 - Rb / 3)
check("Farmer-Larson f = R(1 - xi Phi^2) + beta R^2: m_f^2 = (1 - xi Phi^2)/(6 beta)", "S", sp.simplify(mFL - (1 - xi_ * Phib**2) / (6 * be)) == 0)
```

This result connects directly to the Starobinsky-plus-scalar phenomenology of Ref. [50]. There $f=R(1-\xi\Phi^2)+\beta R^2$, and Eq. (45) reproduces the scalaron mass $m_f^2=(1-\xi\bar\Phi^2)/6\beta$ found in that work, with $\alpha=1-\xi\bar\Phi^2$. In molecular language the scalaron range $\sqrt{6\beta/\alpha}$ measures the ratio of the quadratic to the linear molecular coefficient. Its observational bound is discussed in Sec. VII.

Fifth, channel (a) cannot stand on its own.

**Proposition 19 (The Stelle-type channel alone is unstable).** Let $S=(16\pi G)^{-1}\int\mathrm{d}^4x\sqrt{-g}\,\big(R+c\,R_{abcd}R^{abcd}\big)$ with $c\neq0$, the Einstein term plus the metric-weighted holonomy functional of Proposition 15. The massive scalar and massive spin-2 modes about flat space have $$m_0^2=\frac{1}{2c},\qquad m_2^2=-\frac{1}{4c},\qquad\text{so}\qquad m_0^2=-2\,m_2^2 .
\tag{46}$$ For every $c\neq0$ exactly one of the two modes is tachyonic.

*Proof.* Since $\int\sqrt{-g}\,E$ is topological, $R_{abcd}R^{abcd}\simeq4R_{ab}R^{ab}-R^2$, so the action has the form $R+aR^2+bR_{ab}R^{ab}$ with $a=-c$ and $b=4c$. For this action the masses are $m_0^2=1/[2(3a+b)]$ and $m_2^2=-1/b$ [42]; the normalization is fixed by the $b=0$ limit, which must reproduce Eq. (45) with $\alpha=1$, i.e. $m_0^2=1/6a$. Absence of tachyons requires $3a+b>0$ and $b<0$, that is, $c>0$ and $c<0$ at once. Equivalently, in the Weyl basis $R_{abcd}R^{abcd}\simeq2C^2+R^2/3$ [Eq. (39)], the $R^2$ coefficient $c/3$ and the $C^2$ coefficient $2c$ have the same sign, whereas stability requires opposite signs. $\square$

A viable quadratic sector therefore cannot consist of holonomy traces alone. It must be dominated by channels (b) or (c), or must include additional terms that change the ratio fixed in Eq. (39).

**Verification of Proposition 19 and Eq. (46).** Two independent parametrizations of the masses are compared. (i) Ricci basis, $R+aR^2+bR_{ab}R^{ab}$: $m_0^2=1/[2(3a+b)]$, $m_2^2=-1/b$ (Stelle). (ii) Weyl basis, $R+\beta_RR^2+\alpha_WC^2$: $m_0^2=1/(6\beta_R)$, $m_2^2=-1/(2\alpha_W)$ (the normalization of Eq. (45) and of Alvarez-Gaume et al.). The basis change uses $C^2=\mathrm{Riem}^2-2\mathrm{Ric}^2+\tfrac13R^2$, checked here on the metric of Proposition 15 with $C^2$ computed from the Weyl tensor. [L] Lean theorem `stelle_channel_tachyon`.

```python
Iv = INV.get("static spherical A(r), B(r)", INV.get("Schwarzschild"))
check("C^2 = Riem^2 - 2 Ric^2 + R^2/3 (C^2 from the Weyl tensor)", "S", sp.simplify(Iv['C2'] - (Iv['Riem2'] - 2 * Iv['Ric2'] + Iv['R']**2 / 3)) == 0)
c = sp.symbols('c', real=True)
a_R, b_R = -c, 4 * c                      # Riem^2 = 4 Ric^2 - R^2 + E
beta_W, alpha_W = c / 3, 2 * c            # Riem^2 = 2 C^2 + R^2/3 - E
m0_i, m2_i = 1 / (2 * (3 * a_R + b_R)), -1 / b_R
m0_ii, m2_ii = 1 / (6 * beta_W), -1 / (2 * alpha_W)
a0 = sp.symbols('a0', positive=True)
check("Stelle normalization: b = 0 limit gives m0^2 = 1/(6a), i.e. Eq. (45) with alpha = 1", "S",
      sp.simplify(1 / (2 * (3 * a0 + 0)) - 1 / (6 * a0)) == 0)
check("Eq. (46): both bases give m0^2 = 1/(2c), m2^2 = -1/(4c)", "S",
      sp.simplify(m0_i - m0_ii) == 0 and sp.simplify(m2_i - m2_ii) == 0 and sp.simplify(m0_i - 1 / (2 * c)) == 0 and sp.simplify(m2_i + 1 / (4 * c)) == 0)
sol = sp.reduce_inequalities([3 * a_R + b_R > 0, b_R < 0], c)
check("Prop. 19: no c makes both masses non-tachyonic", "S", sol == sp.false, f"solution set: {sol}")
```

In channel (c) the only coupling is the pure number $3/(64\pi G\Lambda)$, about $5\times10^{119}$ for the observed $\Lambda$. The cosmological-constant hierarchy thus reappears as the size of a dimensionless molecular coefficient. The model does not explain it.

The main open problem is now sharply posed. The $U_{ij}$ must be promoted to quantum link variables, in the manner of Kogut and Susskind [39] and with the corresponding Gauss constraints, and one must determine which combination of the channels in Table III the constraint selects: the coefficients $\alpha$, $\beta$, the Weyl-squared coefficient, and $\Lambda$. A second structural requirement concerns the constraint itself. ${\hat{H}}$ is a sum of local terms but is imposed as a single global constraint, whereas canonical general relativity has local Hamiltonian and momentum constraints whose algebra is the hypersurface-deformation algebra [1, 51]. Refining Eq. (13) into local constraints that reproduce this algebra, or otherwise demonstrating emergent diffeomorphism invariance, is required before the Wheeler--DeWitt interpretation is complete. Two structural hints point to specific channels. The spectrum of ${\hat{T}}$ naturally produces channel (a). The $\mathfrak{so}(4,1)$ structure of Eq. (44), in which the bond molecules are the off-diagonal block, points to channel (c). In self-dual language the linear term is $\Sigma_{AB}\wedge F^{AB}$ with $\Sigma^{AB}=e^{A}{}_{A'}\wedge e^{BA'}$ the Plebański 2-form [20, 52], and $\Sigma^{AB}$ is built from the molecular order parameter $\langle a_Ab_{A'}\rangle$ of two edges meeting at a patch. We therefore *conjecture* that, in molecular-condensate states, closed-loop transport weighted by the patch bivectors produces the linear term.

## VI. De Sitter asymptotics from observation and a dS/CFT completion

Sections II--V are internal to the model: they construct physical states and reconstruct geometry, but they do not select a vacuum. The observed universe does. Two facts fix its asymptotics. First, the expansion is accelerating with an equation of state consistent with a positive cosmological constant, $\Lambda=3\Omega_\Lambda H_0^2/c^2\simeq1.09\times10^{-52}\,{\rm m}^{-2}$ [53], so the far future is approximately de Sitter with radius $\ell=\sqrt{3/\Lambda}\simeq1.66\times10^{26}$ m. Second, the primordial perturbations are adiabatic, nearly Gaussian, and nearly scale invariant [47, 54], the signature of an earlier quasi-de Sitter phase in its Bunch--Davies state. We take these facts as boundary conditions instead of deriving them, and we use the dS/CFT correspondence [55, 56, 57] to turn them into a statement about the physical states of Eq. (13). Boundary conditions restrict states; they do not introduce an ambient spacetime, so background freedom in the sense of Sec. I B is unaffected. Every statement in this section is an exact identity checked in the companion notebook, a postulate imposed by observation, or a labeled conjecture.

### A. The de Sitter boundary condition

**Postulate 1 (de Sitter asymptotics).** Physical states of Eq. (13) are restricted to those whose large-scale geometry, reconstructed as in Sec. V, approaches de Sitter space with the observed $\Lambda$ at late times. Their late-time wave functional is the partition function of a Euclidean three-dimensional conformal field theory on the future boundary $\mathcal{I}^+$, $$\Psi\big[h_{ij},\phi_0\big]=Z_{\rm CFT}\big[h_{ij},\phi_0\big],
\tag{47}$$ where $h_{ij}$ is the conformal class of the late-time spatial metric and $\phi_0$ are the late-time values of the other fields, which act as sources [57].

Equation (47) is the natural partner of a Wheeler--DeWitt constraint: a wave functional with no external time is identified with a partition function on a spacelike surface. In the canonical reading appropriate to a Hamiltonian constraint, the graph $\Gamma$ with ${SU(2)}$ links is a spatial slice and $\mathcal{I}^+$ is the late-time slice. This is also why the Fock inner product selects ${SU(2)}$ (Sec. III E): it is the rotation group of a spacelike surface, and, by Corollary 2, the site-local order parameter of a molecular condensate in a curved region is that surface's timelike normal.

**Where the boundary is.** De Sitter space is conformally flat. In planar coordinates $$\begin{gathered}
\mathrm{d} s^2=\Omega(\eta)^2\big(-\mathrm{d}\eta^2+\delta_{ij}\mathrm{d} x^i\mathrm{d} x^j\big),\\
\Omega(\eta)=\ell/(-\eta),\qquad -\infty<\eta<0,
\end{gathered}
\tag{48}$$ and $\mathcal{I}^+$ is $\eta\to0^-$. By Corollary 1, the only geometric datum a patch carries is its conformal factor $\Omega^2=\sum_{\alpha<\beta}|\omega_{\alpha\beta}|^2$, which measures how strongly its atoms pair. Because the atoms are fermions, the order parameter carried by one flavor is bounded, $|\langle b\,a\rangle|\le\tfrac12$ (Proposition 10). The order parameter of a patch, and with it the pairings $\omega_{\alpha\beta}$, grows only with the number of molecules the patch contains. The future boundary is therefore the limit of divergent pairing, $\Omega^2\to\infty$: the limit in which each coarse-grained patch of the molecular superfluid holds an unbounded number of molecules. It is defined by the states of the atoms alone.

**Verification of Eq. (48) and of the molecular location of $\mathcal{I}^+$.** [N] $\Lambda$ and $\ell$ from the Planck 2018 parameters; [S] planar de Sitter in conformal time is conformally flat ($C^2=0$) with $R=12/\ell^2$, and $\Omega\to\infty$ as $\eta\to0^-$; [N] the Pauli principle bounds the order parameter of one flavor by $\tfrac12$, so the order parameter of a patch grows only with the number of molecules it holds: $N_f$ flavors condensed along the same $\hat\xi$ give $N_f\Delta\,\hat\xi\bar{\hat\xi}$.

```python
c_light = 2.99792458e8; G_N = 6.67430e-11; hbar = 1.054571817e-34
H0_si = 67.36e3 / 3.0857e22; OmL = 0.6847                        # Planck 2018 LCDM
Lam_obs = 3 * OmL * H0_si**2 / c_light**2; ell_dS = np.sqrt(3 / Lam_obs)
check("Sec. VI: Lambda = 3 Omega_L H0^2/c^2 ~ 1.09e-52 m^-2, ell = sqrt(3/Lambda) ~ 1.66e26 m", "N",
      1.07e-52 < Lam_obs < 1.11e-52 and 1.64e26 < ell_dS < 1.68e26, f"Lambda = {Lam_obs:.3e} m^-2, ell = {ell_dS:.3e} m")
eta_c = sp.symbols('eta', negative=True); ellS = sp.symbols('ell', positive=True)
Om_c = ellS / (-eta_c)
G_conf = geometry([eta_c, x, y, zc], sp.diag(Om_c, Om_c, Om_c, Om_c)); I_conf = invariants(G_conf)
check("Eq. (48): planar de Sitter is conformally flat (C^2 = 0) with R = 12/ell^2; Omega -> oo at eta -> 0^-", "S",
      sp.simplify(I_conf['C2']) == 0 and sp.simplify(I_conf['R'] - 12 / ellS**2) == 0 and sp.limit(Om_c, eta_c, 0, '-') == sp.oo)
uu_ = np.linspace(0, np.pi / 2, 2001)
check("Sec. VI: one flavor: |<b a>| = |cos t sin t| <= 1/2 over all condensates (Pauli bound)", "N",
      np.isclose(np.max(np.abs(np.cos(uu_) * np.sin(uu_))), 0.5) and abs(Delta) <= 0.5)
Xal = [(sum(xih[A] * aF2[al][A].T for A in range(2)) @ sum(xih[Ap].conj() * bF2[al][Ap].T for Ap in range(2))).tocsr() for al in range(2)]
Psi2f = vac8.astype(complex)
for al in range(2): Psi2f = uC * Psi2f + vC * (Xal[al] @ Psi2f)
Psi2f /= np.linalg.norm(Psi2f)
op2 = np.array([[sum(np.vdot(Psi2f, bF2[al][Ap] @ (aF2[al][A] @ Psi2f)) for al in range(2)) for Ap in range(2)] for A in range(2)])
check("Sec. VI: N_f = 2 flavors condensed along xi: total order parameter = 2 Delta xi conj(xi) (grows with molecule number)", "N",
      np.allclose(op2, 2 * Delta * ellh), f"|order parameter| = {np.linalg.norm(op2):.4f} = 2 x {np.linalg.norm(Delta * ellh):.4f}")
```

### B. Boundary conformal symmetry from the atom spinor

The isometry group of four-dimensional de Sitter space, $SO(4,1)$, acts on $\mathcal{I}^+$ as the conformal group of three-dimensional Euclidean space. Section V G showed that a pair $({\backslash},{/})$ carries the Dirac representation of $\mathfrak{so}(4,1)$, generated by $J_{ab}\equiv S_D(e_{ab})$ and $P_a=\tfrac12\gamma_a\gamma_5$, where $(e_{ab})_{cd}=\delta_{ac}\delta_{bd}-\delta_{ad}\delta_{bc}$. The atoms therefore carry the boundary conformal algebra.

**Proposition 20 (Boundary conformal algebra).** Define, for $i,j\in\{1,2,3\}$, $$D=P_0,\quad \mathcal{P}_i=J_{0i}+P_i,\quad \mathcal{K}_i=J_{0i}-P_i,\quad M_{ij}=J_{ij}.
\tag{49}$$ Then $[D,\mathcal{P}_i]=\mathcal{P}_i$, $[D,\mathcal{K}_i]=-\mathcal{K}_i$, $[\mathcal{K}_i,\mathcal{P}_j]=2(M_{ij}-\delta_{ij}D)$, and $[D,M_{ij}]=[\mathcal{P}_i,\mathcal{P}_j]=[\mathcal{K}_i,\mathcal{K}_j]=0$, while $\mathcal{P}_i$ and $\mathcal{K}_i$ transform as vectors under the rotations $M_{ij}$. This is the conformal algebra of $\mathbb{R}^3$: $D$ is the dilatation, $\mathcal{P}_i$ the translations, $\mathcal{K}_i$ the special conformal transformations, and $M_{ij}$ the rotations. In the atom representation, $M_{ij}$ is block diagonal and anti-Hermitian, acting on ${/}$ and ${\backslash}$ by the same ${SU(2)}$, whereas $D$ is Hermitian and off diagonal, with eigenvalues $\pm\tfrac12$ on the combinations ${\backslash}\pm{/}$ of a single two-spinor.

*Proof.* Direct computation in the Weyl basis of Sec. V G. The notebook verifies every relation exactly. $\square$

Proposition 20 gives each piece of the model a boundary meaning. The rotations of $\mathcal{I}^+$ are the compact ${SU(2)}$ of the transporters in ${\hat{T}}$, which treat both species alike. The dilatation is the bulk time translation $P_0$. It is the one boundary generator that converts ${/}$ into ${\backslash}$, and its eigenstates are equal-weight superpositions of the two species. Bulk evolution toward $\mathcal{I}^+$ is boundary rescaling. This gives a concrete candidate for the arrow of time left open in Sec. IX: the orientation of information flow (${/}$ counted positive, ${\backslash}$ negative) corresponds to the direction of scale on the boundary, as in the reading of cosmological evolution as inverse renormalization-group flow [58].

**Verification of Proposition 20.** [N] and [S]: with $D=P_0$, $\mathcal P_i=J_{0i}+P_i$, $\mathcal K_i=J_{0i}-P_i$, $M_{ij}=J_{ij}$ built from the Dirac generators of Eq. (44), every relation of the three-dimensional Euclidean conformal algebra holds; $M_{ij}$ is block diagonal and anti-Hermitian, $D$ Hermitian, off diagonal, with eigenvalues $\pm\frac12$.

```python
Comm = lambda A_, B_: A_ @ B_ - B_ @ A_
Dg = 0.5 * gl[0] @ g5
Pc = [Jg(0, i) + 0.5 * gl[i] @ g5 for i in (1, 2, 3)]
Kc = [Jg(0, i) - 0.5 * gl[i] @ g5 for i in (1, 2, 3)]
Mc = {(i, j): (Jg(i, j) if i != j else np.zeros((4, 4), complex)) for i in (1, 2, 3) for j in (1, 2, 3)}   # M_ii = 0
dl = lambda i, j: 1.0 if i == j else 0.0
ok_alg = (all(np.allclose(Comm(Dg, Pc[i]), Pc[i]) and np.allclose(Comm(Dg, Kc[i]), -Kc[i]) for i in range(3))
          and all(np.allclose(Comm(Kc[i], Pc[j]), 2 * (Mc[(i + 1, j + 1)] - dl(i, j) * Dg)) for i in range(3) for j in range(3))
          and all(np.allclose(Comm(Dg, Mc[(i, j)]), 0) for i in (1, 2, 3) for j in (1, 2, 3))
          and all(np.allclose(Comm(Pc[i], Pc[j]), 0) and np.allclose(Comm(Kc[i], Kc[j]), 0) for i in range(3) for j in range(3)))
# vectors under rotations: [M_ij, X_k] = sgn (delta_jk X_i - delta_ik X_j); find the sign once
sgn = None
for s_try in (+1, -1):
    if all(np.allclose(Comm(Mc[(i, j)], X_[k - 1]), s_try * (dl(j, k) * X_[i - 1] - dl(i, k) * X_[j - 1]))
           for X_ in (Pc, Kc) for i in (1, 2, 3) for j in (1, 2, 3) for k in (1, 2, 3)):
        sgn = s_try
check("Prop. 20: D, P_i, K_i, M_ij close into the conformal algebra of R^3", "N", ok_alg and sgn is not None, f"vector rule sign {sgn:+d}" if sgn else "")
evD = np.sort(np.linalg.eigvals(Dg).real)
check("Prop. 20: M_ij block diagonal, anti-Hermitian (common SU(2) on / and \\); D Hermitian, off diagonal, eigenvalues +-1/2", "N",
      all(np.allclose(Mc[(i, j)][:2, 2:], 0) and np.allclose(Mc[(i, j)], -Mc[(i, j)].conj().T) for i, j in [(1, 2), (1, 3), (2, 3)])
      and np.allclose(Dg, Dg.conj().T) and np.allclose(Dg[:2, :2], 0) and np.allclose(Dg[2:, 2:], 0) and np.allclose(evD, [-0.5, -0.5, 0.5, 0.5]))
# exact
JgS = lambda a_, b_: sp.zeros(4) if a_ == b_ else SDs(sp.Matrix(4, 4, lambda c_, d_: (1 if (c_, d_) == (a_, b_) else (-1 if (c_, d_) == (b_, a_) else 0))))
DS = glS[0] * g5S / 2
PS_ = [JgS(0, i) + glS[i] * g5S / 2 for i in (1, 2, 3)]; KS_ = [JgS(0, i) - glS[i] * g5S / 2 for i in (1, 2, 3)]
CS = lambda A_, B_: sp.simplify(A_ * B_ - B_ * A_)
check("Prop. 20: exact relations [D,P]=P, [D,K]=-K, [K_i,P_j]=2(M_ij - delta_ij D)", "S",
      all(CS(DS, PS_[i]) == PS_[i] and CS(DS, KS_[i]) == -KS_[i] for i in range(3))
      and all(CS(KS_[i], PS_[j]) == 2 * (JgS(i + 1, j + 1) - (1 if i == j else 0) * DS) for i in range(3) for j in range(3)))
```

### C. Physical states as boundary operators

A bulk spin-$\tfrac12$ field of mass $m$ approaches $\mathcal{I}^+$ as $(-\eta)^{3/2\pm i\mu}$ with $\mu=m\ell$ [Eq. (51) below], for every $m>0$. In cosmic time, $-\eta=\ell e^{-t/\ell}$, these are $e^{-3t/2\ell}e^{\mp i\mu t/\ell}$: positive and negative frequency, which the boundary theory records as an operator of weight $\Delta=\tfrac32+i\mu$ and its shadow $3-\Delta=\tfrac32-i\mu$. The mirror sectors of Eq. (18) have the same structure, with energy $+\epsilon_k$ for ${/}$ and $-\epsilon_k$ for ${\backslash}$. This suggests the following dictionary.

**Conjecture 1 (Shadow dictionary).** Under Postulate 1, a left atom in normal mode $k$ corresponds to a boundary operator of weight $\Delta_k=\tfrac32+i\mu_k$, and its mirror partner to the shadow weight $3-\Delta_k=\tfrac32-i\mu_k$, with $\mu_k=\epsilon_k/\epsilon_*$ for a single constant $\epsilon_*$.

Because the atoms are spin-$\tfrac12$ fermions, the conjecture has the form of the standard spin-$\tfrac12$ entry of the dictionary, $\mu_k=m_k\ell$, and needs no threshold mass, unlike the scalar case.

If the conjecture holds, the following are exact consequences of Theorem 2 and Proposition 7.

1.  A Fock state $|m,n\rangle$ has total weight $\tfrac32\sum_k(m_k+n_k)+i\sum_k\mu_k(m_k-n_k)$. The constraint ${\hat{H}}|\Psi\rangle =0$ is the statement that the total weight is real: ${\rm Im}\,\Delta_{\rm tot}=E(m,n)/\epsilon_*=0$.

2.  For rationally independent levels the physical states are products of shadow pairs $\mathcal{O}_{\Delta}\mathcal{O}_{3-\Delta}$ [Theorem 2(c)]. Each molecule has weight exactly $3$: it is a marginal operator of the three-dimensional boundary theory, consistent with the absence of an intrinsic scale (Proposition 18).

3.  A state of $r$ molecules has weight $3r$. As a deformation of the boundary theory its coupling scales as $(-\eta)^{3r-3}$, so every interaction among two or more molecules vanishes at $\mathcal{I}^+$, and faster the more molecules take part.

4.  Zero-point cancellation (Proposition 7) is the statement that the spectrum is symmetric under $\Delta\leftrightarrow3-\Delta$.

The conjecture is not derived here. The notebook checks only that it is consistent with the exact solution: the states of real total weight are exactly the kernel of ${\hat{H}}$ in Table I.

**Consistency of Conjecture 1 with Theorem 2.** With $\mu_k=\epsilon_k/\epsilon_*$ for the normal-mode energies of Table I, count the fermionic Fock states (each mode empty or singly occupied) of every sector whose total weight is real, and compare with the kernel dimensions found there. Also tabulate the late-time scaling exponent $3r-3$ of an $r$-molecule deformation.

```python
eps_star = 1.0
mu_k = w / eps_star                                    # normal-mode energies of h (8 modes, Kramers pairs)
def real_weight_count(nL_, nR_):
    cnt = 0; re_ok = True
    for cL in itertools.combinations(range(len(mu_k)), nL_):
        for cR in itertools.combinations(range(len(mu_k)), nR_):
            if abs(sum(mu_k[k] for k in cL) - sum(mu_k[k] for k in cR)) < 1e-9:
                cnt += 1; re_ok &= np.isclose(1.5 * (nL_ + nR_), 3 * nL_)
    return cnt, re_ok
cnts = {(p_, q_): real_weight_count(p_, q_) for p_ in range(NM + 1) for q_ in range(NM + 1)}
check("Conj. 1: real-weight states = ker H in every sector (total 1296), each of weight 3r", "N",
      all(cnts[k_][0] == kerdim[k_] and cnts[k_][1] for k_ in cnts), f"one molecule {cnts[(1, 1)][0]}, two {cnts[(2, 2)][0]}, total {sum(v_[0] for v_ in cnts.values())}")
print("late-time scaling exponent of an r-molecule deformation, 3r-3:", {r_: 3 * r_ - 3 for r_ in range(1, 6)})
```

### D. The cosmological constant as a count of boundary degrees of freedom

In dS/CFT the number of boundary degrees of freedom scales as $\ell^2/G$ [55], the combination that sets the Gibbons--Hawking entropy $S_{\rm dS}=A/4G=\pi\ell^2/G$ [59]. The pure-number coefficient of channel (c) is exactly this entropy: $$\frac{3}{64\pi G\Lambda}=\frac{S_{\rm dS}}{64\pi^2},\qquad S_{\rm dS}=\frac{3\pi}{G\Lambda}\simeq3.3\times10^{122}
\tag{50}$$ (with $\hbar=c=1$). The large number of Sec. V H is therefore the de Sitter entropy, and the observed cosmological constant is $\Lambda=3\pi/(GS_{\rm dS})$: the inverse of the number of boundary degrees of freedom. Because the atoms carry no zero-point energy (Proposition 7), $\Lambda$ is not a vacuum energy of the pre-geometry. This does not explain why $S_{\rm dS}$ is so large, and it does not remove the vacuum energy of matter fields: with Postulate 1 imposed, the matter contribution is part of the observed value.

**Verification of Eq. (50).** [S] $3/(64\pi G\Lambda)=S_{\rm dS}/64\pi^2$ with $S_{\rm dS}=A/4G=\pi\ell^2/G=3\pi/(G\Lambda)$; [N] the numerical values with the observed $\Lambda$.

```python
GS, LS = sp.symbols('G Lambda', positive=True)
ellS2 = sp.sqrt(3 / LS); S_dS = (4 * sp.pi * ellS2**2) / (4 * GS)
check("Eq. (50): 3/(64 pi G Lambda) = S_dS/(64 pi^2), S_dS = A/4G = 3 pi/(G Lambda)", "S",
      sp.simplify(3 / (64 * sp.pi * GS * LS) - S_dS / (64 * sp.pi**2)) == 0 and sp.simplify(S_dS - 3 * sp.pi / (GS * LS)) == 0)
lP2_si = G_N * hbar / c_light**3
S_dS_num = 3 * np.pi / (lP2_si * Lam_obs)
check("Eq. (50): S_dS ~ 3.3e122 and the channel-(c) coefficient ~ 5e119", "N",
      3.2e122 < S_dS_num < 3.4e122 and np.isclose(S_dS_num / (64 * np.pi**2), 3 / (64 * np.pi * lP2_si * Lam_obs)),
      f"S_dS = {S_dS_num:.2e}, S_dS/64pi^2 = {S_dS_num / (64 * np.pi**2):.2e}")
```

### E. Matter: the Standard Model on the molecular network

**Coupling principle.** By Eq. (10), a left atom transforms under the self-dual half of a Lorentz transformation and a right atom under the anti-self-dual half. Every left-handed Weyl field of the Standard Model transforms the same way, as $(\tfrac12,0)$, and every right-handed field as $(0,\tfrac12)$; like the atoms, these fields are fermions. Standard-Model fermions therefore couple to the emergent geometry through the transporters of ${\hat{T}}$ themselves: a left-handed field on patch $j$ is carried to patch $i$ by $U_{ij}$, and a right-handed field by $\bar U_{ij}$ (by the ${SL(2,\mathbb{C})}$ transporters of Sec. V once boosts are included). Gauge fields enter as additional internal link variables $W_{ij}\in SU(3)\times SU(2)_{\rm W}\times U(1)_Y$ on the same graph, in the Kogut--Susskind form [39], and the Higgs doublet as a site variable. The minimal model does not fix the gauge group, the representations, or the number of generations, and nothing in it conflicts with them. The chiral structure of the Standard Model, left-handed doublets and right-handed singlets, uses exactly the two Lorentz representations carried by ${/}$ and ${\backslash}$.

**Boundary content.** Under Postulate 1 each bulk field maps to boundary operators whose weights follow from its mass and spin. For a scalar and a spin-$\tfrac12$ field, $$\Delta_\pm=\tfrac32\pm\sqrt{\tfrac94-m^2\ell^2}\quad\text{and}\quad\Delta_\pm=\tfrac32\pm im\ell ,
\tag{51}$$ respectively. Massless gauge fields give conserved currents with $\Delta=2$, and the graviton gives the boundary stress tensor with $\Delta=3$. Table IV lists the Standard-Model content at the two de Sitter epochs. Among the asymptotic late-time states, only the graviton and the photon give real weights, so unbroken $U(1)_{\rm em}$ is a global symmetry of the boundary theory. Every massive particle appears as a principal-series pair with $\mu\simeq m\ell$, between $10^{31}$ (neutrinos) and $10^{44}$ (top quark); the spin-dependent corrections to $\mu$ are negligible at these values. During inflation (Sec. VI F) every Standard-Model mass is more than ten orders of magnitude below the Hubble rate, so fermions and gauge fields are nearly conformal. Complex weights make the boundary theory non-unitary, as expected for de Sitter duals [55, 60]. The best-studied explicit example, the $Sp(N)$ model dual to higher-spin gravity in de Sitter space, uses anticommuting scalars [60], a boundary theory in which spin and statistics are not tied together. In the bulk model of this paper, by contrast, the atoms obey the usual connection between spin and statistics.

```python
hbarc_GeVm = 1.973269804e-16; MP_red = 2.435e18; A_s = 2.1e-9; N_ref = 55
M_scal = np.sqrt(24 * np.pi**2 * A_s) / N_ref * MP_red; H_inf = M_scal / 2
SM = [("neutrino nu3", 0.050e-9, "4.2e+31", "3e-24"),
      ("electron", 0.51099895e-3, "4.3e+38", "3e-17"), ("W", 80.3692, "6.8e+43", "5e-12"), ("Z", 91.1880, "7.7e+43", "6e-12"),
      ("Higgs", 125.20, "1.1e+44", "8e-12"), ("top", 172.57, "1.5e+44", "1e-11"), ("scalaron", M_scal, "2.6e+55", None)]
ok_tab = True
print(f"{'field':24s} {'mass [GeV]':>12s} {'mu = m ell':>12s} {'m/H_inf':>10s}")
for name, m_, mu_exp, mh_exp in SM:
    mu_ = m_ * ell_dS / hbarc_GeVm; mh = m_ / H_inf
    print(f"{name:24s} {m_:12.4g} {mu_:12.2e} {mh:10.1e}")
    if mu_exp: ok_tab &= f"{mu_:.1e}" == mu_exp
    if mh_exp: ok_tab &= f"{mh:.0e}" == mh_exp
# scalar weight formula: principal series for m ell > 3/2, complementary below
Dl = lambda mell: (1.5 + np.sqrt(complex(2.25 - mell**2)), 1.5 - np.sqrt(complex(2.25 - mell**2)))
check("Table IV: tabulated mu = m ell and m/H_inf reproduced; Eq. (51) gives shadow pairs Delta_+ + Delta_- = 3", "N",
      ok_tab and all(np.isclose(sum(Dl(v_)), 3) for v_ in (0.5, 1.4, 3.0, 1e5)), f"H_inf = {H_inf:.2e} GeV")
```

### F. Inflation as an earlier de Sitter phase

The second observational boundary condition, nearly scale-invariant primordial perturbations, is met by channels already present. Combining channels (b) and (c), $$S=\frac{1}{16\pi G}\int\mathrm{d}^4x\sqrt{-g}\left(R-2\Lambda+\frac{R^2}{6M^2}\right),
\tag{52}$$ which is Eq. (45) with $\alpha=1$ and $\beta=1/6M^2$, so $m_f=M$. The scalaron is the inflaton of Starobinsky's model [46, 62]. In the Einstein frame its potential is $V=\tfrac34M^2M_P^2\big(1-e^{-\sqrt{2/3}\,\phi/M_P}\big)^2$, with $M_P=(8\pi G)^{-1/2}$, and slow roll gives, to leading order in $1/N$, $$n_s\simeq1-\frac2N,\qquad r\simeq\frac{12}{N^2},\qquad A_s\simeq\frac{N^2M^2}{24\pi^2M_P^2},
\tag{53}$$ where $N$ is the number of $e$-folds between horizon exit of the pivot scale and the end of inflation. The measured amplitude $A_s\simeq2.1\times10^{-9}$ [53] fixes $M\simeq3.1\times10^{13}$ GeV and $H_{\rm inf}\simeq M/2$ for $N=55$. Three consistency checks follow. First, the coefficient of $R^2$ in Eq. (52) is the pure number $$\frac{\beta}{16\pi G}=\frac{M_P^2}{12M^2}=\frac{N^2}{288\pi^2A_s}\simeq5\times10^{8},
\tag{54}$$ as Proposition 18 requires of a quadratic coefficient. Second, $\beta/\alpha=1/6M^2\simeq7\times10^{-60}\,{\rm m}^2$, some fifty orders of magnitude below the laboratory bound (55), so the inflaton is the scalaron of P3 and leaves P1--P4 of Sec. VII unchanged. Third, in dS/CFT the inflationary phase is a second, approximate de Sitter boundary [58], with entropy $S_{\rm inf}=8\pi^2M_P^2/H_{\rm inf}^2\simeq2\times10^{12}$. Between the two epochs the number of boundary degrees of freedom grows by a factor of about $10^{110}$, the direction required if cosmological time is inverse renormalization-group flow on the boundary [58].

**Verification of Eqs. (52)–(54).** [S] the Einstein-frame potential of $f=R+R^2/6M^2$ and the leading slow-roll observables; [N] $M$, $H_{\rm inf}$, the pure-number $R^2$ coefficient, $\beta/\alpha$ against the laboratory bound, and the inflationary de Sitter entropy.

```python
Rs_, Ms_, Mp_, ph_ = sp.symbols('R M M_P phi', positive=True)
f_st = Rs_ + Rs_**2 / (6 * Ms_**2); fp_st = sp.diff(f_st, Rs_)
R_of_phi = sp.solve(sp.Eq(fp_st, sp.exp(sp.sqrt(sp.Rational(2, 3)) * ph_ / Mp_)), Rs_)[0]
V_E = sp.simplify((Mp_**2 / 2) * ((Rs_ * fp_st - f_st) / fp_st**2).subs(Rs_, R_of_phi))
V_target = sp.Rational(3, 4) * Ms_**2 * Mp_**2 * (1 - sp.exp(-sp.sqrt(sp.Rational(2, 3)) * ph_ / Mp_))**2
check("Eq. (52): Einstein-frame potential = (3/4) M^2 M_P^2 (1 - exp(-sqrt(2/3) phi/M_P))^2", "S", sp.simplify(V_E - V_target) == 0)
xS, NS_ = sp.symbols('x N', positive=True)          # x = exp(sqrt(2/3) phi/M_P);  N ~ 3x/4 at leading order
eps_x = sp.simplify((Mp_**2 / 2 * (sp.diff(V_target, ph_) / V_target)**2).subs(ph_, sp.sqrt(sp.Rational(3, 2)) * Mp_ * sp.log(xS)))
eta_x = sp.simplify((Mp_**2 * sp.diff(V_target, ph_, 2) / V_target).subs(ph_, sp.sqrt(sp.Rational(3, 2)) * Mp_ * sp.log(xS)))
ns_N = sp.series((1 - 6 * eps_x + 2 * eta_x).subs(xS, 4 * NS_ / 3), NS_, sp.oo, 2).removeO()
r_N = sp.series((16 * eps_x).subs(xS, 4 * NS_ / 3), NS_, sp.oo, 3).removeO()
As_N = sp.simplify((V_target / (24 * sp.pi**2 * Mp_**4 * eps_x.subs(xS, 4 * NS_ / 3))).subs(sp.exp(-sp.sqrt(sp.Rational(2, 3)) * ph_ / Mp_), 0))
check("Eq. (53): n_s = 1 - 2/N, r = 12/N^2 at leading order", "S",
      sp.simplify(ns_N - (1 - 2 / NS_)) == 0 and sp.simplify(sp.limit(r_N * NS_**2, NS_, sp.oo) - 12) == 0)
coef = MP_red**2 / (12 * M_scal**2)
beta_m2 = (hbarc_GeVm / M_scal)**2 / 6
S_inf = 8 * np.pi**2 * (MP_red / H_inf)**2
check("Eqs. (53)-(54): M ~ 3.1e13 GeV; R^2 coefficient M_P^2/12M^2 = N^2/(288 pi^2 A_s) ~ 5e8", "N",
      3.0e13 < M_scal < 3.2e13 and np.isclose(coef, N_ref**2 / (288 * np.pi**2 * A_s)) and 4.5e8 < coef < 5.5e8, f"M = {M_scal:.3e} GeV, coefficient = {coef:.3e}")
check("Sec. VI F: beta/alpha ~ 7e-60 m^2, far below Eq. (55); S_inf ~ 2e12; S_dS/S_inf ~ 1e110", "N",
      beta_m2 < 1e-50 and 1.5e12 < S_inf < 2.5e12 and 109 < np.log10(S_dS_num / S_inf) < 111,
      f"beta = {beta_m2:.1e} m^2, S_inf = {S_inf:.1e}, log10 ratio = {np.log10(S_dS_num / S_inf):.1f}")
```

## VII. Observational consequences

The model is not yet complete enough to predict the values of $\alpha$, $\beta$, or $\Lambda$ from first principles. It does make structural statements that can be confronted with observation, and it converts existing bounds into constraints on its molecular coefficients. We list them in order of robustness and summarize them in Table V. None of them is unique to this model taken singly; together they are the conditions any completion must satisfy, and each would falsify a specific part of the construction if violated.

**P1. Luminal tensor waves.** In every channel the massless spin-2 mode propagates on the light cone of the single emergent metric, the light cone built from the atom pairing (Proposition 3). The gravitational and electromagnetic light cones therefore coincide. GW170817 and GRB 170817A bound the fractional difference between the speeds of gravity and light to lie between $-3\times10^{-15}$ and $+7\times10^{-16}$ [63]. A violation would falsify the identification of the gravitational light cone with the atomic one.

**P2. No long-range scalar.** The scale-invariant limit $\alpha=0$ gives a massless scalaron and $\gamma_{\rm PPN}=\tfrac12$ (Sec. V H), in conflict with the Cassini bound $\gamma_{\rm PPN}-1=(2.1\pm2.3)\times10^{-5}$ [49]. The model therefore requires an induced linear term, $\alpha\neq0$. With $\alpha\neq0$ and a scalaron range far below solar-system distances it predicts $\gamma_{\rm PPN}=1$ there [64].

**P3. A short-range scalaron.** In channel (b) the scalaron adds a Yukawa term of strength $\tfrac13$ and range $\lambda_f=m_f^{-1}=\sqrt{6\beta/\alpha}$ to the Newtonian potential [42, 64]. Torsion-balance tests of the inverse-square law exclude gravitational-strength Yukawa interactions with ranges above $38.6\,\mu$m [65]. For the strength-$\tfrac13$ scalaron of $f=R+a_2R^2/2$, earlier Eöt-Wash data give $|a_2|\lesssim2\times10^{-9}\,{\rm m}^2$, i.e. $\lambda_f=\sqrt{3a_2}\lesssim8\times10^{-5}$ m [64]; the newer data of Ref. [65] can only tighten this, and we use the conservative value. Since $\beta/\alpha=a_2/2$, this becomes $$\frac{\beta}{\alpha}=\frac{\lambda_f^2}{6}\lesssim1\times10^{-9}\ {\rm m}^2 ,
\tag{55}$$ a bound on the ratio of the quadratic to the linear molecular coefficient. The scalar (breathing) polarization can propagate only above the angular frequency $\omega_c=c/\lambda_f\gtrsim3.7\times10^{12}\,{\rm s}^{-1}$, so it is not excited by astrophysical sources [64]. The scalaron mass $1/6\beta$ is fixed by the action rather than by the environment, so the model supplies no screening. Analyses of $R+\beta R^2$ gravity that use $\beta$ far above Eq. (55), such as the pulsar-motivated scales of Ref. [50], therefore implicitly require the scalaron to be screened or decoupled from laboratory matter; within the molecular model no such mechanism has been identified.

**P4. Exactly two polarizations at detector frequencies.** Combining P1--P3: in channel (c) the only propagating mode is the massless graviton, and in channel (b) the scalaron lies above the cutoff of P3. LISA and ground-based detectors should see exactly the two tensor polarizations of general relativity, with no scalar component and no dispersion. Channel (a) alone is excluded by Proposition 19; its massive spin-2 mode would in addition be bounded by the same torsion-balance tests, since it contributes a Yukawa term of strength $-\tfrac43$ [42].

**P5. The atomic vacuum does not gravitate.** Zero-point energies cancel identically between the mirror sectors (Proposition 7), so the pre-geometry contributes no static vacuum energy; whether it also induces no Sakharov-type action is open. The cosmological constant enters only through the coefficient of channel (c), the pure number $3/(64\pi G\Lambda)\approx5\times10^{119}$, which under Postulate 1 is the de Sitter entropy divided by $64\pi^2$ [Eq. (50)]. The model relocates the cosmological-constant problem to the size of this number; it does not solve it.

**P6. Starobinsky tilt and tensor amplitude.** Equation (53) gives $n_s=0.960$--$0.967$ and $r=0.0048$--$0.0033$ for $N=50$--$60$ (Fig. 4). CMB data alone give $n_s=0.9682\pm0.0032$ and $r<0.034$ at 95% confidence (Planck, SPT, ACT, and BICEP/Keck combined [66]), consistent with the prediction for $N\gtrsim55$ (within $1.4\sigma$; $2.6\sigma$ at $N=50$). Adding DESI DR2 BAO raises the tilt to $n_s=0.9728\pm0.0029$ [66], and the Planck--ACT--DESI DR1 combination gives $0.974\pm0.003$ [67]; the former places the prediction $2.1\sigma$ ($N=60$) to $4.4\sigma$ ($N=50$) low. Whether the shift reflects inflationary physics or an inconsistency between BAO and CMB data is debated [68]. The decisive test is $r$. LiteBIRD targets $\delta r<0.001$ [69], enough to detect $r=0.0033$--$0.0048$ at 3--5$\sigma$; its launch is now planned for the 2030s [70], and ground-based polarization experiments will tighten the bound on $r$ in the meantime. A detection of $r\gtrsim0.01$, or a bound $r<0.002$, would exclude the $R^2$ channel as the inflaton.

**P7. A cosmological constant, not dynamical dark energy.** Postulate 1 in its minimal form, with $\Lambda$ supplied by channel (c), requires $w=-1$ at all late times. DESI DR2 BAO combined with the CMB prefers evolving dark energy at $3.1\sigma$, and at 2.8--4.2$\sigma$ when supernova samples are added [71]. Confirmation at high significance would require replacing the constant $\Lambda$ of channel (c) by a dynamical sector and would remove the exact de Sitter future on which Eq. (47) rests.

**P8. Single-field consistency.** With Bunch--Davies initial conditions, the squeezed bispectrum of a single-field model is fixed by the tilt, $f_{\rm NL}^{\rm local}=\tfrac{5}{12}(1-n_s)\simeq0.015$ [57]. In dS/CFT this relation is a Ward identity of the boundary conformal symmetry. Planck gives $f_{\rm NL}^{\rm local}=-0.9\pm5.1$ [54], and SPHEREx, launched in 2025, targets an uncertainty of order unity [72]. A detection of $|f_{\rm NL}^{\rm local}|\gtrsim1$ would exclude the single-scalaron picture.

## VIII. Formal verification in Lean 4 and Rocq

The algebraic results of this paper are machine-checked in two proof assistants with different logical foundations and independent kernels: Lean 4 [73] (`MolecularWDW.lean`) and Rocq, formerly Coq [74] (`MolecularWDW.v`). A single proof assistant certifies a proof relative to its own kernel and encoding; a second, independent formalization guards against errors in either. The two files were written separately from the informal statements and use different encodings. Both construct the fermionic ladder operators on the occupation basis, with each mode empty or singly occupied, and Jordan--Wigner signs for the mode-interleaved ordering $\prod_k(c_k^\dagger)^{m_k}(d_k^\dagger)^{n_k}|0\rangle$. Lean indexes normal modes by `Fin K`, computes the signs from finite sums, and proves polynomial identities over the integers; this implies them in every commutative ring, since an integer-coefficient identity that holds for all integer arguments is an identity of polynomials. Rocq stores occupations in Boolean lists, computes the signs from prefix counts, and proves polynomial identities over an arbitrary commutative ring. Both use only their standard libraries. Table VI maps each formalized result to its theorem in each file.

**Trusted base.** In Lean, every theorem depends at most on the three standard axioms `propext`, `Quot.sound`, and `Classical.choice`; the conformal-algebra computations depend on none. In Rocq, every statement about the exact solution, the spinor identities, and the conformal algebra is closed under the global context, that is, it uses no axioms at all. The six real-number identities use only the two axioms of Rocq's real-number library (`sig_forall_dec` and `functional_extensionality_dep`). Neither file contains `sorry`, `admit`, or `Admitted`, and each prints the assumptions of every main theorem. Both build in one to two minutes: `lean MolecularWDW.lean` (Lean 4.15.0) and `coqc MolecularWDW.v` (Coq 8.18.0).

**Statement fidelity.** A proof assistant certifies that a formal statement follows from its definitions; it does not certify that the formal statement says what the paper says. Each formal statement is therefore written in terms of the paper's own definitions rather than a precomputed polynomial. For example, Proposition 3 is stated with the molecules ${X}^{AA'}_{\alpha\gamma}={/}^A_\alpha\bar{/}^{A'}_\gamma$ and the $\epsilon$ pairing, and Eq. (40) with $S(\theta)$ built from the Pauli matrices, $\epsilon_{ijk}$, $\epsilon_{abcd}$, and $\eta_{ab}$. The normalization differences are listed in Table VI. Both files were prepared with AI assistance by the same process (see Research methods). Their independence lies in language, logical foundations, kernel, and encoding, not in authorship; an independent human check of the correspondence in Table VI is the remaining safeguard.

**Not formalized.** The following are established in the text and checked by computer algebra or numerically, but are not formalized:

-   the reduction of Eq. (16) to normal modes, which rests on the spectral theorem (available in Mathlib, but not in Lean core or the Rocq standard library);

-   Theorem 2(c) for degenerate (Kramers-doublet) levels, which is the per-mode argument applied to level occupations;

-   Propositions 2 and 4 (representation theory);

-   the $\mathfrak{su}(2)$ matrix elements and the binomial-to-Poisson limit of Proposition 9 (the collective commutator follows from the formalized single-mode relations), the site-basis form of Proposition 13, and Corollary 2 (linear algebra);

-   the geometric statements of Sec. V: Eqs. (29) and (33), Table II, Lemma 1, Propositions 14--17, and Eq. (44);

-   the four-dimensional Weyl-tensor identity behind Eq. (39);

-   the slow-roll asymptotics of Eq. (53), and all observational statements.

Conjecture 1 is a conjecture and is not claimed.

**Formal verification [L].** The cell writes the Lean source (identical to the ancillary `MolecularWDW.lean`) if it is not already present, and compiles it when a Lean 4 toolchain is available (set `LEAN_BIN` if `lean` is not on the PATH). The compiler output is parsed: the file must compile without errors or `sorry`, and each main theorem may depend only on Lean's three standard axioms. Without a toolchain (for example in JupyterLite) the compile checks are reported as skipped; a static scan of the source still runs.

```python
LEAN_SRC = r'''/-
  MolecularWDW.lean
  =================

  Formal verification (Lean 4 core, no external libraries) of the algebraic
  statements underlying the paper

      "Spacetime Molecules: A Pre-geometric Wheeler-DeWitt Model Built from
       Fermionic Chiral Spinor Atoms"

  WHAT IS PROVED HERE
  -------------------
  Work in the normal-mode Fock representation of the model (paper, Sec. IV):
  K normal modes in the left-atom ("/") sector with ladder operators c_k, c_k^dag
  and K mirror modes in the right-atom ("\") sector with d_k, d_k^dag.  THE ATOMS ARE
  FERMIONS.  The Wheeler-DeWitt (Hamiltonian-constraint) operator is

        H = sum_k eps_k ( c_k^dag c_k  -  d_k^dag d_k ) .

  We construct the fermionic ladder operators explicitly on the occupation basis
  (occupations 0 or 1; mode-interleaved Jordan-Wigner signs) and prove, from those
  definitions alone:

   (1) canonical anticommutation relations: {c_k, c_l^dag} = delta_kl,
       {d_k, d_k^dag} = 1, c_k^dag c_l^dag = - c_l^dag c_k^dag, d_k^dag d_l^dag = -d_l^dag d_k^dag,
       c_k^dag d_l^dag = - d_l^dag c_k^dag, and the Pauli principle (c_k^dag)^2 = (d_k^dag)^2 = 0;
   (2) MOLECULES ARE BOSONS: X_k^dag = c_k^dag d_k^dag satisfy X_k^dag X_l^dag = X_l^dag X_k^dag,
       commute with the atom operators of every other mode, [X_k, X_l^dag] = 0 (k /= l),
       and obey the hard-core algebra [X_k, X_k^dag] = 1 - n^c_k - n^d_k, (X_k^dag)^2 = 0;
   (3) H is diagonal with eigenvalue E(m,n) = sum_k eps_k (m_k - n_k);
   (4) mirror antisymmetry  H M = - M H  (M exchanges "/" and "\");
   (5) H commutes with every molecule operator;
   (6) H |0> = 0;
   (7) MAIN THEOREM: every multi-molecule state X_{k_1}^dag ... X_{k_r}^dag |0> satisfies
       the Wheeler-DeWitt equation  H Psi = 0;
   (8) for distinct modes each such state IS the basis vector |occ, occ> (coefficient 1),
       states on different sets of modes are linearly independent, and a repeated mode
       gives zero (Pauli);
   (9) every BCS-type molecular condensate prod_k (u_k + v_k X_k^dag)|0> satisfies H Psi = 0;
  (10) the pair shift [H, d_l c_k] = (eps_l - eps_k) d_l c_k (molecular field equation);
  (11) for any split H = T + V,  H Psi = 0  is equivalent to  T Psi = - V Psi;
  (12) antisymmetric (Weyl) ordering: the fermionic zero-point constants cancel exactly,
       sum_k eps_k[(c^dag c - c c^dag) - (d^dag d - d d^dag)] = 2 H;
  (13) no-go lemma (expectation form);
  (14) scale freedom: for lam /= 0, ker(lam H) = ker H;
  (15) the sign argument of the Stelle-channel proposition;
  (16) rational independence of the mode energies forces m_k = n_k on every
       zero-energy configuration (paper `thm:kernel`(c), per normal mode).

  Further sections (namespaces Spinor, Conformal, FieldIdentities):
  (17) `eq:etaeps`: -eps eps X Y = 2 eta(X,Y) and det X = -eta(X,X) (unnormalized soldering);
  (18) `prop:null`: det(xi xi^dagger) = 0;
  (19) the dyad identity of App. A and `prop:tetrad` in spinor form;
  (20) the chiral trace multiplied by 16; Kramers: sigma_y conj(U) sigma_y = U;
  (21) the boundary conformal algebra) and the Clifford algebra, by exact
       kernel computation with Gaussian-integer matrices;
  (22) cleared-denominator forms of the scalaron mass, Stelle masses, de Sitter
       entropy, Starobinsky potential and R^2 coefficient.
  Polynomial identities are proved over Int.  An identity between polynomials with
  integer coefficients that holds for all integer arguments is an identity of
  polynomials, and therefore holds in every commutative ring, R and C included.

  An independent formalization of the same statements in Rocq (Coq), with a
  different encoding, is in the companion file MolecularWDW.v.

  WHAT IS NOT PROVED HERE (and why)
  ---------------------------------
  * The reduction of the site-basis constraint  H = T + V  to normal modes uses
    the spectral theorem for the 2N x 2N Hermitian one-atom matrix h and the
    isospectrality of h and its complex conjugate. That is standard linear
    algebra (Mathlib: Matrix.IsHermitian.spectral_theorem); it is checked
    numerically in the companion notebook. Lean core has no matrix library.
  * Scalars: Lean core has no real or complex numbers with proofs, so
    amplitudes and energies are taken in the integers.  Every statement below
    is a polynomial identity proved without division, so the identical proof
    goes through over any commutative ring (R or C) in a Mathlib port.

  Build:  lean MolecularWDW.lean      (tested with Lean 4.15.0, core only)
-/

namespace MolecularWDW

/-! ## 1. Configurations, vectors, finite sums -/

/-- Occupation numbers (0 or 1: the atoms are fermions) of the `K` normal modes in each
    sector: `m` for the left-atom ("/") modes, `n` for the right-atom ("\") modes. -/
structure Config (K : Nat) where
  m : Fin K → Bool
  n : Fin K → Bool

/-- A vector in the occupation basis, given by its coefficient function.  The basis vector
    for `s` is `Π_k (c_k†)^{m_k} (d_k†)^{n_k} |0⟩`, with the product ordered by mode
    (mode-interleaved Jordan-Wigner ordering). -/
abbrev Vec (K : Nat) := Config K → Int

/-- Set the occupation of mode `k` to `b`. -/
def setb {K : Nat} (f : Fin K → Bool) (k : Fin K) (b : Bool) : Fin K → Bool :=
  fun j => if j = k then b else f j

@[simp] theorem setb_self {K : Nat} (f : Fin K → Bool) (k : Fin K) (b : Bool) :
    setb f k b k = b := by simp [setb]

theorem setb_ne {K : Nat} (f : Fin K → Bool) {k j : Fin K} (b : Bool) (h : j ≠ k) :
    setb f k b j = f j := by simp [setb, h]

@[simp] theorem setb_setb {K : Nat} (f : Fin K → Bool) (k : Fin K) (b b' : Bool) :
    setb (setb f k b) k b' = setb f k b' := by
  funext j; unfold setb; by_cases h : j = k <;> simp [h]

theorem setb_eq {K : Nat} (f : Fin K → Bool) (k : Fin K) (b : Bool) (h : f k = b) :
    setb f k b = f := by
  funext j; unfold setb
  by_cases hj : j = k
  · subst hj; simp [h]
  · simp [hj]

theorem setb_comm {K : Nat} (f : Fin K → Bool) {k l : Fin K} (b b' : Bool) (h : k ≠ l) :
    setb (setb f k b) l b' = setb (setb f l b') k b := by
  funext j; unfold setb
  by_cases hl : j = l <;> by_cases hk : j = k
  · exact absurd (hk.symm.trans hl) h
  · rw [if_pos hl, if_neg hk, if_pos hl]
  · rw [if_neg hl, if_pos hk, if_pos hk]
  · rw [if_neg hl, if_neg hk, if_neg hk, if_neg hl]

/-- `0`/`1` value of an occupation. -/
def ind (b : Bool) : Int := if b then 1 else 0
@[simp] theorem ind_true : ind true = 1 := rfl
@[simp] theorem ind_false : ind false = 0 := rfl

/-- Finite sum over `Fin K`. -/
def sumFin : (K : Nat) → (Fin K → Int) → Int
  | 0, _ => 0
  | K + 1, f => sumFin K (fun i => f i.castSucc) + f (Fin.last K)

theorem sumFin_congr {K : Nat} {f g : Fin K → Int} (h : ∀ i, f i = g i) :
    sumFin K f = sumFin K g := by
  have : f = g := funext h
  rw [this]

theorem sumFin_zero {K : Nat} (f : Fin K → Int) (h : ∀ i, f i = 0) :
    sumFin K f = 0 := by
  induction K with
  | zero => rfl
  | succ K ih =>
    simp only [sumFin]
    rw [ih (fun i => f i.castSucc) (fun i => h _), h]
    rfl

theorem sumFin_mul_right {K : Nat} (f : Fin K → Int) (x : Int) :
    sumFin K (fun i => f i * x) = sumFin K f * x := by
  induction K with
  | zero => simp [sumFin]
  | succ K ih =>
    simp only [sumFin]
    rw [ih (fun i => f i.castSucc), Int.add_mul]

theorem sumFin_neg {K : Nat} (f : Fin K → Int) :
    sumFin K (fun i => - f i) = - sumFin K f := by
  induction K with
  | zero => rfl
  | succ K ih =>
    simp only [sumFin]
    rw [ih (fun i => f i.castSucc), Int.neg_add]

theorem sumFin_two_mul {K : Nat} (f : Fin K → Int) :
    sumFin K (fun i => 2 * f i) = 2 * sumFin K f := by
  induction K with
  | zero => rfl
  | succ K ih =>
    simp only [sumFin]
    rw [ih (fun i => f i.castSucc), Int.mul_add]

/-- Two functions that agree except at one point have sums differing by the jump there. -/
theorem sumFin_point {K : Nat} (f g : Fin K → Int) (k : Fin K)
    (h : ∀ j, j ≠ k → f j = g j) : sumFin K f = sumFin K g + (f k - g k) := by
  induction K with
  | zero => exact k.elim0
  | succ K ih =>
    simp only [sumFin]
    by_cases hk : k.val < K
    · have e : (⟨k.val, hk⟩ : Fin K).castSucc = k := Fin.eq_of_val_eq rfl
      have h1 := ih (fun i => f i.castSucc) (fun i => g i.castSucc) ⟨k.val, hk⟩ (by
        intro j hj
        apply h
        intro hc
        apply hj
        apply Fin.eq_of_val_eq
        have := congrArg Fin.val hc
        simpa using this)
      have h2 : f (Fin.last K) = g (Fin.last K) := h _ (by
        intro hc; have := congrArg Fin.val hc; simp at this; omega)
      rw [h1, h2]; simp only [e]; omega
    · have e : k = Fin.last K := Fin.eq_of_val_eq (by have := k.isLt; simp; omega)
      subst e
      have h1 : sumFin K (fun i => f i.castSucc) = sumFin K (fun i => g i.castSucc) :=
        sumFin_congr (fun i => h _ (by
          intro hc; have := congrArg Fin.val hc; simp at this; omega))
      rw [h1]; omega

theorem Config.ext' {K : Nat} {a b : Config K} (hm : a.m = b.m) (hn : a.n = b.n) :
    a = b := by
  cases a; cases b; simp_all

/-! ## 2. Fermionic ladder operators (Jordan-Wigner signs)

  Basis ordering: modes in increasing order, and within mode `k` first `c_k†` then `d_k†`.
  Then `c_k†` carries the sign `(-1)^(number of atoms in modes below k)` and `d_k†` the sign
  `(-1)^(atoms below k + m_k)`.
-/

variable {K : Nat}

/-- number of atoms (both species) in modes strictly below `k` -/
def below (s : Config K) (k : Fin K) : Int :=
  sumFin K (fun j => if j.val < k.val then ind (s.m j) + ind (s.n j) else 0)

/-- `(-1)^x` -/
def sgn (x : Int) : Int := if x % 2 = 0 then 1 else -1

theorem sgn_sq (x : Int) : sgn x * sgn x = 1 := by
  unfold sgn; split <;> rfl

theorem sgn_add_one (x : Int) : sgn (x + 1) = - sgn x := by
  unfold sgn; split <;> split <;> omega

theorem sgn_sub_one (x : Int) : sgn (x - 1) = - sgn x := by
  have h := sgn_add_one (x - 1)
  rw [Int.sub_add_cancel] at h
  rw [h, Int.neg_neg]

theorem sgn_sub_two (x : Int) : sgn (x - 2) = sgn x := by
  have h1 := sgn_sub_one x
  have h2 := sgn_sub_one (x - 1)
  rw [show x - 1 - 1 = x - 2 by omega, h1, Int.neg_neg] at h2
  exact h2

/-- changing an occupation at or above `k` does not change `below · k` -/
theorem below_setm_ge (s : Config K) {l k : Fin K} (b : Bool) (h : ¬ l.val < k.val) :
    below ⟨setb s.m l b, s.n⟩ k = below s k := by
  unfold below; apply sumFin_congr; intro j
  by_cases hj : j.val < k.val
  · have : j ≠ l := by intro e; subst e; exact h hj
    simp [hj, setb_ne _ _ this]
  · simp [hj]

theorem below_setn_ge (s : Config K) {l k : Fin K} (b : Bool) (h : ¬ l.val < k.val) :
    below ⟨s.m, setb s.n l b⟩ k = below s k := by
  unfold below; apply sumFin_congr; intro j
  by_cases hj : j.val < k.val
  · have : j ≠ l := by intro e; subst e; exact h hj
    simp [hj, setb_ne _ _ this]
  · simp [hj]

/-- changing an occupation below `k` shifts `below · k` by the change -/
theorem below_setm_lt (s : Config K) {l k : Fin K} (b : Bool) (h : l.val < k.val) :
    below ⟨setb s.m l b, s.n⟩ k = below s k + (ind b - ind (s.m l)) := by
  unfold below
  rw [sumFin_point _ (fun j => if j.val < k.val then ind (s.m j) + ind (s.n j) else 0) l]
  · simp [h]; omega
  · intro j hj; simp [setb_ne _ _ hj]

theorem below_setn_lt (s : Config K) {l k : Fin K} (b : Bool) (h : l.val < k.val) :
    below ⟨s.m, setb s.n l b⟩ k = below s k + (ind b - ind (s.n l)) := by
  unfold below
  rw [sumFin_point _ (fun j => if j.val < k.val then ind (s.m j) + ind (s.n j) else 0) l]
  · simp [h]; omega
  · intro j hj; simp [setb_ne _ _ hj]

/-- sign of `d_k†` -/
def dsg (s : Config K) (k : Fin K) : Int := sgn (below s k + ind (s.m k))

/-- creation of a left atom ("/") in normal mode `k` -/
def cdag (k : Fin K) (ψ : Vec K) : Vec K := fun s =>
  if s.m k then sgn (below s k) * ψ ⟨setb s.m k false, s.n⟩ else 0

/-- annihilation of a left atom in normal mode `k` -/
def c (k : Fin K) (ψ : Vec K) : Vec K := fun s =>
  if s.m k then 0 else sgn (below s k) * ψ ⟨setb s.m k true, s.n⟩

/-- creation of a right atom ("\") in mirror mode `k` -/
def ddag (k : Fin K) (ψ : Vec K) : Vec K := fun s =>
  if s.n k then dsg s k * ψ ⟨s.m, setb s.n k false⟩ else 0

/-- annihilation of a right atom in mirror mode `k` -/
def d (k : Fin K) (ψ : Vec K) : Vec K := fun s =>
  if s.n k then 0 else dsg s k * ψ ⟨s.m, setb s.n k true⟩

/-- (1a) number operator of the "/" sector -/
theorem number_c (k : Fin K) (ψ : Vec K) (s : Config K) :
    cdag k (c k ψ) s = ind (s.m k) * ψ s := by
  unfold cdag c
  cases h : s.m k
  · simp
  · simp only [if_true, setb_self, Bool.false_eq_true, if_false, below_setm_ge s false (Nat.lt_irrefl _),
      setb_setb, setb_eq s.m k true h, ind_true, Int.one_mul]
    rw [← Int.mul_assoc, sgn_sq, Int.one_mul]

theorem number_d (k : Fin K) (ψ : Vec K) (s : Config K) :
    ddag k (d k ψ) s = ind (s.n k) * ψ s := by
  unfold ddag d dsg
  cases h : s.n k
  · simp
  · simp only [if_true, setb_self, Bool.false_eq_true, if_false, below_setn_ge s false (Nat.lt_irrefl _),
      setb_setb, setb_eq s.n k true h, ind_true, Int.one_mul]
    rw [← Int.mul_assoc, sgn_sq, Int.one_mul]

theorem anti_number_c (k : Fin K) (ψ : Vec K) (s : Config K) :
    c k (cdag k ψ) s = (1 - ind (s.m k)) * ψ s := by
  unfold cdag c
  cases h : s.m k
  · simp only [Bool.false_eq_true, if_false, setb_self, if_true, below_setm_ge s true (Nat.lt_irrefl _),
      setb_setb, setb_eq s.m k false h, ind_false, Int.sub_zero, Int.one_mul]
    rw [← Int.mul_assoc, sgn_sq, Int.one_mul]
  · simp

theorem anti_number_d (k : Fin K) (ψ : Vec K) (s : Config K) :
    d k (ddag k ψ) s = (1 - ind (s.n k)) * ψ s := by
  unfold ddag d dsg
  cases h : s.n k
  · simp only [Bool.false_eq_true, if_false, setb_self, if_true, below_setn_ge s true (Nat.lt_irrefl _),
      setb_setb, setb_eq s.n k false h, ind_false, Int.sub_zero, Int.one_mul]
    rw [← Int.mul_assoc, sgn_sq, Int.one_mul]
  · simp

/-- (1b) canonical anticommutation relation `{c_k, c_k†} = 1` -/
theorem car_c (k : Fin K) (ψ : Vec K) (s : Config K) :
    c k (cdag k ψ) s + cdag k (c k ψ) s = ψ s := by
  rw [anti_number_c, number_c, Int.sub_mul, Int.one_mul]; omega

/-- (1b') `{d_k, d_k†} = 1` -/
theorem car_d (k : Fin K) (ψ : Vec K) (s : Config K) :
    d k (ddag k ψ) s + ddag k (d k ψ) s = ψ s := by
  rw [anti_number_d, number_d, Int.sub_mul, Int.one_mul]; omega

/-- (1c) Pauli principle: `(c_k†)^2 = 0`, `(d_k†)^2 = 0` -/
theorem pauli_c (k : Fin K) (ψ : Vec K) : cdag k (cdag k ψ) = fun _ => 0 := by
  funext s; unfold cdag; cases s.m k <;> simp

theorem pauli_d (k : Fin K) (ψ : Vec K) : ddag k (ddag k ψ) = fun _ => 0 := by
  funext s; unfold ddag; cases s.n k <;> simp

theorem fin_lt_or_gt {k l : Fin K} (h : k ≠ l) : k.val < l.val ∨ l.val < k.val := by
  have : k.val ≠ l.val := fun e => h (Fin.eq_of_val_eq e)
  omega

/-- (1d) atoms of different modes anticommute: `c_k† c_l† = - c_l† c_k†` -/
theorem cdag_anticomm (k l : Fin K) (h : k ≠ l) (ψ : Vec K) (s : Config K) :
    cdag k (cdag l ψ) s = - cdag l (cdag k ψ) s := by
  unfold cdag
  have hlk : l ≠ k := fun e => h e.symm
  simp only [setb_ne _ _ hlk, setb_ne _ _ h]
  cases hk : s.m k <;> cases hl : s.m l <;> simp
  rw [setb_comm s.m false false h]
  rcases fin_lt_or_gt h with hlt | hgt
  · rw [below_setm_lt s false hlt, below_setm_ge s false (Nat.not_lt.mpr (Nat.le_of_lt hlt)), hk]
    simp only [ind_false, ind_true, Int.zero_sub, ← Int.sub_eq_add_neg, sgn_sub_one]
    simp only [Int.neg_mul, Int.mul_neg, Int.neg_neg, ← Int.mul_assoc, Int.mul_comm (sgn (below s k))]
  · rw [below_setm_lt s false hgt, below_setm_ge s false (Nat.not_lt.mpr (Nat.le_of_lt hgt)), hl]
    simp only [ind_false, ind_true, Int.zero_sub, ← Int.sub_eq_add_neg, sgn_sub_one]
    simp only [Int.neg_mul, Int.mul_neg, Int.neg_neg, ← Int.mul_assoc, Int.mul_comm (sgn (below s l))]

/-- (1d') `{c_k, c_l†} = 0` for `k ≠ l` -/
theorem c_cdag_anticomm (k l : Fin K) (h : k ≠ l) (ψ : Vec K) (s : Config K) :
    c k (cdag l ψ) s = - cdag l (c k ψ) s := by
  unfold cdag c
  have hlk : l ≠ k := fun e => h e.symm
  simp only [setb_ne _ _ hlk, setb_ne _ _ h]
  cases hk : s.m k <;> cases hl : s.m l <;> simp
  rw [setb_comm s.m true false h]
  rcases fin_lt_or_gt h with hlt | hgt
  · rw [below_setm_lt s true hlt, below_setm_ge s false (Nat.not_lt.mpr (Nat.le_of_lt hlt)), hk]
    simp only [ind_false, ind_true, Int.sub_zero, sgn_add_one]
    simp only [Int.neg_mul, Int.mul_neg, Int.neg_neg, ← Int.mul_assoc, Int.mul_comm (sgn (below s k))]
  · rw [below_setm_ge s true (Nat.not_lt.mpr (Nat.le_of_lt hgt)), below_setm_lt s false hgt, hl]
    simp only [ind_false, ind_true, Int.zero_sub, ← Int.sub_eq_add_neg, sgn_sub_one]
    simp only [Int.neg_mul, Int.mul_neg, Int.neg_neg, ← Int.mul_assoc, Int.mul_comm (sgn (below s l))]

/-- (1d'') `d_k† d_l† = - d_l† d_k†` -/
theorem ddag_anticomm (k l : Fin K) (h : k ≠ l) (ψ : Vec K) (s : Config K) :
    ddag k (ddag l ψ) s = - ddag l (ddag k ψ) s := by
  unfold ddag dsg
  have hlk : l ≠ k := fun e => h e.symm
  simp only [setb_ne _ _ hlk, setb_ne _ _ h]
  cases hk : s.n k <;> cases hl : s.n l <;> simp
  rw [setb_comm s.n false false h]
  rcases fin_lt_or_gt h with hlt | hgt
  · rw [below_setn_lt s false hlt, below_setn_ge s false (Nat.not_lt.mpr (Nat.le_of_lt hlt)), hk]
    simp only [ind_false, ind_true, Int.zero_sub, ← Int.sub_eq_add_neg]
    rw [show below s l - 1 + ind (s.m l) = below s l + ind (s.m l) - 1 by omega, sgn_sub_one]
    simp only [Int.neg_mul, Int.mul_neg, Int.neg_neg, ← Int.mul_assoc,
      Int.mul_comm (sgn (below s k + ind (s.m k)))]
  · rw [below_setn_lt s false hgt, below_setn_ge s false (Nat.not_lt.mpr (Nat.le_of_lt hgt)), hl]
    simp only [ind_false, ind_true, Int.zero_sub, ← Int.sub_eq_add_neg]
    rw [show below s k - 1 + ind (s.m k) = below s k + ind (s.m k) - 1 by omega, sgn_sub_one]
    simp only [Int.neg_mul, Int.mul_neg, Int.neg_neg, ← Int.mul_assoc,
      Int.mul_comm (sgn (below s l + ind (s.m l)))]

/-- (1e) "/" and "\" atoms anticommute: `c_k† d_l† = - d_l† c_k†` (all `k, l`) -/
theorem cdag_ddag_anticomm (k l : Fin K) (ψ : Vec K) (s : Config K) :
    cdag k (ddag l ψ) s = - ddag l (cdag k ψ) s := by
  unfold cdag ddag dsg
  dsimp only
  cases hk : s.m k <;> cases hl : s.n l <;> simp
  by_cases hkl : k = l
  · subst hkl
    rw [below_setm_ge s false (Nat.lt_irrefl _), below_setn_ge s false (Nat.lt_irrefl _), hk]
    simp only [setb_self, ind_false, ind_true, Int.add_zero, sgn_add_one, Int.neg_mul, Int.neg_neg]
  · have hlk : l ≠ k := fun e => hkl e.symm
    rw [setb_ne _ _ hlk]
    rcases fin_lt_or_gt hkl with hlt | hgt
    · rw [below_setm_lt s false hlt, below_setn_ge s false (Nat.not_lt.mpr (Nat.le_of_lt hlt)), hk]
      simp only [ind_false, ind_true, Int.zero_sub, ← Int.sub_eq_add_neg]
      rw [show below s l - 1 + ind (s.m l) = below s l + ind (s.m l) - 1 by omega, sgn_sub_one]
      simp only [Int.neg_mul, Int.mul_neg, Int.neg_neg, ← Int.mul_assoc,
        Int.mul_comm (sgn (below s k))]
    · rw [below_setm_ge s false (Nat.not_lt.mpr (Nat.le_of_lt hgt)), below_setn_lt s false hgt, hl]
      simp only [ind_false, ind_true, Int.zero_sub, ← Int.sub_eq_add_neg, sgn_sub_one]
      simp only [Int.neg_mul, Int.mul_neg, Int.neg_neg, ← Int.mul_assoc,
        Int.mul_comm (sgn (below s k))]

/-! ## 3. Molecules are bosons

  The molecule operator `X_k† = c_k† d_k†` creates one "/" and one "\" atom in mode `k`.
  In the mode-interleaved ordering its two Jordan-Wigner signs cancel, so it acts on the
  occupation basis with no sign at all. -/

/-- molecule creation `X_k† = c_k† d_k†` -/
def Xdag (k : Fin K) (ψ : Vec K) : Vec K := cdag k (ddag k ψ)

/-- molecule annihilation `X_k = d_k c_k` (the adjoint of `X_k†`) -/
def X (k : Fin K) (ψ : Vec K) : Vec K := d k (c k ψ)

theorem Xdag_apply (k : Fin K) (ψ : Vec K) (s : Config K) :
    Xdag k ψ s = if s.m k then (if s.n k then ψ ⟨setb s.m k false, setb s.n k false⟩ else 0)
                 else 0 := by
  unfold Xdag cdag ddag dsg
  dsimp only
  cases hm : s.m k <;> cases hn : s.n k <;> simp
  rw [below_setm_ge s false (Nat.lt_irrefl _)]
  simp only [setb_self, ind_false, Int.add_zero, ← Int.mul_assoc, sgn_sq, Int.one_mul]

theorem X_apply (k : Fin K) (ψ : Vec K) (s : Config K) :
    X k ψ s = if s.m k then 0 else (if s.n k then 0 else ψ ⟨setb s.m k true, setb s.n k true⟩) := by
  unfold X c d dsg
  dsimp only
  cases hm : s.m k <;> cases hn : s.n k <;> simp [hm]
  rw [below_setn_ge s true (Nat.lt_irrefl _)]
  simp only [ind_false, Int.add_zero, ← Int.mul_assoc, sgn_sq, Int.one_mul]

/-- (2a) MOLECULES ARE BOSONS: molecule creation operators commute, `X_k† X_l† = X_l† X_k†`. -/
theorem molecules_commute (k l : Fin K) (ψ : Vec K) :
    Xdag k (Xdag l ψ) = Xdag l (Xdag k ψ) := by
  funext s
  by_cases h : k = l
  · subst h; rfl
  · have hlk : l ≠ k := fun e => h e.symm
    simp only [Xdag_apply, setb_ne _ _ hlk, setb_ne _ _ h]
    cases s.m k <;> cases s.n k <;> cases s.m l <;> cases s.n l <;> simp
    rw [setb_comm s.m false false h, setb_comm s.n false false h]

/-- (2b) hard core (Pauli): `(X_k†)^2 = 0`, at most one molecule per mode -/
theorem molecule_pauli (k : Fin K) (ψ : Vec K) : Xdag k (Xdag k ψ) = fun _ => 0 := by
  funext s; simp only [Xdag_apply, setb_self]
  cases s.m k <;> cases s.n k <;> simp

/-- (2c) hard-core boson algebra: `[X_k, X_k†] = 1 - n^c_k - n^d_k` -/
theorem molecule_hardcore (k : Fin K) (ψ : Vec K) (s : Config K) :
    X k (Xdag k ψ) s - Xdag k (X k ψ) s = (1 - ind (s.m k) - ind (s.n k)) * ψ s := by
  simp only [X_apply, Xdag_apply, setb_self, setb_setb]
  cases hm : s.m k <;> cases hn : s.n k <;> simp [setb_eq _ _ _ hm, setb_eq _ _ _ hn]

/-- (2d) different modes: `[X_k, X_l†] = 0` for `k ≠ l` -/
theorem molecule_cross (k l : Fin K) (h : k ≠ l) (ψ : Vec K) :
    X k (Xdag l ψ) = Xdag l (X k ψ) := by
  funext s
  have hlk : l ≠ k := fun e => h e.symm
  simp only [X_apply, Xdag_apply, setb_ne _ _ hlk, setb_ne _ _ h]
  cases s.m k <;> cases s.n k <;> cases s.m l <;> cases s.n l <;> simp
  rw [setb_comm s.m true false h, setb_comm s.n true false h]

/-- below is unchanged when a whole molecule is removed from mode `k`, up to an even shift -/
theorem sgn_below_mol (s : Config K) {k l : Fin K} (hm : s.m k = true) (hn : s.n k = true) :
    sgn (below ⟨setb s.m k false, setb s.n k false⟩ l) = sgn (below s l) := by
  by_cases hkl : k.val < l.val
  · have e1 := below_setn_lt (⟨setb s.m k false, s.n⟩ : Config K) false hkl
    have e2 := below_setm_lt s false hkl
    simp only at e1
    rw [e1, e2, hm, hn]
    simp only [ind_false, ind_true]
    rw [show below s l + (0 - 1) + (0 - 1) = below s l - 2 by omega, sgn_sub_two]
  · have e1 := below_setn_ge (⟨setb s.m k false, s.n⟩ : Config K) false hkl
    have e2 := below_setm_ge s false hkl
    simp only at e1
    rw [e1, e2]

/-- (2e) molecules commute with atoms of every other mode: `c_l† X_k† = X_k† c_l†` -/
theorem molecule_atom_comm_c (k l : Fin K) (h : l ≠ k) (ψ : Vec K) :
    cdag l (Xdag k ψ) = Xdag k (cdag l ψ) := by
  funext s
  have hkl : k ≠ l := fun e => h e.symm
  simp only [Xdag_apply, cdag, setb_ne _ _ h, setb_ne _ _ hkl]
  cases hmk : s.m k <;> cases hnk : s.n k <;> cases hml : s.m l <;> simp
  rw [sgn_below_mol s hmk hnk, setb_comm s.m false false hkl]

/-- (2e') `d_l† X_k† = X_k† d_l†` for `l ≠ k` -/
theorem molecule_atom_comm_d (k l : Fin K) (h : l ≠ k) (ψ : Vec K) :
    ddag l (Xdag k ψ) = Xdag k (ddag l ψ) := by
  funext s
  have hkl : k ≠ l := fun e => h e.symm
  simp only [Xdag_apply, ddag, dsg, setb_ne _ _ h, setb_ne _ _ hkl]
  cases hmk : s.m k <;> cases hnk : s.n k <;> cases hnl : s.n l <;> simp
  have e := sgn_below_mol s (l := l) hmk hnk
  have e2 : sgn (below ⟨setb s.m k false, setb s.n k false⟩ l + ind (s.m l))
          = sgn (below s l + ind (s.m l)) := by
    unfold sgn at e ⊢
    split at e <;> split at e <;> split <;> split <;> omega
  rw [e2, setb_comm s.n false false hkl]

/-! ## 4. The Wheeler-DeWitt operator -/

/-- `H = Σ_k ε_k (c_k† c_k − d_k† d_k)`, built from the fermionic ladder operators. -/
def H (ε : Fin K → Int) (ψ : Vec K) : Vec K := fun s =>
  sumFin K (fun k => ε k * (cdag k (c k ψ) s - ddag k (d k ψ) s))

/-- Energy imbalance of a configuration: `E(m,n) = Σ_k ε_k (m_k − n_k)`. -/
def energy (ε : Fin K → Int) (s : Config K) : Int :=
  sumFin K (fun k => ε k * (ind (s.m k) - ind (s.n k)))

/-- (3) `H` is diagonal in the occupation basis with eigenvalue `E(m,n)`. -/
theorem H_diag (ε : Fin K → Int) (ψ : Vec K) (s : Config K) :
    H ε ψ s = energy ε s * ψ s := by
  unfold H energy
  rw [← sumFin_mul_right]
  apply sumFin_congr
  intro k
  rw [number_c, number_d, ← Int.sub_mul, Int.mul_assoc]

/-- Mirror map: exchange the "/" and "\" sectors. -/
def mirror (ψ : Vec K) : Vec K := fun s => ψ ⟨s.n, s.m⟩

theorem energy_mirror (ε : Fin K → Int) (s : Config K) :
    energy ε ⟨s.n, s.m⟩ = - energy ε s := by
  unfold energy
  rw [← sumFin_neg]
  apply sumFin_congr
  intro k
  simp only
  rw [← Int.mul_neg, Int.neg_sub]

/-- (4) mirror antisymmetry of the constraint: `H M = − M H`. -/
theorem H_mirror (ε : Fin K → Int) (ψ : Vec K) (s : Config K) :
    H ε (mirror ψ) s = - (mirror (H ε ψ)) s := by
  rw [H_diag]
  unfold mirror
  rw [H_diag, energy_mirror, Int.neg_mul]
  simp

/-- removing a whole molecule does not change the energy imbalance -/
theorem energy_clear_both (ε : Fin K → Int) (s : Config K) (k : Fin K)
    (hm : s.m k = true) (hn : s.n k = true) :
    energy ε ⟨setb s.m k false, setb s.n k false⟩ = energy ε s := by
  unfold energy
  apply sumFin_congr
  intro j
  simp only
  by_cases hj : j = k
  · subst hj; simp [hm, hn]
  · simp [setb_ne _ _ hj]

/-- (5) `H` commutes with every molecule creation operator. -/
theorem H_Xdag_comm (ε : Fin K → Int) (k : Fin K) (ψ : Vec K) :
    H ε (Xdag k ψ) = Xdag k (H ε ψ) := by
  funext s
  rw [H_diag, Xdag_apply, Xdag_apply]
  cases hm : s.m k <;> cases hn : s.n k <;> simp
  rw [H_diag, energy_clear_both ε s k hm hn]

/-! ## 5. Vacuum, molecular states, condensates -/

/-- the empty configuration -/
def zeroCfg (K : Nat) : Config K := ⟨fun _ => false, fun _ => false⟩

open Classical in
/-- the Fock vacuum `|0⟩` -/
noncomputable def vac (K : Nat) : Vec K := fun s => if s = zeroCfg K then 1 else 0

theorem energy_zero (ε : Fin K → Int) : energy ε (zeroCfg K) = 0 := by
  unfold energy zeroCfg
  apply sumFin_zero
  intro k
  simp

/-- (6) the vacuum solves the constraint -/
theorem H_vac (ε : Fin K → Int) : H ε (vac K) = fun _ => 0 := by
  funext s
  rw [H_diag]
  unfold vac
  by_cases h : s = zeroCfg K
  · subst h; simp [energy_zero]
  · simp [h]

/-- `X_{k_1}† ⋯ X_{k_r}† ψ` -/
def molecules (ks : List (Fin K)) (ψ : Vec K) : Vec K :=
  ks.foldr (fun k φ => Xdag k φ) ψ

theorem Xdag_zero (k : Fin K) : Xdag k (fun _ => (0 : Int)) = fun _ => 0 := by
  funext s
  rw [Xdag_apply]
  split <;> (try split) <;> rfl

/-- (7) MAIN THEOREM: every multi-molecule state solves the Wheeler-DeWitt
    equation `H Ψ = 0`. -/
theorem wdw_molecular_solution (ε : Fin K → Int) (ks : List (Fin K)) :
    H ε (molecules ks (vac K)) = fun _ => 0 := by
  induction ks with
  | nil => exact H_vac ε
  | cons k ks ih =>
    show H ε (Xdag k (molecules ks (vac K))) = _
    rw [H_Xdag_comm, ih, Xdag_zero]

/-- modes occupied by a list of molecules -/
def occ (ks : List (Fin K)) : Fin K → Bool :=
  ks.foldr (fun k f => setb f k true) (fun _ => false)

/-- no mode is used twice -/
def fresh : List (Fin K) → Prop
  | [] => True
  | k :: ks => occ ks k = false ∧ fresh ks

/-- (8a) a molecular state with distinct modes has coefficient 1 on `|occ, occ⟩` -/
theorem molecules_at_occ (ks : List (Fin K)) (hf : fresh ks) :
    molecules ks (vac K) ⟨occ ks, occ ks⟩ = 1 := by
  induction ks with
  | nil => simp [molecules, occ, vac, zeroCfg]
  | cons k ks ih =>
    show Xdag k (molecules ks (vac K)) ⟨setb (occ ks) k true, setb (occ ks) k true⟩ = 1
    rw [Xdag_apply]
    simp only [setb_self, if_true, setb_setb, setb_eq _ _ _ hf.1]
    exact ih hf.2

/-- (8b) ... and vanishes on every other configuration: it IS the basis vector
    `|occ, occ⟩`. -/
theorem molecules_support (ks : List (Fin K)) (s : Config K)
    (h : molecules ks (vac K) s ≠ 0) : s = ⟨occ ks, occ ks⟩ := by
  induction ks generalizing s with
  | nil =>
    simp only [molecules, List.foldr] at h
    unfold vac at h
    by_cases hs : s = zeroCfg K
    · subst hs; rfl
    · simp [hs] at h
  | cons k ks ih =>
    have h' : Xdag k (molecules ks (vac K)) s ≠ 0 := h
    rw [Xdag_apply] at h'
    cases hm : s.m k <;> cases hn : s.n k <;> simp [hm, hn] at h'
    have e := ih _ h'
    have em : setb s.m k false = occ ks := congrArg Config.m e
    have en : setb s.n k false = occ ks := congrArg Config.n e
    apply Config.ext'
    · show s.m = setb (occ ks) k true
      rw [← em, setb_setb, setb_eq _ _ _ hm]
    · show s.n = setb (occ ks) k true
      rw [← en, setb_setb, setb_eq _ _ _ hn]

/-- (8c) molecular states built on different sets of modes are different basis vectors,
    hence linearly independent -/
theorem distinct_solutions (ks ks' : List (Fin K)) (h : occ ks ≠ occ ks') :
    molecules ks (vac K) ⟨occ ks', occ ks'⟩ = 0 := by
  apply Classical.byContradiction
  intro hne
  have e := molecules_support ks _ hne
  exact h (congrArg Config.m e).symm

/-- (8d) Pauli: a mode used twice gives the zero vector -/
theorem molecules_repeat (k : Fin K) (ks : List (Fin K)) (ψ : Vec K) :
    molecules (k :: k :: ks) ψ = fun _ => 0 := by
  show Xdag k (Xdag k (molecules ks ψ)) = _
  exact molecule_pauli k _

/-- BCS-type molecular condensate `Π_k (u_k + v_k X_k†) ψ` -/
def bcs (u v : Fin K → Int) : List (Fin K) → Vec K → Vec K
  | [], ψ => ψ
  | k :: ks, ψ => fun s => u k * bcs u v ks ψ s + v k * Xdag k (bcs u v ks ψ) s

/-- (9) MOLECULAR CONDENSATES SOLVE THE CONSTRAINT: every BCS-type state
    `Π_k (u_k + v_k X_k†)|0⟩` satisfies `H Ψ = 0`. -/
theorem bcs_solution (ε u v : Fin K → Int) (ks : List (Fin K)) :
    H ε (bcs u v ks (vac K)) = fun _ => 0 := by
  induction ks with
  | nil => exact H_vac ε
  | cons k ks ih =>
    funext s
    have h1 : ∀ t, energy ε t * bcs u v ks (vac K) t = 0 := by
      intro t; rw [← H_diag, ih]
    have h2 : ∀ t, energy ε t * Xdag k (bcs u v ks (vac K)) t = 0 := by
      intro t; rw [← H_diag, H_Xdag_comm, ih, Xdag_zero]
    rw [H_diag]
    show energy ε s * (u k * bcs u v ks (vac K) s + v k * Xdag k (bcs u v ks (vac K)) s) = 0
    rw [Int.mul_add, Int.mul_left_comm, h1, Int.mul_left_comm, h2]
    simp

/-- energy change when a "/" atom is added to mode `k` and a "\" atom to mode `l` -/
theorem energy_pair (ε : Fin K → Int) (s : Config K) (k l : Fin K)
    (hm : s.m k = false) (hn : s.n l = false) :
    energy ε ⟨setb s.m k true, setb s.n l true⟩ = energy ε s + ε k - ε l := by
  have step1 : energy ε ⟨setb s.m k true, s.n⟩ = energy ε s + ε k := by
    unfold energy
    rw [sumFin_point _ (fun j => ε j * (ind (s.m j) - ind (s.n j))) k]
    · simp only [setb_self, hm, ind_true, ind_false]
      simp only [Int.mul_sub, Int.mul_one, Int.mul_zero]; omega
    · intro j hj; simp [setb_ne _ _ hj]
  have step2 : energy ε ⟨setb s.m k true, setb s.n l true⟩
             = energy ε ⟨setb s.m k true, s.n⟩ - ε l := by
    unfold energy
    rw [sumFin_point _ (fun j => ε j * (ind (setb s.m k true j) - ind (s.n j))) l]
    · simp only [setb_self, hn, ind_true, ind_false]
      simp only [Int.mul_sub, Int.mul_one, Int.mul_zero]; omega
    · intro j hj; simp [setb_ne _ _ hj]
  rw [step2, step1]

/-- (10) PAIR (MOLECULAR-FIELD) SHIFT: `[H, d_l c_k] = (ε_l − ε_k) d_l c_k`.  Taking the
    expectation value in any physical state gives `(ε_l − ε_k) ⟨d_l c_k⟩ = 0`: the molecular
    order parameter pairs only modes of equal energy. -/
theorem pair_shift (ε : Fin K → Int) (k l : Fin K) (ψ : Vec K) (s : Config K) :
    H ε (d l (c k ψ)) s - d l (c k (H ε ψ)) s = (ε l - ε k) * d l (c k ψ) s := by
  rw [H_diag]
  unfold d c
  dsimp only
  cases hn : s.n l <;> cases hm : s.m k <;> simp
  rw [H_diag, energy_pair ε s k l hm hn]
  simp only [Int.mul_sub, Int.sub_mul, Int.add_mul, Int.mul_add]
  generalize energy ε s = E
  generalize dsg s l = A
  generalize sgn (below ⟨s.m, setb s.n l true⟩ k) = B
  generalize ψ ⟨setb s.m k true, setb s.n l true⟩ = P
  simp only [Int.mul_comm, Int.mul_assoc, Int.mul_left_comm]
  omega

/-! ## 6. The balanced form, zero-point energy, no-go, scale, sign argument -/

/-- (11) For any decomposition `H = T + V`,  `H Ψ = 0  ↔  T Ψ = − V Ψ`. -/
theorem balance_iff (Hop T V : Vec K → Vec K)
    (hsplit : ∀ ψ s, Hop ψ s = T ψ s + V ψ s) (Ψ : Vec K) :
    (∀ s, Hop Ψ s = 0) ↔ (∀ s, T Ψ s = - V Ψ s) := by
  constructor
  · intro h s; have := h s; rw [hsplit] at this; omega
  · intro h s; rw [hsplit, h s]; omega

/-- Antisymmetrically (Weyl) ordered fermionic constraint, times two:
    `Σ ε_k [(c† c − c c†) − (d† d − d d†)]`. -/
def H_weyl2 (ε : Fin K → Int) (ψ : Vec K) : Vec K := fun s =>
  sumFin K (fun k => ε k * ((cdag k (c k ψ) s - c k (cdag k ψ) s)
                          - (ddag k (d k ψ) s - d k (ddag k ψ) s)))

/-- (12) Zero-point cancellation: each "/" mode carries the fermionic zero-point constant
    `−ε_k/2`, its mirror "\" partner `+ε_k/2`; Weyl ordering gives exactly `2H`. -/
theorem zero_point_cancellation (ε : Fin K → Int) (ψ : Vec K) (s : Config K) :
    H_weyl2 ε ψ s = 2 * H ε ψ s := by
  unfold H_weyl2 H
  rw [← sumFin_two_mul]
  apply sumFin_congr
  intro k
  rw [anti_number_c, anti_number_d, number_c, number_d]
  simp only [Int.sub_mul, Int.one_mul]
  have h1 : ind (s.m k) * ψ s - (ψ s - ind (s.m k) * ψ s)
            - (ind (s.n k) * ψ s - (ψ s - ind (s.n k) * ψ s))
          = 2 * (ind (s.m k) * ψ s - ind (s.n k) * ψ s) := by omega
  rw [h1, ← Int.mul_assoc, Int.mul_comm (ε k) 2, Int.mul_assoc]

/-- (13) No-go lemma (expectation form). -/
theorem no_go (t v : Int) (ht : 0 ≤ t) (hv : 0 ≤ v) (hsum : t + v = 0) :
    t = 0 ∧ v = 0 := by
  omega

/-- (14) Scale freedom (paper, `prop:scale`): `ker(λH) = ker H` for `λ ≠ 0`. -/
theorem kernel_scale (Hop : Vec K → Vec K) (lam : Int) (hl : lam ≠ 0) (Ψ : Vec K) :
    (∀ s, lam * Hop Ψ s = 0) ↔ (∀ s, Hop Ψ s = 0) := by
  constructor
  · intro h s
    rcases Int.mul_eq_zero.mp (h s) with h1 | h1
    · exact absurd h1 hl
    · exact h1
  · intro h s
    rw [h s, Int.mul_zero]

/-- (15) Sign argument of the Stelle-channel proposition: with `a = -c`, `b = 4c` the
    no-tachyon conditions `3a + b > 0` and `b < 0` cannot hold together. -/
theorem stelle_channel_tachyon (c : Int) :
    ¬ (3 * (-c) + 4 * c > 0 ∧ 4 * c < 0) := by
  omega

/-- (16) Rational independence (`thm:kernel`(c), per normal mode): every zero-energy
    configuration has equal occupations in the two mirror sectors. -/
def rat_indep (ε : Fin K → Int) : Prop :=
  ∀ cf : Fin K → Int, sumFin K (fun k => cf k * ε k) = 0 → ∀ k, cf k = 0

theorem kernel_molecular (ε : Fin K → Int) (s : Config K) (RI : rat_indep ε)
    (hE : energy ε s = 0) : ∀ k, s.m k = s.n k := by
  intro k
  have h := RI (fun j => ind (s.m j) - ind (s.n j)) (by
    rw [← hE]; unfold energy; apply sumFin_congr; intro j; exact Int.mul_comm _ _) k
  cases hm : s.m k <;> cases hn : s.n k <;> simp [hm, hn] at h ⊢

/-! ## 7. Sanity checks on a concrete instance (K = 2) -/

section Examples
def ε2 : Fin 2 → Int := fun k => if k = 0 then 3 else 7

-- One molecule in each mode solves the constraint.
example : H ε2 (molecules [0, 1] (vac 2)) = fun _ => 0 :=
  wdw_molecular_solution ε2 _
-- Its energy imbalance is zero; an unpaired atom has non-zero imbalance.
example : energy ε2 ⟨occ [0, 1], occ [0, 1]⟩ = 0 := by decide
example : energy ε2 ⟨fun k => if k = 0 then true else false, fun _ => false⟩ = 3 := by decide
-- Two molecules in the same mode vanish (Pauli).
example : molecules [0, 0] (vac 2) = fun _ => 0 := molecules_repeat 0 [] _
end Examples

end MolecularWDW

/-! ## 9. Spinor kinematics (paper Sec. II, App. A); polynomial identities over Int -/
namespace Spinor
/-- Gaussian-style complex numbers with integer components.  Every statement below is a
    polynomial identity with integer coefficients; one that holds for all integer arguments
    holds as an identity of polynomials, hence in every commutative ring (R, C included). -/
structure Cx where
  re : Int
  im : Int

@[simp] def cadd (a b : Cx) : Cx := ⟨a.re + b.re, a.im + b.im⟩
@[simp] def csub (a b : Cx) : Cx := ⟨a.re - b.re, a.im - b.im⟩
@[simp] def cneg (a : Cx) : Cx := ⟨-a.re, -a.im⟩
@[simp] def cmul (a b : Cx) : Cx := ⟨a.re * b.re - a.im * b.im, a.re * b.im + a.im * b.re⟩
@[simp] def cconj (a : Cx) : Cx := ⟨a.re, -a.im⟩
@[simp] def creal (x : Int) : Cx := ⟨x, 0⟩

inductive I2 | i0 | i1
open I2

@[simp] def epsC : I2 → I2 → Cx
  | i0, i1 => ⟨1, 0⟩
  | i1, i0 => ⟨-1, 0⟩
  | _, _ => ⟨0, 0⟩

@[simp] def sum2 (f : I2 → Cx) : Cx := cadd (f i0) (f i1)

@[simp] def Xt (x0 x1 x2 x3 : Int) : I2 → I2 → Cx
  | i0, i0 => creal (x0 + x3)
  | i0, i1 => ⟨x1, -x2⟩
  | i1, i0 => ⟨x1, x2⟩
  | i1, i1 => creal (x0 - x3)

@[simp] def etaI (x0 x1 x2 x3 y0 y1 y2 y3 : Int) : Int := -(x0 * y0) + x1 * y1 + x2 * y2 + x3 * y3

theorem eta_from_pairing (x0 x1 x2 x3 y0 y1 y2 y3 : Int) :
    cneg (sum2 fun A => sum2 fun B => sum2 fun A' => sum2 fun B' =>
      cmul (cmul (epsC A B) (epsC A' B')) (cmul (Xt x0 x1 x2 x3 A A') (Xt y0 y1 y2 y3 B B')))
    = creal (etaI x0 x1 x2 x3 y0 y1 y2 y3 + etaI x0 x1 x2 x3 y0 y1 y2 y3) := by
  simp only [sum2, cadd, cneg, cmul, epsC, Xt, creal, etaI, Cx.mk.injEq]
  constructor <;>
  simp only [Int.add_mul, Int.mul_add, Int.sub_mul, Int.mul_sub, Int.neg_mul, Int.mul_neg,
    Int.mul_one, Int.one_mul, Int.mul_zero, Int.zero_mul, Int.neg_neg, Int.neg_zero,
    Int.mul_assoc, Int.mul_comm, Int.mul_left_comm] <;> omega

@[simp] def omegaC (psi phi : I2 → Cx) : Cx :=
  sum2 fun A => sum2 fun B => cmul (epsC A B) (cmul (psi A) (phi B))

@[simp] def atom (o0 o1 io0 io1 : Cx) : I2 → I2 → Cx
  | i0, i0 => o0
  | i0, i1 => o1
  | i1, i0 => io0
  | i1, i1 => io1

set_option maxHeartbeats 4000000 in
theorem tetrad_metric_spinor (o0 o1 io0 io1 : Cx) (A A' B B' : I2) :
    cneg (sum2 fun al => sum2 fun be => sum2 fun ga => sum2 fun de =>
      cmul (cmul (epsC al be) (epsC ga de))
        (cmul (cmul (atom o0 o1 io0 io1 al A) (cconj (atom o0 o1 io0 io1 ga A')))
              (cmul (atom o0 o1 io0 io1 be B) (cconj (atom o0 o1 io0 io1 de B')))))
    = cneg (cmul (cmul (omegaC (atom o0 o1 io0 io1 i0) (atom o0 o1 io0 io1 i1))
                       (cconj (omegaC (atom o0 o1 io0 io1 i0) (atom o0 o1 io0 io1 i1))))
                 (cmul (epsC A B) (epsC A' B'))) := by
  obtain ⟨a, b⟩ := o0; obtain ⟨c, d⟩ := o1; obtain ⟨e, f⟩ := io0; obtain ⟨g, h⟩ := io1
  cases A <;> cases A' <;> cases B <;> cases B' <;>
  simp only [sum2, cadd, cneg, cmul, cconj, epsC, atom, omegaC, Cx.mk.injEq] <;>
  constructor <;>
  simp only [Int.add_mul, Int.mul_add, Int.sub_mul, Int.mul_sub, Int.neg_mul, Int.mul_neg,
    Int.mul_one, Int.one_mul, Int.mul_zero, Int.zero_mul, Int.neg_neg, Int.neg_zero,
    Int.sub_neg, Int.add_zero, Int.zero_add, Int.sub_zero, Int.zero_sub,
    Int.mul_assoc, Int.mul_comm, Int.mul_left_comm] <;> omega


@[simp] def czero : Cx := ⟨0, 0⟩

theorem det_is_minus_eta (x0 x1 x2 x3 : Int) :
    csub (cmul (Xt x0 x1 x2 x3 i0 i0) (Xt x0 x1 x2 x3 i1 i1)) (cmul (Xt x0 x1 x2 x3 i0 i1) (Xt x0 x1 x2 x3 i1 i0))
    = creal (-(etaI x0 x1 x2 x3 x0 x1 x2 x3)) := by
  simp only [csub, cmul, Xt, creal, etaI, Cx.mk.injEq]
  constructor <;>
  simp only [Int.add_mul, Int.mul_add, Int.sub_mul, Int.mul_sub, Int.neg_mul, Int.mul_neg,
    Int.mul_one, Int.one_mul, Int.mul_zero, Int.zero_mul, Int.neg_neg, Int.neg_zero,
    Int.mul_assoc, Int.mul_comm, Int.mul_left_comm] <;> omega

/-- `prop:null`: a single molecule xi^A conj(xi^{A'}) has zero determinant (it is null). -/
theorem single_molecule_null (xi0 xi1 : Cx) :
    csub (cmul (cmul xi0 (cconj xi0)) (cmul xi1 (cconj xi1)))
         (cmul (cmul xi0 (cconj xi1)) (cmul xi1 (cconj xi0))) = czero := by
  obtain ⟨a, b⟩ := xi0; obtain ⟨c, d⟩ := xi1
  simp only [csub, cmul, cconj, czero, Cx.mk.injEq]
  constructor <;>
  simp only [Int.add_mul, Int.mul_add, Int.sub_mul, Int.mul_sub, Int.neg_mul, Int.mul_neg,
    Int.mul_one, Int.one_mul, Int.mul_zero, Int.zero_mul, Int.neg_neg, Int.neg_zero, Int.sub_neg,
    Int.mul_assoc, Int.mul_comm, Int.mul_left_comm] <;> omega

@[simp] def spin2 (p0 p1 : Cx) : I2 → Cx
  | i0 => p0
  | i1 => p1

/-- App. A, dyad identity: psi^A phi^B - phi^A psi^B = omega(psi, phi) eps^{AB}. -/
theorem dyad_identity (p0 p1 q0 q1 : Cx) (A B : I2) :
    csub (cmul (spin2 p0 p1 A) (spin2 q0 q1 B)) (cmul (spin2 q0 q1 A) (spin2 p0 p1 B))
    = cmul (omegaC (spin2 p0 p1) (spin2 q0 q1)) (epsC A B) := by
  obtain ⟨a, b⟩ := p0; obtain ⟨c, d⟩ := p1; obtain ⟨e, f⟩ := q0; obtain ⟨g, h⟩ := q1
  cases A <;> cases B <;>
  simp only [csub, cmul, sum2, cadd, spin2, omegaC, epsC, Cx.mk.injEq] <;>
  constructor <;>
  simp only [Int.add_mul, Int.mul_add, Int.sub_mul, Int.mul_sub, Int.neg_mul, Int.mul_neg,
    Int.mul_one, Int.one_mul, Int.mul_zero, Int.zero_mul, Int.neg_neg, Int.neg_zero, Int.sub_neg,
    Int.add_zero, Int.zero_add, Int.sub_zero, Int.zero_sub,
    Int.mul_assoc, Int.mul_comm, Int.mul_left_comm] <;> omega

@[simp] def ci : Cx := ⟨0, 1⟩
inductive I4 | e0 | e1 | e2 | e3
open I4
@[simp] def sum4 (f : I4 → Cx) : Cx := cadd (cadd (f e0) (f e1)) (cadd (f e2) (f e3))
/-- antisymmetric theta_ab from six independent components -/
@[simp] def theta (t01 t02 t03 t12 t13 t23 : Int) : I4 → I4 → Int
  | e0, e1 => t01 | e0, e2 => t02 | e0, e3 => t03 | e1, e2 => t12 | e1, e3 => t13 | e2, e3 => t23
  | e1, e0 => -t01 | e2, e0 => -t02 | e3, e0 => -t03 | e2, e1 => -t12 | e3, e1 => -t13 | e3, e2 => -t23
  | _, _ => 0
@[simp] def etaD : I4 → Int | e0 => -1 | _ => 1
@[simp] def lc3 : I4 → I4 → I4 → Int
  | e1, e2, e3 => 1 | e2, e3, e1 => 1 | e3, e1, e2 => 1
  | e1, e3, e2 => -1 | e3, e2, e1 => -1 | e2, e1, e3 => -1
  | _, _, _ => 0
@[simp] def lc4 : I4 → I4 → I4 → I4 → Int
  | e0,e1,e2,e3 => 1 | e0,e1,e3,e2 => -1 | e0,e2,e1,e3 => -1 | e0,e2,e3,e1 => 1 | e0,e3,e1,e2 => 1 | e0,e3,e2,e1 => -1
  | e1,e0,e2,e3 => -1 | e1,e0,e3,e2 => 1 | e1,e2,e0,e3 => 1 | e1,e2,e3,e0 => -1 | e1,e3,e0,e2 => -1 | e1,e3,e2,e0 => 1
  | e2,e0,e1,e3 => 1 | e2,e0,e3,e1 => -1 | e2,e1,e0,e3 => -1 | e2,e1,e3,e0 => 1 | e2,e3,e0,e1 => 1 | e2,e3,e1,e0 => -1
  | e3,e0,e1,e2 => -1 | e3,e0,e2,e1 => 1 | e3,e1,e0,e2 => 1 | e3,e1,e2,e0 => -1 | e3,e2,e0,e1 => -1 | e3,e2,e1,e0 => 1
  | _,_,_,_ => 0
@[simp] def sigma : I4 → I2 → I2 → Cx
  | e0, i0, i0 => ⟨1, 0⟩ | e0, i1, i1 => ⟨1, 0⟩
  | e1, i0, i1 => ⟨1, 0⟩ | e1, i1, i0 => ⟨1, 0⟩
  | e2, i0, i1 => ⟨0, -1⟩ | e2, i1, i0 => ⟨0, 1⟩
  | e3, i0, i0 => ⟨1, 0⟩ | e3, i1, i1 => ⟨-1, 0⟩
  | _, _, _ => ⟨0, 0⟩
@[simp] def mmul (P Q : I2 → I2 → Cx) (A B : I2) : Cx := sum2 fun C => cmul (P A C) (Q C B)
@[simp] def tr2 (P : I2 → I2 → Cx) : Cx := cadd (P i0 i0) (P i1 i1)
/-- S~ = 4 S(theta) = -2 theta_{0i} sigma_i + i eps_{ijk} theta_{ij} sigma_k  (the self-dual generator `eq:Stheta` times 4) -/
@[simp] def St (t01 t02 t03 t12 t13 t23 : Int) (A B : I2) : Cx :=
  cadd (sum4 fun i => match i with
          | e0 => ⟨0, 0⟩
          | i => cmul (creal (-(theta t01 t02 t03 t12 t13 t23 e0 i + theta t01 t02 t03 t12 t13 t23 e0 i))) (sigma i A B))
       (sum4 fun k => match k with
          | e0 => ⟨0, 0⟩
          | k => cmul (cmul ci (sum4 fun i => sum4 fun j => creal (lc3 i j k * theta t01 t02 t03 t12 t13 t23 i j))) (sigma k A B))
@[simp] def up (th : I4 → I4 → Int) (a b : I4) : Int := etaD a * etaD b * th a b

set_option maxHeartbeats 8000000 in
/-- The chiral-trace identity `eq:chiral_trace` times 16:  tr S~ S~' = -4 theta_ab theta'^ab + 2 i eps_abcd theta^ab theta'^cd -/
theorem chiral_trace_times16 (t01 t02 t03 t12 t13 t23 u01 u02 u03 u12 u13 u23 : Int) :
    let th := theta t01 t02 t03 t12 t13 t23
    let th' := theta u01 u02 u03 u12 u13 u23
    let X := (sum4 fun a => sum4 fun b => creal (th a b * up th' a b)).re
    let E := sum4 fun a => sum4 fun b => sum4 fun c => sum4 fun d => creal (lc4 a b c d * up th a b * up th' c d)
    tr2 (mmul (St t01 t02 t03 t12 t13 t23) (St u01 u02 u03 u12 u13 u23))
      = cadd (creal (-(X + X + X + X))) (cmul ci (cadd E E)) := by
  intro th th' X E
  simp only [th, th', X, E, tr2, mmul, St, sum2, sum4, cadd, cmul, creal, ci, theta, etaD, lc3, lc4, sigma, up, Cx.mk.injEq]
  constructor <;>
  simp only [Int.add_mul, Int.mul_add, Int.sub_mul, Int.mul_sub, Int.neg_mul, Int.mul_neg,
    Int.mul_one, Int.one_mul, Int.mul_zero, Int.zero_mul, Int.neg_neg, Int.neg_zero,
    Int.sub_neg, Int.add_zero, Int.zero_add, Int.sub_zero, Int.zero_sub, Int.neg_add,
    Int.mul_assoc, Int.mul_comm, Int.mul_left_comm] <;> omega
@[simp] def su2m (a b : Cx) : I2 → I2 → Cx
  | i0, i0 => a
  | i0, i1 => b
  | i1, i0 => ⟨-b.re, b.im⟩
  | i1, i1 => ⟨a.re, -a.im⟩

/-- `prop:kramers`: for U = [[a, b], [-conj b, conj a]] (SU(2) form), sigma_y conj(U) sigma_y = U, the identity
    behind the Kramers degeneracy and the mirror isospectrality hbar = Theta^-1 h Theta. -/
theorem su2_pseudoreal (a b : Cx) (A B : I2) :
    (sum2 fun C => sum2 fun D => cmul (cmul (sigma e2 A C) (cconj (su2m a b C D))) (sigma e2 D B)) = su2m a b A B := by
  obtain ⟨p, q⟩ := a; obtain ⟨r, s⟩ := b
  cases A <;> cases B <;>
  simp only [sum2, cadd, cmul, cconj, sigma, su2m, Cx.mk.injEq] <;>
  constructor <;>
  simp only [Int.add_mul, Int.mul_add, Int.sub_mul, Int.mul_sub, Int.neg_mul, Int.mul_neg,
    Int.mul_one, Int.one_mul, Int.mul_zero, Int.zero_mul, Int.neg_neg, Int.neg_zero, Int.sub_neg,
    Int.add_zero, Int.zero_add, Int.sub_zero, Int.zero_sub] <;> omega
end Spinor

/-! ## 10. Boundary conformal algebra (paper `prop:conformal`), by exact computation -/
namespace Conformal
structure GI where
  re : Int
  im : Int
def gadd (a b : GI) : GI := ⟨a.re + b.re, a.im + b.im⟩
def gsub (a b : GI) : GI := ⟨a.re - b.re, a.im - b.im⟩
def gmul (a b : GI) : GI := ⟨a.re * b.re - a.im * b.im, a.re * b.im + a.im * b.re⟩
def gscale (z : Int) (a : GI) : GI := ⟨z * a.re, z * a.im⟩
def gconj (a : GI) : GI := ⟨a.re, -a.im⟩
def gz : GI := ⟨0, 0⟩
def g1 : GI := ⟨1, 0⟩
def geqb (a b : GI) : Bool := a.re == b.re && a.im == b.im
inductive I4 | e0 | e1 | e2 | e3
open I4
def idx : List I4 := [e0, e1, e2, e3]
abbrev M := I4 → I4 → GI
def msum (f : I4 → GI) : GI := idx.foldr (fun k acc => gadd (f k) acc) gz
def mmul (A B : M) : M := fun i j => msum (fun k => gmul (A i k) (B k j))
def madd (A B : M) : M := fun i j => gadd (A i j) (B i j)
def msub (A B : M) : M := fun i j => gsub (A i j) (B i j)
def mscale (z : Int) (A : M) : M := fun i j => gscale z (A i j)
def comm (A B : M) : M := msub (mmul A B) (mmul B A)
def mzero : M := fun _ _ => gz
def mid : M := fun i j => match i, j with
  | e0, e0 => g1 | e1, e1 => g1 | e2, e2 => g1 | e3, e3 => g1 | _, _ => gz
def meqb (A B : M) : Bool := idx.all fun i => idx.all fun j => geqb (A i j) (B i j)
def adj (A : M) : M := fun i j => gconj (A j i)
/-- Pauli blocks: sigma^a = (1, sx, sy, sz), sigmabar^a = (-1, sx, sy, sz) -/
def sig : I4 → Bool → Bool → GI
  | e0, false, false => g1 | e0, true, true => g1
  | e1, false, true => g1 | e1, true, false => g1
  | e2, false, true => ⟨0, -1⟩ | e2, true, false => ⟨0, 1⟩
  | e3, false, false => g1 | e3, true, true => ⟨-1, 0⟩
  | _, _, _ => gz
def sigbar (a : I4) (r c : Bool) : GI := match a with
  | e0 => gscale (-1) (sig e0 r c) | _ => sig a r c
/-- rows/columns e0,e1: upper block (dual of right atoms); e2,e3: lower block (left atoms) -/
def blk : I4 → Bool | e0 => false | e1 => false | e2 => true | e3 => true
def sub : I4 → Bool | e0 => false | e1 => true | e2 => false | e3 => true
def gam (a : I4) : M := fun i j =>
  match blk i, blk j with
  | false, true => sig a (sub i) (sub j)
  | true, false => sigbar a (sub i) (sub j)
  | _, _ => gz
def g5 : M := fun i j => match i, j with
  | e0, e0 => g1 | e1, e1 => g1 | e2, e2 => ⟨-1, 0⟩ | e3, e3 => ⟨-1, 0⟩ | _, _ => gz
def etaZ : I4 → Int | e0 => -1 | _ => 1
def gamlow (a : I4) : M := mscale (etaZ a) (gam a)
/-- generators times 2: Jt_ab = gamma^a gamma^b (a ≠ b), Pt_a = gamma_a gamma_5 -/
def Jt (a b : I4) : M := mmul (gam a) (gam b)
def Pt (a : I4) : M := mmul (gamlow a) g5
def Dt : M := Pt e0
def Pc (i : I4) : M := madd (Jt e0 i) (Pt i)
def Kc (i : I4) : M := msub (Jt e0 i) (Pt i)
def delta : I4 → I4 → Int | e1, e1 => 1 | e2, e2 => 1 | e3, e3 => 1 | _, _ => 0
def Mt (i j : I4) : M := match i, j with
  | e1, e1 => mzero | e2, e2 => mzero | e3, e3 => mzero | _, _ => Jt i j
def spat : List I4 := [e1, e2, e3]
def etaDiag : I4 → I4 → Int | e0, e0 => -2 | e1, e1 => 2 | e2, e2 => 2 | e3, e3 => 2 | _, _ => 0
def clifford_ok : Bool := idx.all fun a => idx.all fun b =>
  meqb (madd (mmul (gam a) (gam b)) (mmul (gam b) (gam a))) (mscale (etaDiag a b) mid)
def conformal_ok : Bool := spat.all fun i =>
  meqb (comm Dt (Pc i)) (mscale 2 (Pc i)) && meqb (comm Dt (Kc i)) (mscale (-2) (Kc i)) &&
  spat.all fun j =>
    meqb (comm (Kc i) (Pc j)) (mscale 4 (msub (Mt i j) (mscale (delta i j) Dt))) &&
    meqb (comm Dt (Mt i j)) mzero && meqb (comm (Pc i) (Pc j)) mzero && meqb (comm (Kc i) (Kc j)) mzero &&
    spat.all fun k =>
      meqb (comm (Mt i j) (Pc k)) (mscale 2 (msub (mscale (delta j k) (Pc i)) (mscale (delta i k) (Pc j)))) &&
      meqb (comm (Mt i j) (Kc k)) (mscale 2 (msub (mscale (delta j k) (Kc i)) (mscale (delta i k) (Kc j))))
def blockdiag (A : M) : Bool := idx.all fun i => idx.all fun j => if blk i == blk j then true else geqb (A i j) gz
def offdiag (A : M) : Bool := idx.all fun i => idx.all fun j => if blk i == blk j then geqb (A i j) gz else true
/-- M_ij block diagonal and anti-Hermitian (same SU(2) on / and \); D Hermitian, off diagonal, Dt^2 = 1
    (so D = Dt/2 has eigenvalues +-1/2). -/
def structure_ok : Bool :=
  (spat.all fun i => spat.all fun j => blockdiag (Mt i j) && meqb (adj (Mt i j)) (mscale (-1) (Mt i j)))
  && meqb (adj Dt) Dt && offdiag Dt && meqb (mmul Dt Dt) mid
theorem clifford_algebra : clifford_ok = true := by decide
/-- `prop:conformal` with generators doubled (Dt = 2D, Pc = 2P, Kc = 2K, Mt = 2M):
    [Dt,Pc_i] = 2Pc_i, [Dt,Kc_i] = -2Kc_i, [Kc_i,Pc_j] = 4(Mt_ij - delta_ij Dt), [Dt,Mt_ij] = 0,
    [Pc_i,Pc_j] = [Kc_i,Kc_j] = 0, [Mt_ij,X_k] = 2(delta_jk X_i - delta_ik X_j). -/
theorem boundary_conformal_algebra : conformal_ok = true := by decide
theorem conformal_generator_structure : structure_ok = true := by decide
end Conformal

/-! ## 11. Cleared-denominator forms of the field-theory formulas -/
namespace FieldIdentities
/-- Scalaron mass `eq:scalaron_mass`, numerator: f' - Rbar f'' = alpha for f = alpha R + beta R^2, so
    m_f^2 = (f'/f'' - Rbar)/3 = alpha/(3 f'') = alpha/(6 beta). -/
theorem scalaron_numerator (α β R : Int) : (α + 2 * β * R) - R * (2 * β) = α := by
  simp only [Int.mul_comm, Int.mul_assoc, Int.mul_left_comm]; omega
/-- de Sitter entropy `eq:entropy`, cross-multiplied:  3/(64 π G Λ) = S_dS/(64 π^2) with S_dS = 3π/(GΛ). -/
theorem desitter_entropy_cleared (p G L : Int) :
    3 * (64 * (p * p)) * (G * L) = (3 * p) * (64 * p * G * L) := by
  simp only [Int.mul_comm, Int.mul_assoc, Int.mul_left_comm]
/-- R^2 coefficient `eq:R2coef`, cross-multiplied: with A_s (24 π^2 M_P^2) = N^2 M^2, M_P^2 (288 π^2 A_s) = 12 M^2 N^2. -/
theorem R2_coefficient_cleared (N M Mp p As : Int) (hAs : 24 * (p * p) * (Mp * Mp) * As = N * N * (M * M)) :
    Mp * Mp * (288 * (p * p) * As) = 12 * (M * M) * (N * N) := by
  have h12 : Mp * Mp * (288 * (p * p) * As) = 12 * (24 * (p * p) * (Mp * Mp) * As) := by
    rw [show (288 : Int) = 12 * 24 from rfl]; simp only [Int.mul_comm, Int.mul_assoc, Int.mul_left_comm]
  rw [h12, hAs]; simp only [Int.mul_comm, Int.mul_assoc, Int.mul_left_comm]
/-- Starobinsky potential (`sec:inflation`), cleared: with R = 3 M^2 (F - 1), i.e. F = f'(R) for f = R + R^2/(6 M^2),
    6 M^2 (R F - f) = R^2, so (M_P^2/2)(R F - f)/F^2 = (3/4) M^2 M_P^2 (1 - 1/F)^2. -/
theorem starobinsky_cleared (M F R : Int) (hR : 3 * (M * M) * (F - 1) = R) :
    6 * (M * M) * (R * F) - (6 * (M * M) * R + R * R) = R * R := by
  have key : 6 * (M * M) * (R * F) - (6 * (M * M) * R + R * R) - R * R
           = R * (3 * (M * M) * (F - 1) - R) + R * (3 * (M * M) * (F - 1) - R) := by
    rw [show (6 : Int) = 3 + 3 from rfl]
    simp only [Int.mul_sub, Int.sub_mul, Int.mul_add, Int.add_mul, Int.mul_one, Int.one_mul, Int.mul_comm, Int.mul_assoc, Int.mul_left_comm]
    omega
  rw [hR, Int.sub_self, Int.mul_zero, Int.add_zero] at key; omega
/-- Stelle masses `eq:stelle_masses`: Ricci basis a = -c, b = 4c and Weyl basis beta_R = c/3, alpha_W = 2c give the same
    masses: 2(3a+b) = 6 beta_R = 2c (scalar) and b = 2 alpha_W = 4c (spin 2). -/
theorem stelle_bases_agree (c bR aW : Int) (h1 : 3 * bR = c) (h2 : aW = 2 * c) :
    2 * (3 * (-c) + 4 * c) = 2 * c ∧ 6 * bR = 2 * c ∧ 4 * c = 2 * aW := by omega
theorem scalaron_numerator_FL (xi Phi2 β R : Int) : ((1 - xi * Phi2) + 2 * β * R) - R * (2 * β) = 1 - xi * Phi2 := by
  simp only [Int.mul_comm, Int.mul_assoc, Int.mul_left_comm]; omega
end FieldIdentities

/-! ## 12. Audit: no `sorry`, only Lean's standard axioms -/
#print axioms MolecularWDW.car_c
#print axioms MolecularWDW.car_d
#print axioms MolecularWDW.pauli_c
#print axioms MolecularWDW.cdag_anticomm
#print axioms MolecularWDW.c_cdag_anticomm
#print axioms MolecularWDW.ddag_anticomm
#print axioms MolecularWDW.cdag_ddag_anticomm
#print axioms MolecularWDW.molecules_commute
#print axioms MolecularWDW.molecule_pauli
#print axioms MolecularWDW.molecule_hardcore
#print axioms MolecularWDW.molecule_cross
#print axioms MolecularWDW.molecule_atom_comm_c
#print axioms MolecularWDW.molecule_atom_comm_d
#print axioms MolecularWDW.H_diag
#print axioms MolecularWDW.H_mirror
#print axioms MolecularWDW.H_Xdag_comm
#print axioms MolecularWDW.wdw_molecular_solution
#print axioms MolecularWDW.molecules_at_occ
#print axioms MolecularWDW.molecules_support
#print axioms MolecularWDW.distinct_solutions
#print axioms MolecularWDW.molecules_repeat
#print axioms MolecularWDW.bcs_solution
#print axioms MolecularWDW.pair_shift
#print axioms MolecularWDW.balance_iff
#print axioms MolecularWDW.zero_point_cancellation
#print axioms MolecularWDW.no_go
#print axioms MolecularWDW.kernel_scale
#print axioms MolecularWDW.stelle_channel_tachyon
#print axioms MolecularWDW.kernel_molecular
#print axioms Spinor.eta_from_pairing
#print axioms Spinor.det_is_minus_eta
#print axioms Spinor.single_molecule_null
#print axioms Spinor.dyad_identity
#print axioms Spinor.tetrad_metric_spinor
#print axioms Spinor.chiral_trace_times16
#print axioms Spinor.su2_pseudoreal
#print axioms Conformal.clifford_algebra
#print axioms Conformal.boundary_conformal_algebra
#print axioms Conformal.conformal_generator_structure
#print axioms FieldIdentities.scalaron_numerator
#print axioms FieldIdentities.scalaron_numerator_FL
#print axioms FieldIdentities.stelle_bases_agree
#print axioms FieldIdentities.desitter_entropy_cleared
#print axioms FieldIdentities.R2_coefficient_cleared
#print axioms FieldIdentities.starobinsky_cleared
'''
if not os.path.exists('MolecularWDW.lean'):
    open('MolecularWDW.lean', 'w').write(LEAN_SRC)
same = hashlib.sha256(open('MolecularWDW.lean').read().encode()).hexdigest() == hashlib.sha256(LEAN_SRC.encode()).hexdigest()
MAIN = ["MolecularWDW." + n_ for n_ in ["car_c", "car_d", "pauli_c", "cdag_anticomm", "c_cdag_anticomm", "ddag_anticomm",
        "cdag_ddag_anticomm", "molecules_commute", "molecule_pauli", "molecule_hardcore", "molecule_cross",
        "molecule_atom_comm_c", "molecule_atom_comm_d", "H_diag", "H_mirror", "H_Xdag_comm", "wdw_molecular_solution",
        "molecules_at_occ", "molecules_support", "distinct_solutions", "molecules_repeat", "bcs_solution", "pair_shift",
        "balance_iff", "zero_point_cancellation", "no_go", "kernel_scale", "stelle_channel_tachyon", "kernel_molecular"]] + \
       ["Spinor." + n_ for n_ in ["eta_from_pairing", "det_is_minus_eta", "single_molecule_null", "dyad_identity",
        "tetrad_metric_spinor", "chiral_trace_times16", "su2_pseudoreal"]] + \
       ["Conformal." + n_ for n_ in ["clifford_algebra", "boundary_conformal_algebra", "conformal_generator_structure"]] + \
       ["FieldIdentities." + n_ for n_ in ["scalaron_numerator", "scalaron_numerator_FL", "stelle_bases_agree",
        "desitter_entropy_cleared", "R2_coefficient_cleared", "starobinsky_cleared"]]
src_ok = same and "sorry" not in LEAN_SRC.replace("no `sorry`", "") and all(f"theorem {m_.split('.')[1]}" in LEAN_SRC for m_ in MAIN)
check("Lean source: identical to ancillary file, no sorry, all main theorems present", "L", src_ok)
lean = shutil.which('lean') or os.environ.get('LEAN_BIN')
if lean:
    out = subprocess.run([lean, 'MolecularWDW.lean'], capture_output=True, text=True)
    txt = out.stdout + out.stderr
    print(txt[-1500:])
    allowed = {"propext", "Classical.choice", "Quot.sound"}
    check("Lean: file compiles without errors or sorry", "L", out.returncode == 0 and "error" not in txt and "sorry" not in txt)
    for m_ in MAIN:
        mm = re.search(rf"'{re.escape(m_)}' depends on axioms: \[([^\]]*)\]", txt)
        none_ = re.search(rf"'{re.escape(m_)}' does not depend on any axioms", txt)
        axs = set() if none_ else (set(a_.strip() for a_ in mm.group(1).split(",")) if mm else None)
        check(f"Lean theorem {m_}: standard axioms only", "L", axs is not None and axs <= allowed, (", ".join(sorted(axs)) or "no axioms") if axs is not None else "not found")
else:
    check("Lean: compilation", "L", None, "no Lean toolchain found")
```

**Independent formal verification [R].** The cell writes the Rocq source (identical to the ancillary `MolecularWDW.v`) if it is not already present and compiles it when `coqc` (Coq ≥ 8.18, or the Rocq Prover) is available; set `COQC_BIN` otherwise. The file must compile; every theorem about the exact solution, the spinor identities, and the conformal algebra must be closed under the global context (no axioms), and the real-number identities may use only the two axioms of Rocq's real-number library. Without a toolchain the compile checks are reported as skipped; a static scan of the source still runs.

```python
ROCQ_SRC = r'''(* ===========================================================================
   MolecularWDW.v
   Independent Rocq (Coq >= 8.18) formalization accompanying the paper

     "Spacetime Molecules: A Pre-geometric Wheeler-DeWitt Model
      Built from Fermionic Chiral Spinor Atoms"

   This file was written separately from the companion Lean 4 file
   MolecularWDW.lean and uses a different encoding (occupations are
   finite lists of booleans, not Fin-indexed functions; Jordan-Wigner signs from
   prefix counts, not Fin sums; polynomial identities are proved
   over an arbitrary commutative ring, not over the integers).  Only the
   informal statements are shared.

   CONTENTS (paper labels in brackets)
   Part A  Exact solution of the constraint with FERMIONIC atoms (Secs. III-IV)
     A1  fermionic ladder operators on occupation lists (mode-interleaved
         Jordan-Wigner signs): {c_k,c_k^dag} = {d_k,d_k^dag} = 1, {c_k,c_j^dag} = 0,
         c^dag c^dag, d^dag d^dag and c^dag d^dag anticommute, Pauli (c_k^dag)^2 = 0
     A2  MOLECULES ARE BOSONS: X_k^dag = c_k^dag d_k^dag commute with each other and with
         the atoms of other modes; hard-core algebra [X_k, X_k^dag] = 1 - n^c_k - n^d_k,
         [X_k, X_j^dag] = 0 (k <> j), (X_k^dag)^2 = 0               [prop:hardcore]
     A3  H = sum_k eps_k (c_k^dag c_k - d_k^dag d_k) is diagonal, eigenvalue E(m,n);
         mirror antisymmetry H M = - M H; [H, X_k^dag] = 0
     A4  H|0> = 0; every multi-molecule state solves H Psi = 0; molecular states on
         different sets of modes are distinct basis vectors; a repeated mode gives
         zero; every BCS condensate prod (u + v X^dag)|0> solves H Psi = 0;
         pair shift [H, d_j c_k] = (eps_j - eps_k) d_j c_k  [thm:kernel, prop:condensate,
         prop:fieldeq]
     A5  H Psi = 0 <-> T Psi = - V Psi; fermionic zero-point cancellation [prop:zp];
         no-go [thm:nogo]; scale freedom [prop:scale]; Stelle sign [prop:stelle];
         rational independence forces m_k = n_k [thm:kernel(c)]
   Part B  Spinor kinematics, over an arbitrary commutative ring (Sec. II, App. A)
     B1  [eq:etaeps]: -eps eps X Y = 2 eta(X,Y) and det X = -eta(X,X) (unnormalized soldering)
     B2  [prop:null]: det(xi xi^dagger) = 0
     B3  dyad identity (App. A) and [prop:tetrad] in spinor form
     B4  [eq:chiral_trace], the chiral trace, multiplied by 16
     B5  [prop:kramers]: sigma_y conj(U) sigma_y = U for U of SU(2) form
   Part C  Boundary conformal algebra [prop:conformal], by exact computation with
           Gaussian-integer matrices; Clifford algebra of the Weyl-basis gammas
   Part D  Field identities over the real numbers (scalaron mass, Stelle masses,
           de Sitter entropy, Starobinsky potential, R^2 coefficient)

   Normalizations: the paper's soldering carries a factor 1/sqrt 2; Part B uses
   X~ = X^0 1 + X^i sigma_i, which multiplies [eq:etaeps] by 2.  Part C uses the
   generators multiplied by 2 so that all entries are Gaussian integers.

   Build:  coqc MolecularWDW.v      (tested with Coq 8.18.0, standard library only)
   The final section prints the assumptions of every main theorem.
   =========================================================================== *)

From Coq Require Import ZArith Lia List Arith Bool Ring Field.
From Coq Require Import Reals Lra.
Import ListNotations.

(* ===================================================================== *)
(** * Part A. Exact solution of the constraint                           *)
(* ===================================================================== *)

Module FockSpace.
Open Scope nat_scope.

(** Occupation lists of booleans (fermions: each mode is empty or singly occupied).
    Mode k of a list l has occupation [get l k]. *)
Fixpoint upd (l : list bool) (k : nat) (b : bool) : list bool :=
  match l, k with
  | [], _ => []
  | x :: t, O => b :: t
  | x :: t, S k' => x :: upd t k' b
  end.

Definition get (l : list bool) (k : nat) : bool := nth k l false.
Definition b2n (b : bool) : nat := if b then 1 else 0.
Fixpoint cnt (l : list bool) : nat :=
  match l with [] => 0 | x :: t => b2n x + cnt t end.

Lemma length_upd : forall l k b, length (upd l k b) = length l.
Proof. induction l as [|x t IH]; intros [|k] b; simpl; auto. Qed.

Lemma get_upd_same : forall l k b, k < length l -> get (upd l k b) k = b.
Proof.
  unfold get; induction l as [|x t IH]; intros k b H; simpl in *; [lia|].
  destruct k as [|k]; simpl; [reflexivity|]. apply IH; lia.
Qed.

Lemma get_upd_other : forall l k j b, j <> k -> get (upd l k b) j = get l j.
Proof.
  unfold get; induction l as [|x t IH]; intros k j b H; simpl; [reflexivity|].
  destruct k as [|k], j as [|j]; simpl; try reflexivity; try lia.
  apply IH; lia.
Qed.

Lemma get_true_lt : forall l k, get l k = true -> k < length l.
Proof.
  unfold get; induction l as [|x t IH]; intros k H; simpl in *.
  - destruct k; discriminate.
  - destruct k as [|k]; [lia|]. apply IH in H. lia.
Qed.

Lemma upd_upd : forall l k b b', upd (upd l k b) k b' = upd l k b'.
Proof. induction l as [|x t IH]; intros [|k] b b'; simpl; try rewrite IH; reflexivity. Qed.

Lemma upd_get : forall l k, upd l k (get l k) = l.
Proof. unfold get; induction l as [|x t IH]; intros [|k]; simpl; try rewrite IH; reflexivity. Qed.

Lemma upd_eq : forall l k b, get l k = b -> upd l k b = l.
Proof. intros l k b <-. apply upd_get. Qed.

Lemma upd_comm : forall l k j b b', k <> j -> upd (upd l k b) j b' = upd (upd l j b') k b.
Proof.
  induction l as [|x t IH]; intros k j b b' H; simpl; [reflexivity|].
  destruct k as [|k], j as [|j]; simpl; try reflexivity; try lia.
  rewrite IH by lia. reflexivity.
Qed.

Lemma firstn_upd : forall l k j b, firstn k (upd l j b) = upd (firstn k l) j b.
Proof.
  induction l as [|x t IH]; intros k j b.
  - destruct k; simpl; reflexivity.
  - destruct k as [|k]; simpl; [reflexivity|].
    destruct j as [|j]; simpl; [reflexivity|]. rewrite IH. reflexivity.
Qed.

Lemma upd_oob : forall l j b, length l <= j -> upd l j b = l.
Proof.
  induction l as [|x t IH]; intros j b H; simpl in *; [reflexivity|].
  destruct j as [|j]; [lia|]. rewrite IH by lia. reflexivity.
Qed.

Lemma cnt_upd : forall l j b, j < length l -> cnt (upd l j b) + b2n (get l j) = cnt l + b2n b.
Proof.
  unfold get; induction l as [|x t IH]; intros j b H; simpl in *; [lia|].
  destruct j as [|j]; simpl.
  - lia.
  - specialize (IH j b ltac:(lia)). lia.
Qed.

Lemma get_firstn : forall l k j, j < k -> get (firstn k l) j = get l j.
Proof.
  unfold get; induction l as [|x t IH]; intros k j H.
  - destruct k, j; reflexivity.
  - destruct k as [|k]; [lia|]. destruct j as [|j]; simpl; [reflexivity|]. apply IH; lia.
Qed.

Lemma cnt_firstn_upd_ge : forall l k j b, k <= j -> cnt (firstn k (upd l j b)) = cnt (firstn k l).
Proof.
  intros l k j b H. rewrite firstn_upd, upd_oob; [reflexivity|]. rewrite firstn_length. lia.
Qed.

Lemma cnt_firstn_upd_lt : forall l k j b, j < k -> j < length l ->
  cnt (firstn k (upd l j b)) + b2n (get l j) = cnt (firstn k l) + b2n b.
Proof.
  intros l k j b Hjk Hj. rewrite firstn_upd. rewrite <- (get_firstn l k j Hjk).
  apply cnt_upd. rewrite firstn_length. lia.
Qed.

Lemma get_repeat_false : forall K k, get (repeat false K) k = false.
Proof. unfold get; induction K as [|K IH]; intros [|k]; simpl; auto. Qed.

Close Scope nat_scope.
Open Scope Z_scope.

Definition Config : Type := (list bool * list bool)%type.
Definition Vec : Type := Config -> Z.
Definition b2z (b : bool) : Z := if b then 1 else 0.

(** Basis ordering: modes in increasing order, within mode k first c_k^dag then d_k^dag
    (mode-interleaved Jordan-Wigner).  [below s k] counts the atoms in modes below k. *)
Definition below (s : Config) (k : nat) : nat :=
  (cnt (firstn k (fst s)) + cnt (firstn k (snd s)))%nat.

Definition sgn (x : nat) : Z := if Nat.even x then 1 else -1.

Lemma sgn_S : forall x, sgn (S x) = - sgn x.
Proof. intros x. unfold sgn. rewrite Nat.even_succ, <- Nat.negb_even. destruct (Nat.even x); reflexivity. Qed.

Lemma sgn_sq : forall x, sgn x * sgn x = 1.
Proof. intros x. unfold sgn. destruct (Nat.even x); reflexivity. Qed.

Lemma sgn_SS : forall x, sgn (S (S x)) = sgn x.
Proof. intros x. rewrite !sgn_S. ring. Qed.

Lemma below_m_ge : forall m n k j b, (k <= j)%nat -> below (upd m j b, n) k = below (m, n) k.
Proof. intros. unfold below; cbn [fst snd]. rewrite cnt_firstn_upd_ge by assumption. reflexivity. Qed.

Lemma below_n_ge : forall m n k j b, (k <= j)%nat -> below (m, upd n j b) k = below (m, n) k.
Proof. intros. unfold below; cbn [fst snd]. rewrite cnt_firstn_upd_ge by assumption. reflexivity. Qed.

Lemma below_m_lt : forall m n k j b, (j < k)%nat -> (j < length m)%nat ->
  (below (upd m j b, n) k + b2n (get m j) = below (m, n) k + b2n b)%nat.
Proof. intros m n k j b H1 H2. unfold below; cbn [fst snd]. pose proof (cnt_firstn_upd_lt m k j b H1 H2). lia. Qed.

Lemma below_n_lt : forall m n k j b, (j < k)%nat -> (j < length n)%nat ->
  (below (m, upd n j b) k + b2n (get n j) = below (m, n) k + b2n b)%nat.
Proof. intros m n k j b H1 H2. unfold below; cbn [fst snd]. pose proof (cnt_firstn_upd_lt n k j b H1 H2). lia. Qed.

Definition cdag (k : nat) (psi : Vec) : Vec := fun s =>
  if get (fst s) k then sgn (below s k) * psi (upd (fst s) k false, snd s) else 0.
Definition c (k : nat) (psi : Vec) : Vec := fun s =>
  if get (fst s) k then 0 else sgn (below s k) * psi (upd (fst s) k true, snd s).
Definition dsg (s : Config) (k : nat) : Z := sgn (below s k + b2n (get (fst s) k)).
Definition ddag (k : nat) (psi : Vec) : Vec := fun s =>
  if get (snd s) k then dsg s k * psi (fst s, upd (snd s) k false) else 0.
Definition d (k : nat) (psi : Vec) : Vec := fun s =>
  if get (snd s) k then 0 else dsg s k * psi (fst s, upd (snd s) k true).

(** A1: canonical anticommutation relations and the Pauli principle *)
Lemma number_c : forall k psi s, cdag k (c k psi) s = b2z (get (fst s) k) * psi s.
Proof.
  intros k psi [m n]; unfold cdag, c; cbn [fst snd].
  destruct (get m k) eqn:E; [|unfold b2z; ring].
  pose proof (get_true_lt m k E) as Hk.
  rewrite get_upd_same by exact Hk. rewrite below_m_ge by lia.
  rewrite upd_upd, upd_eq by exact E.
  unfold b2z. rewrite Z.mul_assoc, sgn_sq. ring.
Qed.

Lemma number_d : forall k psi s, ddag k (d k psi) s = b2z (get (snd s) k) * psi s.
Proof.
  intros k psi [m n]; unfold ddag, d, dsg; cbn [fst snd].
  destruct (get n k) eqn:E; [|unfold b2z; ring].
  pose proof (get_true_lt n k E) as Hk.
  rewrite get_upd_same by exact Hk. rewrite below_n_ge by lia.
  rewrite upd_upd, upd_eq by exact E.
  unfold b2z. rewrite Z.mul_assoc, sgn_sq. ring.
Qed.

Lemma anti_number_c : forall k psi s, (k < length (fst s))%nat ->
  c k (cdag k psi) s = (1 - b2z (get (fst s) k)) * psi s.
Proof.
  intros k psi [m n] H; simpl in H; unfold cdag, c; cbn [fst snd].
  destruct (get m k) eqn:E; [unfold b2z; ring|].
  rewrite get_upd_same by exact H. rewrite below_m_ge by lia.
  rewrite upd_upd, upd_eq by exact E.
  unfold b2z. rewrite Z.mul_assoc, sgn_sq. ring.
Qed.

Lemma anti_number_d : forall k psi s, (k < length (snd s))%nat ->
  d k (ddag k psi) s = (1 - b2z (get (snd s) k)) * psi s.
Proof.
  intros k psi [m n] H; simpl in H; unfold ddag, d, dsg; cbn [fst snd].
  destruct (get n k) eqn:E; [unfold b2z; ring|].
  rewrite get_upd_same by exact H. rewrite below_n_ge by lia.
  rewrite upd_upd, upd_eq by exact E.
  unfold b2z. rewrite Z.mul_assoc, sgn_sq. ring.
Qed.

Theorem car_c : forall k psi s, (k < length (fst s))%nat ->
  c k (cdag k psi) s + cdag k (c k psi) s = psi s.
Proof. intros. rewrite anti_number_c, number_c by assumption. ring. Qed.

Theorem car_d : forall k psi s, (k < length (snd s))%nat ->
  d k (ddag k psi) s + ddag k (d k psi) s = psi s.
Proof. intros. rewrite anti_number_d, number_d by assumption. ring. Qed.

Theorem pauli_c : forall k psi s, cdag k (cdag k psi) s = 0.
Proof.
  intros k psi [m n]; unfold cdag; cbn [fst snd].
  destruct (get m k) eqn:E; [|reflexivity].
  rewrite get_upd_same by (apply get_true_lt; exact E). ring.
Qed.

Theorem pauli_d : forall k psi s, ddag k (ddag k psi) s = 0.
Proof.
  intros k psi [m n]; unfold ddag; cbn [fst snd].
  destruct (get n k) eqn:E; [|reflexivity].
  rewrite get_upd_same by (apply get_true_lt; exact E). ring.
Qed.

Theorem cdag_anticomm : forall k j psi s, k <> j ->
  cdag k (cdag j psi) s = - cdag j (cdag k psi) s.
Proof.
  intros k j psi [m n] Hkj; unfold cdag; cbn [fst snd].
  rewrite (get_upd_other m k j false) by lia. rewrite (get_upd_other m j k false) by lia.
  destruct (get m k) eqn:Ek, (get m j) eqn:Ej; try ring.
  pose proof (get_true_lt m k Ek) as Hk. pose proof (get_true_lt m j Ej) as Hj.
  rewrite (upd_comm m k j false false) by lia.
  destruct (Nat.lt_ge_cases k j) as [Hlt|Hge].
  - pose proof (below_m_lt m n j k false Hlt Hk) as B. rewrite Ek in B. simpl b2n in B.
    rewrite (below_m_ge m n k j false) by lia.
    replace (below (m, n) j) with (S (below (upd m k false, n) j)) by lia.
    rewrite sgn_S. ring.
  - assert (Hlt : (j < k)%nat) by lia.
    pose proof (below_m_lt m n k j false Hlt Hj) as B. rewrite Ej in B. simpl b2n in B.
    rewrite (below_m_ge m n j k false) by lia.
    replace (below (m, n) k) with (S (below (upd m j false, n) k)) by lia.
    rewrite sgn_S. ring.
Qed.

Theorem c_cdag_anticomm : forall k j psi s, k <> j ->
  c k (cdag j psi) s = - cdag j (c k psi) s.
Proof.
  intros k j psi [m n] Hkj; unfold cdag, c; cbn [fst snd].
  rewrite (get_upd_other m k j true) by lia. rewrite (get_upd_other m j k false) by lia.
  destruct (get m k) eqn:Ek, (get m j) eqn:Ej; try ring.
  pose proof (get_true_lt m j Ej) as Hj.
  rewrite (upd_comm m k j true false) by lia.
  destruct (Nat.lt_ge_cases k j) as [Hlt|Hge].
  - pose proof (below_m_lt m n j k true Hlt ltac:(lia)) as B. rewrite Ek in B. simpl b2n in B.
    rewrite (below_m_ge m n k j false) by lia.
    replace (below (upd m k true, n) j) with (S (below (m, n) j)) by lia.
    rewrite sgn_S. ring.
  - assert (Hlt : (j < k)%nat) by lia.
    pose proof (below_m_lt m n k j false Hlt Hj) as B. rewrite Ej in B. simpl b2n in B.
    rewrite (below_m_ge m n j k true) by lia.
    replace (below (m, n) k) with (S (below (upd m j false, n) k)) by lia.
    rewrite sgn_S. ring.
Qed.

Theorem ddag_anticomm : forall k j psi s, k <> j ->
  ddag k (ddag j psi) s = - ddag j (ddag k psi) s.
Proof.
  intros k j psi [m n] Hkj; unfold ddag, dsg; cbn [fst snd].
  rewrite (get_upd_other n k j false) by lia. rewrite (get_upd_other n j k false) by lia.
  destruct (get n k) eqn:Ek, (get n j) eqn:Ej; try ring.
  pose proof (get_true_lt n k Ek) as Hk. pose proof (get_true_lt n j Ej) as Hj.
  rewrite (upd_comm n k j false false) by lia.
  destruct (Nat.lt_ge_cases k j) as [Hlt|Hge].
  - pose proof (below_n_lt m n j k false Hlt Hk) as B. rewrite Ek in B. simpl b2n in B.
    rewrite (below_n_ge m n k j false) by lia.
    replace (below (m, n) j) with (S (below (m, upd n k false) j)) by lia.
    rewrite Nat.add_succ_l, sgn_S. ring.
  - assert (Hlt : (j < k)%nat) by lia.
    pose proof (below_n_lt m n k j false Hlt Hj) as B. rewrite Ej in B. simpl b2n in B.
    rewrite (below_n_ge m n j k false) by lia.
    replace (below (m, n) k) with (S (below (m, upd n j false) k)) by lia.
    rewrite Nat.add_succ_l, sgn_S. ring.
Qed.

Theorem cdag_ddag_anticomm : forall k j psi s,
  cdag k (ddag j psi) s = - ddag j (cdag k psi) s.
Proof.
  intros k j psi [m n]; unfold cdag, ddag, dsg; cbn [fst snd].
  destruct (get m k) eqn:Ek, (get n j) eqn:Ej; try ring.
  pose proof (get_true_lt m k Ek) as Hk. pose proof (get_true_lt n j Ej) as Hj.
  destruct (Nat.lt_total k j) as [Hlt|[Heq|Hgt]].
  - pose proof (below_m_lt m n j k false Hlt Hk) as B. rewrite Ek in B. simpl b2n in B.
    rewrite (get_upd_other m k j false) by lia.
    rewrite (below_n_ge m n k j false) by lia.
    replace (below (m, n) j) with (S (below (upd m k false, n) j)) by lia.
    rewrite Nat.add_succ_l, sgn_S. ring.
  - subst j. rewrite get_upd_same by exact Hk. rewrite Ek.
    rewrite (below_m_ge m n k k false), (below_n_ge m n k k false) by lia.
    simpl b2n. rewrite Nat.add_0_r, Nat.add_1_r, sgn_S. ring.
  - pose proof (below_n_lt m n k j false Hgt Hj) as B. rewrite Ej in B. simpl b2n in B.
    rewrite (get_upd_other m k j false) by lia.
    rewrite (below_m_ge m n j k false) by lia.
    replace (below (m, n) k) with (S (below (m, upd n j false) k)) by lia.
    rewrite sgn_S. ring.
Qed.

(** A2: molecules are bosons *)
Definition Xdag (k : nat) (psi : Vec) : Vec := cdag k (ddag k psi).
Definition X (k : nat) (psi : Vec) : Vec := d k (c k psi).

Lemma Xdag_apply : forall k psi s, Xdag k psi s =
  if get (fst s) k then (if get (snd s) k then psi (upd (fst s) k false, upd (snd s) k false) else 0)
  else 0.
Proof.
  intros k psi [m n]. unfold Xdag, cdag, ddag, dsg; cbn [fst snd].
  destruct (get m k) eqn:Em; [|reflexivity].
  destruct (get n k) eqn:En; [|ring].
  rewrite get_upd_same by (apply get_true_lt; exact Em).
  rewrite below_m_ge by lia. simpl b2n. rewrite Nat.add_0_r, Z.mul_assoc, sgn_sq. ring.
Qed.

Lemma X_apply : forall k psi s, X k psi s =
  if get (snd s) k then 0
  else (if get (fst s) k then 0 else psi (upd (fst s) k true, upd (snd s) k true)).
Proof.
  intros k psi [m n]. unfold X, c, d, dsg; cbn [fst snd].
  destruct (get n k) eqn:En; [reflexivity|].
  destruct (get m k) eqn:Em; [ring|].
  rewrite below_n_ge by lia. simpl b2n. rewrite Nat.add_0_r, Z.mul_assoc, sgn_sq. ring.
Qed.

(** MOLECULES ARE BOSONS: molecule creation operators commute. *)
Theorem molecules_commute : forall k j psi s, Xdag k (Xdag j psi) s = Xdag j (Xdag k psi) s.
Proof.
  intros k j psi [m n].
  destruct (Nat.eq_dec k j) as [->|Hkj]; [reflexivity|].
  rewrite !Xdag_apply; cbn [fst snd].
  rewrite (get_upd_other m k j), (get_upd_other n k j), (get_upd_other m j k), (get_upd_other n j k) by lia.
  destruct (get m k), (get n k), (get m j), (get n j); try reflexivity.
  rewrite (upd_comm m k j), (upd_comm n k j) by lia. reflexivity.
Qed.

Theorem molecule_pauli : forall k psi s, Xdag k (Xdag k psi) s = 0.
Proof.
  intros k psi [m n]. rewrite !Xdag_apply; cbn [fst snd].
  destruct (get m k) eqn:Em; [|reflexivity]. destruct (get n k); [|reflexivity].
  rewrite get_upd_same by (apply get_true_lt; exact Em). reflexivity.
Qed.

Theorem molecule_hardcore : forall k psi s,
  (k < length (fst s))%nat -> (k < length (snd s))%nat ->
  X k (Xdag k psi) s - Xdag k (X k psi) s = (1 - b2z (get (fst s) k) - b2z (get (snd s) k)) * psi s.
Proof.
  intros k psi [m n] Hm Hn; simpl in Hm, Hn.
  rewrite (X_apply k (Xdag k psi)), (Xdag_apply k (X k psi)); cbn [fst snd].
  destruct (get m k) eqn:Em, (get n k) eqn:En; unfold b2z.
  - rewrite X_apply; cbn [fst snd]. rewrite !get_upd_same by assumption.
    rewrite !upd_upd, !upd_eq by assumption. ring.
  - ring.
  - ring.
  - rewrite Xdag_apply; cbn [fst snd]. rewrite !get_upd_same by assumption.
    rewrite !upd_upd, !upd_eq by assumption. ring.
Qed.

Theorem molecule_cross : forall k j psi s, k <> j -> X k (Xdag j psi) s = Xdag j (X k psi) s.
Proof.
  intros k j psi [m n] Hkj. rewrite (X_apply k (Xdag j psi)), (Xdag_apply j (X k psi)); cbn [fst snd].
  destruct (get n k) eqn:Enk, (get m k) eqn:Emk, (get m j) eqn:Emj, (get n j) eqn:Enj;
    rewrite ?Xdag_apply, ?X_apply; cbn [fst snd];
    rewrite ?(get_upd_other m k j), ?(get_upd_other n k j), ?(get_upd_other m j k), ?(get_upd_other n j k) by lia;
    rewrite ?Enk, ?Emk, ?Emj, ?Enj; try reflexivity.
  rewrite (upd_comm m k j), (upd_comm n k j) by lia. reflexivity.
Qed.

Lemma below_mol : forall m n k j, get m k = true -> get n k = true ->
  sgn (below (upd m k false, upd n k false) j) = sgn (below (m, n) j).
Proof.
  intros m n k j Em En.
  pose proof (get_true_lt m k Em) as Hm. pose proof (get_true_lt n k En) as Hn.
  destruct (Nat.lt_ge_cases k j) as [Hlt|Hge].
  - pose proof (below_n_lt (upd m k false) n j k false Hlt Hn) as B1.
    pose proof (below_m_lt m n j k false Hlt Hm) as B2.
    rewrite En in B1. rewrite Em in B2. simpl b2n in B1, B2.
    replace (below (m, n) j) with (S (S (below (upd m k false, upd n k false) j))) by lia.
    apply eq_sym, sgn_SS.
  - rewrite (below_n_ge (upd m k false) n j k false Hge), (below_m_ge m n j k false Hge). reflexivity.
Qed.

Theorem molecule_atom_comm_c : forall k j psi s, j <> k ->
  cdag j (Xdag k psi) s = Xdag k (cdag j psi) s.
Proof.
  intros k j psi [m n] Hjk. unfold cdag at 1; rewrite Xdag_apply; cbn [fst snd].
  destruct (get m j) eqn:Emj, (get m k) eqn:Emk, (get n k) eqn:Enk;
    rewrite ?Xdag_apply; unfold cdag; cbn [fst snd];
    rewrite ?(get_upd_other m j k), ?(get_upd_other m k j), ?(get_upd_other n k j) by lia;
    rewrite ?Emj, ?Emk, ?Enk; try ring.
  rewrite below_mol by assumption. rewrite (upd_comm m j k) by lia. reflexivity.
Qed.

Theorem molecule_atom_comm_d : forall k j psi s, j <> k ->
  ddag j (Xdag k psi) s = Xdag k (ddag j psi) s.
Proof.
  intros k j psi [m n] Hjk. unfold ddag at 1; rewrite Xdag_apply; unfold dsg; cbn [fst snd].
  destruct (get n j) eqn:Enj, (get m k) eqn:Emk, (get n k) eqn:Enk;
    rewrite ?Xdag_apply; unfold ddag, dsg; cbn [fst snd];
    rewrite ?(get_upd_other n j k), ?(get_upd_other m k j), ?(get_upd_other n k j) by lia;
    rewrite ?Enj, ?Emk, ?Enk; try ring.
  assert (E : sgn (below (upd m k false, upd n k false) j + b2n (get m j))
            = sgn (below (m, n) j + b2n (get m j))).
  { pose proof (below_mol m n k j Emk Enk) as B. unfold sgn in *.
    rewrite !Nat.even_add. destruct (Nat.even (below (upd m k false, upd n k false) j)),
      (Nat.even (below (m, n) j)); try discriminate; reflexivity. }
  rewrite E, (upd_comm n j k) by lia. reflexivity.
Qed.

(** Finite sums over the mode list [seq 0 K]. *)
Definition sumZ (f : nat -> Z) (l : list nat) : Z := fold_right (fun k acc => f k + acc) 0 l.
Lemma sumZ_cons : forall f a l, sumZ f (a :: l) = f a + sumZ f l.
Proof. reflexivity. Qed.

Lemma sumZ_ext : forall f g l, (forall k, In k l -> f k = g k) -> sumZ f l = sumZ g l.
Proof.
  intros f g l; induction l as [|a l IH]; intros H; [reflexivity|].
  rewrite !sumZ_cons, H by (left; reflexivity).
  rewrite IH by (intros; apply H; right; assumption). reflexivity.
Qed.

Lemma sumZ_mul_r : forall f x l, sumZ (fun k => f k * x) l = sumZ f l * x.
Proof. intros f x l; induction l as [|a l IH]; [reflexivity|]. rewrite !sumZ_cons, IH. ring. Qed.

Lemma sumZ_opp : forall f l, sumZ (fun k => - f k) l = - sumZ f l.
Proof. intros f l; induction l as [|a l IH]; [reflexivity|]. rewrite !sumZ_cons, IH. ring. Qed.

Lemma sumZ_zero : forall f l, (forall k, In k l -> f k = 0) -> sumZ f l = 0.
Proof.
  intros f l; induction l as [|a l IH]; intros H; [reflexivity|].
  rewrite sumZ_cons, H by (left; reflexivity).
  rewrite IH by (intros; apply H; right; assumption). reflexivity.
Qed.

Lemma sumZ_scale_2 : forall f l, sumZ (fun k => 2 * f k) l = 2 * sumZ f l.
Proof. intros f l; induction l as [|a l IH]; [reflexivity|]. rewrite !sumZ_cons, IH. ring. Qed.

Lemma sumZ_point : forall f g l k, NoDup l -> In k l -> (forall j, j <> k -> f j = g j) ->
  sumZ f l = sumZ g l + (f k - g k).
Proof.
  intros f g l k ND; induction ND as [|a l Ha ND IH]; intros Hin Hfg; [destruct Hin|].
  rewrite !sumZ_cons. destruct (Nat.eq_dec a k) as [->|Hak].
  - rewrite (sumZ_ext f g l) by (intros j Hj; apply Hfg; intros ->; contradiction). ring.
  - destruct Hin as [->|Hin]; [contradiction|].
    rewrite (Hfg a Hak), IH by assumption. ring.
Qed.

(** A3: the Wheeler-DeWitt operator *)
Definition H (eps : nat -> Z) (K : nat) (psi : Vec) : Vec := fun s =>
  sumZ (fun k => eps k * (cdag k (c k psi) s - ddag k (d k psi) s)) (seq 0 K).

Definition energy (eps : nat -> Z) (K : nat) (s : Config) : Z :=
  sumZ (fun k => eps k * (b2z (get (fst s) k) - b2z (get (snd s) k))) (seq 0 K).

Theorem H_diag : forall eps K psi s, H eps K psi s = energy eps K s * psi s.
Proof.
  intros. unfold H, energy. rewrite <- sumZ_mul_r. apply sumZ_ext; intros k _.
  rewrite number_c, number_d. ring.
Qed.

Definition mirror (psi : Vec) : Vec := fun s => psi (snd s, fst s).

Lemma energy_mirror : forall eps K s, energy eps K (snd s, fst s) = - energy eps K s.
Proof.
  intros. unfold energy. rewrite <- sumZ_opp. apply sumZ_ext; intros k _. cbn [fst snd]. ring.
Qed.

Theorem H_mirror : forall eps K psi s, H eps K (mirror psi) s = - mirror (H eps K psi) s.
Proof.
  intros eps K psi [m n]. rewrite H_diag. unfold mirror; cbn [fst snd].
  rewrite H_diag. pose proof (energy_mirror eps K (m, n)) as E; simpl in E. rewrite E. ring.
Qed.

Lemma energy_clear_both : forall eps K m n k, get m k = true -> get n k = true ->
  energy eps K (upd m k false, upd n k false) = energy eps K (m, n).
Proof.
  intros eps K m n k Hm Hn. unfold energy. apply sumZ_ext; intros j _. cbn [fst snd]. f_equal.
  destruct (Nat.eq_dec j k) as [->|Hjk].
  - rewrite !get_upd_same by (apply get_true_lt; assumption). rewrite Hm, Hn. reflexivity.
  - rewrite !get_upd_other by exact Hjk. reflexivity.
Qed.

Theorem H_Xdag_comm : forall eps K k psi s, H eps K (Xdag k psi) s = Xdag k (H eps K psi) s.
Proof.
  intros eps K k psi [m n]. rewrite H_diag, !Xdag_apply; cbn [fst snd].
  destruct (get m k) eqn:E1; [|ring].
  destruct (get n k) eqn:E2; [|ring].
  rewrite H_diag, energy_clear_both by assumption. reflexivity.
Qed.

(** A4: vacuum, molecular states, condensates *)
Definition vac (K : nat) : Vec := fun s =>
  if list_eq_dec bool_dec (fst s) (repeat false K) then
    if list_eq_dec bool_dec (snd s) (repeat false K) then 1 else 0
  else 0.

Theorem H_vac : forall eps K s, H eps K (vac K) s = 0.
Proof.
  intros eps K [m n]. rewrite H_diag. unfold vac; cbn [fst snd].
  destruct (list_eq_dec bool_dec m (repeat false K)) as [Em|]; [|ring].
  destruct (list_eq_dec bool_dec n (repeat false K)) as [En|]; [|ring].
  subst. replace (energy eps K (repeat false K, repeat false K)) with 0; [ring|].
  symmetry. unfold energy. apply sumZ_zero; intros k _. cbn [fst snd]. rewrite get_repeat_false. ring.
Qed.

Definition molecules (ks : list nat) (psi : Vec) : Vec := fold_right Xdag psi ks.

(** MAIN THEOREM: every multi-molecule state solves the Wheeler-DeWitt equation. *)
Theorem wdw_molecular_solution : forall eps K ks s, H eps K (molecules ks (vac K)) s = 0.
Proof.
  intros eps K ks; induction ks as [|k ks IH]; intros s; simpl.
  - apply H_vac.
  - rewrite H_Xdag_comm, Xdag_apply.
    destruct (get (fst s) k); [|reflexivity].
    destruct (get (snd s) k); [|reflexivity]. apply IH.
Qed.

Definition occ (K : nat) (ks : list nat) : list bool :=
  fold_right (fun k l => upd l k true) (repeat false K) ks.

Lemma length_occ : forall K ks, length (occ K ks) = K.
Proof.
  intros K ks; induction ks as [|k ks IH]; simpl.
  - apply repeat_length.
  - rewrite length_upd. exact IH.
Qed.

Fixpoint fresh (K : nat) (ks : list nat) : Prop :=
  match ks with
  | [] => True
  | k :: t => (k < K)%nat /\ get (occ K t) k = false /\ fresh K t
  end.

Theorem molecules_at_occ : forall K ks, fresh K ks ->
  molecules ks (vac K) (occ K ks, occ K ks) = 1.
Proof.
  intros K ks; induction ks as [|k ks IH]; intros Hf; simpl.
  - unfold vac; cbn [fst snd].
    destruct (list_eq_dec bool_dec (repeat false K) (repeat false K)); [|congruence]. reflexivity.
  - destruct Hf as [Hk [Hocc Hf]].
    assert (Hk' : (k < length (occ K ks))%nat) by (rewrite length_occ; exact Hk).
    rewrite Xdag_apply; cbn [fst snd]. rewrite get_upd_same by exact Hk'.
    rewrite upd_upd, upd_eq by exact Hocc. apply IH, Hf.
Qed.

Theorem molecules_support : forall K ks s, molecules ks (vac K) s <> 0 -> s = (occ K ks, occ K ks).
Proof.
  intros K ks; induction ks as [|k ks IH]; intros [m n] Hs; cbn [molecules fold_right occ] in *.
  - unfold vac in Hs; cbn [fst snd] in Hs.
    destruct (list_eq_dec bool_dec m (repeat false K)) as [->|]; [|congruence].
    destruct (list_eq_dec bool_dec n (repeat false K)) as [->|]; [|congruence]. reflexivity.
  - rewrite Xdag_apply in Hs; cbn [fst snd] in Hs.
    destruct (get m k) eqn:E1; [|congruence].
    destruct (get n k) eqn:E2; [|congruence].
    apply IH in Hs. injection Hs as Hm Hn.
    rewrite <- (upd_eq m k true E1), <- (upd_eq n k true E2).
    rewrite <- (upd_upd m k false true), <- (upd_upd n k false true), Hm, Hn. reflexivity.
Qed.

Theorem distinct_solutions : forall K ks ks', occ K ks <> occ K ks' ->
  molecules ks (vac K) (occ K ks', occ K ks') = 0.
Proof.
  intros K ks ks' Hne.
  destruct (Z.eq_dec (molecules ks (vac K) (occ K ks', occ K ks')) 0) as [E|E]; [exact E|].
  exfalso. apply molecules_support in E. injection E as E1 _. congruence.
Qed.

Theorem molecules_repeat : forall k ks psi s, molecules (k :: k :: ks) psi s = 0.
Proof. intros. apply molecule_pauli. Qed.

Fixpoint bcs (u v : nat -> Z) (ks : list nat) (psi : Vec) : Vec :=
  match ks with
  | [] => psi
  | k :: t => fun s => u k * bcs u v t psi s + v k * Xdag k (bcs u v t psi) s
  end.

(** MOLECULAR CONDENSATES SOLVE THE CONSTRAINT. *)
Theorem bcs_solution : forall eps K u v ks s, H eps K (bcs u v ks (vac K)) s = 0.
Proof.
  intros eps K u v ks; induction ks as [|k ks IH]; intros s; simpl.
  - apply H_vac.
  - rewrite H_diag.
    assert (h1 : energy eps K s * bcs u v ks (vac K) s = 0) by (rewrite <- H_diag; apply IH).
    assert (h2 : energy eps K s * Xdag k (bcs u v ks (vac K)) s = 0).
    { rewrite <- H_diag, H_Xdag_comm, Xdag_apply.
      destruct (get (fst s) k); [|reflexivity]. destruct (get (snd s) k); [|reflexivity]. apply IH. }
    transitivity (u k * (energy eps K s * bcs u v ks (vac K) s)
                  + v k * (energy eps K s * Xdag k (bcs u v ks (vac K)) s)); [ring|].
    rewrite h1, h2. ring.
Qed.

Lemma energy_pair : forall eps K m n k j, length m = K -> length n = K -> (k < K)%nat -> (j < K)%nat ->
  get m k = false -> get n j = false ->
  energy eps K (upd m k true, upd n j true) = energy eps K (m, n) + eps k - eps j.
Proof.
  intros eps K m n k j Lm Ln Hk Hj Em En.
  assert (S1 : energy eps K (upd m k true, n) = energy eps K (m, n) + eps k).
  { unfold energy; cbn [fst snd].
    rewrite (sumZ_point _ (fun i => eps i * (b2z (get m i) - b2z (get n i))) (seq 0 K) k).
    - rewrite get_upd_same by lia. rewrite Em. unfold b2z. ring.
    - apply seq_NoDup.
    - apply in_seq; lia.
    - intros i Hi. rewrite get_upd_other by exact Hi. reflexivity. }
  assert (S2 : energy eps K (upd m k true, upd n j true) = energy eps K (upd m k true, n) - eps j).
  { unfold energy; cbn [fst snd].
    rewrite (sumZ_point _ (fun i => eps i * (b2z (get (upd m k true) i) - b2z (get n i))) (seq 0 K) j).
    - rewrite (get_upd_same n j) by lia. rewrite En. unfold b2z. ring.
    - apply seq_NoDup.
    - apply in_seq; lia.
    - intros i Hi. rewrite (get_upd_other n j i) by exact Hi. reflexivity. }
  rewrite S2, S1. ring.
Qed.

(** PAIR (MOLECULAR-FIELD) SHIFT: [H, d_j c_k] = (eps_j - eps_k) d_j c_k. *)
Theorem pair_shift : forall eps K k j psi s,
  length (fst s) = K -> length (snd s) = K -> (k < K)%nat -> (j < K)%nat ->
  H eps K (d j (c k psi)) s - d j (c k (H eps K psi)) s = (eps j - eps k) * d j (c k psi) s.
Proof.
  intros eps K k j psi [m n] Lm Ln Hk Hj; cbn [fst snd] in Lm, Ln.
  rewrite H_diag. unfold d, c; cbn [fst snd].
  destruct (get n j) eqn:En; [ring|].
  destruct (get m k) eqn:Em; [ring|].
  rewrite H_diag, energy_pair by assumption. ring.
Qed.

(** A5: balanced form, zero point, no-go, scale, Stelle sign, rational independence *)
Theorem balance_iff : forall (Hop T V : Vec -> Vec) psi,
  (forall phi s, Hop phi s = T phi s + V phi s) ->
  ((forall s, Hop psi s = 0) <-> (forall s, T psi s = - V psi s)).
Proof. intros Hop T V psi Hs; split; intros Hz s; specialize (Hz s); rewrite Hs in *; lia. Qed.

Definition H_weyl2 (eps : nat -> Z) (K : nat) (psi : Vec) : Vec := fun s =>
  sumZ (fun k => eps k * ((cdag k (c k psi) s - c k (cdag k psi) s)
                        - (ddag k (d k psi) s - d k (ddag k psi) s))) (seq 0 K).

Theorem zero_point_cancellation : forall eps K psi s,
  length (fst s) = K -> length (snd s) = K -> H_weyl2 eps K psi s = 2 * H eps K psi s.
Proof.
  intros eps K psi s L1 L2. unfold H_weyl2, H. rewrite <- sumZ_scale_2. apply sumZ_ext; intros k Hk.
  apply in_seq in Hk.
  rewrite anti_number_c, anti_number_d, number_c, number_d by lia. ring.
Qed.

Theorem no_go : forall t v : Z, 0 <= t -> 0 <= v -> t + v = 0 -> t = 0 /\ v = 0.
Proof. lia. Qed.

Theorem kernel_scale : forall (Hop : Vec -> Vec) (lam : Z) psi, lam <> 0 ->
  ((forall s, lam * Hop psi s = 0) <-> (forall s, Hop psi s = 0)).
Proof.
  intros Hop lam psi Hl; split; intros Hz s; specialize (Hz s).
  - apply Z.mul_eq_0 in Hz. destruct Hz; [contradiction|assumption].
  - rewrite Hz. ring.
Qed.

Theorem stelle_channel_tachyon : forall cS : Z, ~ (3 * (- cS) + 4 * cS > 0 /\ 4 * cS < 0).
Proof. lia. Qed.

Definition rat_indep (eps : nat -> Z) (K : nat) : Prop :=
  forall cf : nat -> Z, sumZ (fun k => cf k * eps k) (seq 0 K) = 0 -> forall k, (k < K)%nat -> cf k = 0.

Theorem kernel_molecular : forall eps K s, rat_indep eps K -> energy eps K s = 0 ->
  forall k, (k < K)%nat -> get (fst s) k = get (snd s) k.
Proof.
  intros eps K s RI E k Hk.
  pose proof (RI (fun j => b2z (get (fst s) j) - b2z (get (snd s) j))) as R.
  assert (Hsum : sumZ (fun j => (b2z (get (fst s) j) - b2z (get (snd s) j)) * eps j) (seq 0 K) = 0).
  { rewrite <- E. unfold energy. apply sumZ_ext; intros; ring. }
  specialize (R Hsum k Hk). cbn beta in R. unfold b2z in R.
  destruct (get (fst s) k), (get (snd s) k); try reflexivity; lia.
Qed.

End FockSpace.

(* ===================================================================== *)
(** * Part B. Spinor kinematics over an arbitrary commutative ring        *)
(* ===================================================================== *)

Module SpinorIdentities.
Section Generic.
Variable R : Type.
Variables (rO rI : R) (radd rmul rsub : R -> R -> R) (ropp : R -> R).
Hypothesis Rth : ring_theory rO rI radd rmul rsub ropp (@eq R).
Add Ring GenericRing : Rth.

Declare Scope cr_scope.
Delimit Scope cr_scope with cr.
Local Infix "+" := radd : cr_scope.
Local Infix "*" := rmul : cr_scope.
Local Infix "-" := rsub : cr_scope.
Local Notation "- x" := (ropp x) : cr_scope.
Local Notation "0" := rO : cr_scope.
Local Notation "1" := rI : cr_scope.
Local Open Scope cr_scope.

(** Complex numbers over R *)
Record cx := mkC { re : R; im : R }.
Definition cadd (a b : cx) := mkC (re a + re b) (im a + im b).
Definition csub (a b : cx) := mkC (re a - re b) (im a - im b).
Definition cneg (a : cx) := mkC (- re a) (- im a).
Definition cmul (a b : cx) := mkC (re a * re b - im a * im b) (re a * im b + im a * re b).
Definition cconj (a : cx) := mkC (re a) (- im a).
Definition c0 := mkC 0 0.
Definition c1 := mkC 1 0.
Definition ci := mkC 0 1.
Definition creal (x : R) := mkC x 0.

Lemma cx_eq : forall a b, re a = re b -> im a = im b -> a = b.
Proof. intros [a1 a2] [b1 b2]; simpl; intros -> ->; reflexivity. Qed.

(** Two-component spinor index and the antisymmetric pairing eps_01 = +1 *)
Inductive I2 := i0 | i1.
Definition epsC (A B : I2) : cx :=
  match A, B with i0, i1 => c1 | i1, i0 => cneg c1 | _, _ => c0 end.
Definition sum2 (f : I2 -> cx) : cx := cadd (f i0) (f i1).

(** B1: [eq:etaeps].  X~^{AA'} = X^0 1 + X^i sigma_i (unnormalized soldering). *)
Definition Xt (x0 x1 x2 x3 : R) (A A' : I2) : cx :=
  match A, A' with
  | i0, i0 => creal (x0 + x3) | i0, i1 => mkC x1 (- x2)
  | i1, i0 => mkC x1 x2       | i1, i1 => creal (x0 - x3)
  end.
Definition etaR (x0 x1 x2 x3 y0 y1 y2 y3 : R) : R := - (x0 * y0) + x1 * y1 + x2 * y2 + x3 * y3.

Theorem eta_from_pairing : forall x0 x1 x2 x3 y0 y1 y2 y3,
  cneg (sum2 (fun A => sum2 (fun B => sum2 (fun A' => sum2 (fun B' =>
     cmul (cmul (epsC A B) (epsC A' B')) (cmul (Xt x0 x1 x2 x3 A A') (Xt y0 y1 y2 y3 B B')))))))
  = creal ((1 + 1) * etaR x0 x1 x2 x3 y0 y1 y2 y3).
Proof. intros; apply cx_eq; cbv; ring. Qed.

Theorem det_is_minus_eta : forall x0 x1 x2 x3,
  csub (cmul (Xt x0 x1 x2 x3 i0 i0) (Xt x0 x1 x2 x3 i1 i1)) (cmul (Xt x0 x1 x2 x3 i0 i1) (Xt x0 x1 x2 x3 i1 i0))
  = creal (- etaR x0 x1 x2 x3 x0 x1 x2 x3).
Proof. intros; apply cx_eq; cbv; ring. Qed.

(** B2: [prop:null], a single molecule xi xi^dagger has zero determinant (is null). *)
Theorem single_molecule_null : forall xi : I2 -> cx,
  csub (cmul (cmul (xi i0) (cconj (xi i0))) (cmul (xi i1) (cconj (xi i1))))
       (cmul (cmul (xi i0) (cconj (xi i1))) (cmul (xi i1) (cconj (xi i0)))) = c0.
Proof. intros; apply cx_eq; cbv; ring. Qed.

(** B3: dyad identity (App. A) and [prop:tetrad] in spinor form *)
Definition omega (psi phi : I2 -> cx) : cx :=
  sum2 (fun A => sum2 (fun B => cmul (epsC A B) (cmul (psi A) (phi B)))).

Theorem dyad_identity : forall (psi phi : I2 -> cx) (A B : I2),
  csub (cmul (psi A) (phi B)) (cmul (phi A) (psi B)) = cmul (omega psi phi) (epsC A B).
Proof. intros psi phi A B; destruct A, B; apply cx_eq; cbv; ring. Qed.

(** Atoms of two flavors: alpha = i0 is o, alpha = i1 is iota.  The molecule
    X_{alpha gamma}^{A A'} = atom_alpha^A * conj(atom_gamma^{A'}).  [prop:tetrad] states
    - eps^{alpha beta} eps^{gamma delta} X_{alpha gamma} X_{beta delta}
      = - |omega_12|^2 eps^{AB} eps^{A'B'},
    which by [eq:etaeps] is |omega_12|^2 eta^{ab}. *)
Definition atom (o io : I2 -> cx) (alpha : I2) : I2 -> cx := match alpha with i0 => o | i1 => io end.

Theorem tetrad_metric_spinor : forall (o io : I2 -> cx) (A A' B B' : I2),
  cneg (sum2 (fun al => sum2 (fun be => sum2 (fun ga => sum2 (fun de =>
     cmul (cmul (epsC al be) (epsC ga de))
          (cmul (cmul (atom o io al A) (cconj (atom o io ga A')))
                (cmul (atom o io be B) (cconj (atom o io de B')))))))))
  = cneg (cmul (cmul (omega o io) (cconj (omega o io))) (cmul (epsC A B) (epsC A' B'))).
Proof. intros o io A A' B B'; destruct A, A', B, B'; apply cx_eq; cbv; ring. Qed.

(** B4: [eq:chiral_trace] times 16.  S~ = 4 S(theta) = -2 theta_{0i} sigma_i + i eps_{ijk} theta_{ij} sigma_k. *)
Inductive I4 := e0 | e1 | e2 | e3.
Definition sum4 (f : I4 -> cx) : cx := cadd (cadd (f e0) (f e1)) (cadd (f e2) (f e3)).

(* antisymmetric theta_{ab} from six independent components *)
Definition theta (t01 t02 t03 t12 t13 t23 : R) (a b : I4) : R :=
  match a, b with
  | e0, e1 => t01 | e0, e2 => t02 | e0, e3 => t03 | e1, e2 => t12 | e1, e3 => t13 | e2, e3 => t23
  | e1, e0 => - t01 | e2, e0 => - t02 | e3, e0 => - t03 | e2, e1 => - t12 | e3, e1 => - t13 | e3, e2 => - t23
  | _, _ => 0
  end.
Definition etaD (a : I4) : R := match a with e0 => - (1) | _ => 1 end.
(* Levi-Civita symbols eps_{0123} = +1 and eps_{123} = +1 (spatial indices e1 e2 e3) *)
Definition lc3 (i j k : I4) : R :=
  match i, j, k with
  | e1, e2, e3 => 1 | e2, e3, e1 => 1 | e3, e1, e2 => 1
  | e1, e3, e2 => - (1) | e3, e2, e1 => - (1) | e2, e1, e3 => - (1)
  | _, _, _ => 0
  end.
(** We define eps_{abcd} directly by its value on all 24 permutations of (0,1,2,3). *)
Definition lc4 (a b c d : I4) : R :=
  match a, b, c, d with
  | e0,e1,e2,e3 => 1 | e0,e1,e3,e2 => - (1) | e0,e2,e1,e3 => - (1) | e0,e2,e3,e1 => 1 | e0,e3,e1,e2 => 1 | e0,e3,e2,e1 => - (1)
  | e1,e0,e2,e3 => - (1) | e1,e0,e3,e2 => 1 | e1,e2,e0,e3 => 1 | e1,e2,e3,e0 => - (1) | e1,e3,e0,e2 => - (1) | e1,e3,e2,e0 => 1
  | e2,e0,e1,e3 => 1 | e2,e0,e3,e1 => - (1) | e2,e1,e0,e3 => - (1) | e2,e1,e3,e0 => 1 | e2,e3,e0,e1 => 1 | e2,e3,e1,e0 => - (1)
  | e3,e0,e1,e2 => - (1) | e3,e0,e2,e1 => 1 | e3,e1,e0,e2 => 1 | e3,e1,e2,e0 => - (1) | e3,e2,e0,e1 => - (1) | e3,e2,e1,e0 => 1
  | _,_,_,_ => 0
  end.

(* 2 x 2 complex matrices *)
Definition M2 := I2 -> I2 -> cx.
Definition madd (P Q : M2) : M2 := fun A B => cadd (P A B) (Q A B).
Definition mmul (P Q : M2) : M2 := fun A B => sum2 (fun C => cmul (P A C) (Q C B)).
Definition mscal (z : cx) (P : M2) : M2 := fun A B => cmul z (P A B).
Definition mzero : M2 := fun _ _ => c0.
Definition tr2 (P : M2) : cx := cadd (P i0 i0) (P i1 i1).
Definition sigma (k : I4) : M2 := fun A B =>
  match k, A, B with
  | e1, i0, i1 => c1 | e1, i1, i0 => c1
  | e2, i0, i1 => cneg ci | e2, i1, i0 => ci
  | e3, i0, i0 => c1 | e3, i1, i1 => cneg c1
  | e0, i0, i0 => c1 | e0, i1, i1 => c1
  | _, _, _ => c0
  end.
Definition sumM (f : I4 -> M2) : M2 := fun A B => sum4 (fun k => f k A B).
Definition spatial (k : I4) : bool := match k with e0 => false | _ => true end.

Definition St (t01 t02 t03 t12 t13 t23 : R) : M2 :=
  madd (sumM (fun i => if spatial i then mscal (creal (- (1 + 1) * theta t01 t02 t03 t12 t13 t23 e0 i)) (sigma i) else mzero))
       (sumM (fun k => if spatial k then
          mscal (cmul ci (sum4 (fun i => sum4 (fun j => creal (lc3 i j k * theta t01 t02 t03 t12 t13 t23 i j))))) (sigma k)
        else mzero)).

Definition up (th : I4 -> I4 -> R) (a b : I4) : R := etaD a * etaD b * th a b.

Theorem chiral_trace_times16 : forall t01 t02 t03 t12 t13 t23 u01 u02 u03 u12 u13 u23,
  let th := theta t01 t02 t03 t12 t13 t23 in let th' := theta u01 u02 u03 u12 u13 u23 in
  tr2 (mmul (St t01 t02 t03 t12 t13 t23) (St u01 u02 u03 u12 u13 u23))
  = cadd (creal (- (1 + 1 + 1 + 1) * re (sum4 (fun a => sum4 (fun b => creal (th a b * up th' a b))))))
         (cmul (creal (1 + 1)) (cmul ci (sum4 (fun a => sum4 (fun b => sum4 (fun c => sum4 (fun d =>
            creal (lc4 a b c d * up th a b * up th' c d)))))))).
Proof. intros; apply cx_eq; cbv; ring. Qed.

(** [prop:kramers]: for U = [[a, b], [-conj b, conj a]] (the SU(2) form), sigma_y conj(U) sigma_y = U.
    This is the identity behind the Kramers degeneracy and the mirror isospectrality hbar = Theta^-1 h Theta. *)
Definition su2m (a b : cx) (A B : I2) : cx :=
  match A, B with i0, i0 => a | i0, i1 => b | i1, i0 => cneg (cconj b) | i1, i1 => cconj a end.

Theorem su2_pseudoreal : forall (a b : cx) (A B : I2),
  sum2 (fun C => sum2 (fun D => cmul (cmul (sigma e2 A C) (cconj (su2m a b C D))) (sigma e2 D B))) = su2m a b A B.
Proof. intros a b A B; destruct A, B; apply cx_eq; cbv; ring. Qed.

End Generic.
End SpinorIdentities.

(* ===================================================================== *)
(** * Part C. Boundary conformal algebra by exact computation [prop:conformal] *)
(* ===================================================================== *)

Module Conformal.
Open Scope Z_scope.

(* Gaussian integers *)
Record gi := G { gre : Z; gim : Z }.
Definition gadd a b := G (gre a + gre b) (gim a + gim b).
Definition gsub a b := G (gre a - gre b) (gim a - gim b).
Definition gmul a b := G (gre a * gre b - gim a * gim b) (gre a * gim b + gim a * gre b).
Definition gscale (z : Z) a := G (z * gre a) (z * gim a).
Definition gconj a := G (gre a) (- gim a).
Definition gz := G 0 0.  Definition g1 := G 1 0.  Definition gi1 := G 0 1.
Definition geqb a b := Z.eqb (gre a) (gre b) && Z.eqb (gim a) (gim b).

Inductive I4 := e0 | e1 | e2 | e3.
Definition idx := [e0; e1; e2; e3].
Definition M := I4 -> I4 -> gi.
Definition msum (f : I4 -> gi) : gi := fold_right (fun k acc => gadd (f k) acc) gz idx.
Definition mmul (A B : M) : M := fun i j => msum (fun k => gmul (A i k) (B k j)).
Definition madd (A B : M) : M := fun i j => gadd (A i j) (B i j).
Definition msub (A B : M) : M := fun i j => gsub (A i j) (B i j).
Definition mscale (z : Z) (A : M) : M := fun i j => gscale z (A i j).
Definition comm (A B : M) : M := msub (mmul A B) (mmul B A).
Definition mzero : M := fun _ _ => gz.
Definition mid : M := fun i j => match i, j with e0,e0 | e1,e1 | e2,e2 | e3,e3 => g1 | _,_ => gz end.
Definition meqb (A B : M) : bool := forallb (fun i => forallb (fun j => geqb (A i j) (B i j)) idx) idx.
Definition adj (A : M) : M := fun i j => gconj (A j i).

Lemma geqb_eq : forall a b, geqb a b = true -> a = b.
Proof.
  intros [a1 a2] [b1 b2]; unfold geqb; simpl; intros Hb.
  apply andb_true_iff in Hb as [H1 H2]. apply Z.eqb_eq in H1. apply Z.eqb_eq in H2. subst; reflexivity.
Qed.

Lemma meqb_sound : forall A B, meqb A B = true -> forall i j, A i j = B i j.
Proof.
  intros A B Hb i j; unfold meqb in Hb. rewrite forallb_forall in Hb.
  assert (Hi : In i idx) by (destruct i; simpl; tauto). specialize (Hb i Hi).
  rewrite forallb_forall in Hb. assert (Hj : In j idx) by (destruct j; simpl; tauto).
  apply geqb_eq, Hb, Hj.
Qed.

(* Weyl-basis gamma matrices gamma^a = [[0, sigma^a], [sigmabar^a, 0]], sigma = (1, s), sigmabar = (-1, s);
   rows/columns 0,1 = upper block (dual of right atoms), 2,3 = lower block (left atoms). *)
Definition sig (a : I4) (r c : nat) : gi :=
  match a, r, c with
  | e0, 0%nat, 0%nat => g1 | e0, 1%nat, 1%nat => g1
  | e1, 0%nat, 1%nat => g1 | e1, 1%nat, 0%nat => g1
  | e2, 0%nat, 1%nat => G 0 (-1) | e2, 1%nat, 0%nat => gi1
  | e3, 0%nat, 0%nat => g1 | e3, 1%nat, 1%nat => G (-1) 0
  | _, _, _ => gz end.
Definition sigbar (a : I4) (r c : nat) : gi := match a with e0 => gscale (-1) (sig e0 r c) | _ => sig a r c end.
Definition blk (i : I4) : nat := match i with e0 | e1 => 0%nat | e2 | e3 => 1%nat end.
Definition sub (i : I4) : nat := match i with e0 | e2 => 0%nat | e1 | e3 => 1%nat end.
Definition gam (a : I4) : M := fun i j =>
  match blk i, blk j with
  | 0%nat, 1%nat => sig a (sub i) (sub j)
  | 1%nat, 0%nat => sigbar a (sub i) (sub j)
  | _, _ => gz end.
Definition g5 : M := fun i j => match i, j with e0,e0 | e1,e1 => g1 | e2,e2 | e3,e3 => G (-1) 0 | _,_ => gz end.
Definition etaZ (a : I4) : Z := match a with e0 => -1 | _ => 1 end.
Definition gamlow (a : I4) : M := mscale (etaZ a) (gam a).

(* Generators multiplied by 2: Jt_ab = 2 J_ab = gamma^a gamma^b (a <> b), Pt_a = 2 P_a = gamma_a gamma_5 *)
Definition Jt (a b : I4) : M := mmul (gam a) (gam b).
Definition Pt (a : I4) : M := mmul (gamlow a) g5.
Definition Dt : M := Pt e0.
Definition Pc (i : I4) : M := madd (Jt e0 i) (Pt i).
Definition Kc (i : I4) : M := msub (Jt e0 i) (Pt i).
Definition delta (i j : I4) : Z := match i, j with e1,e1 | e2,e2 | e3,e3 => 1 | _,_ => 0 end.
Definition Mt (i j : I4) : M := match i, j with e1,e1 | e2,e2 | e3,e3 => mzero | _,_ => Jt i j end.
Definition spat := [e1; e2; e3].

(* Clifford algebra {gamma^a, gamma^b} = 2 eta^{ab} 1 *)
Definition clifford_ok : bool :=
  forallb (fun a => forallb (fun b =>
    meqb (madd (mmul (gam a) (gam b)) (mmul (gam b) (gam a)))
         (mscale (match a, b with e0,e0 => -2 | e1,e1 | e2,e2 | e3,e3 => 2 | _,_ => 0 end) mid)) idx) idx.

(* [prop:conformal], scaled: [Dt,Pc_i] = 2 Pc_i, [Dt,Kc_i] = -2 Kc_i, [Kc_i,Pc_j] = 4 (Mt_ij - delta_ij Dt),
   [Dt,Mt_ij] = 0, [Pc_i,Pc_j] = [Kc_i,Kc_j] = 0, [Mt_ij, X_k] = 2 (delta_jk X_i - delta_ik X_j) *)
Definition conformal_ok : bool :=
  forallb (fun i =>
    meqb (comm Dt (Pc i)) (mscale 2 (Pc i)) && meqb (comm Dt (Kc i)) (mscale (-2) (Kc i)) &&
    forallb (fun j =>
      meqb (comm (Kc i) (Pc j)) (mscale 4 (msub (Mt i j) (mscale (delta i j) Dt))) &&
      meqb (comm Dt (Mt i j)) mzero && meqb (comm (Pc i) (Pc j)) mzero && meqb (comm (Kc i) (Kc j)) mzero &&
      forallb (fun k =>
        meqb (comm (Mt i j) (Pc k)) (mscale 2 (msub (mscale (delta j k) (Pc i)) (mscale (delta i k) (Pc j)))) &&
        meqb (comm (Mt i j) (Kc k)) (mscale 2 (msub (mscale (delta j k) (Kc i)) (mscale (delta i k) (Kc j))))) spat) spat) spat.

(* block structure: Mt block diagonal and anti-Hermitian; Dt Hermitian and off diagonal; Dt^2 = 1 (eigenvalues of D = Dt/2 are +-1/2) *)
Definition blockdiag (A : M) : bool := forallb (fun i => forallb (fun j =>
  if Nat.eqb (blk i) (blk j) then true else geqb (A i j) gz) idx) idx.
Definition offdiag (A : M) : bool := forallb (fun i => forallb (fun j =>
  if Nat.eqb (blk i) (blk j) then geqb (A i j) gz else true) idx) idx.
Definition structure_ok : bool :=
  forallb (fun i => forallb (fun j => blockdiag (Mt i j) && meqb (adj (Mt i j)) (mscale (-1) (Mt i j))) spat) spat
  && meqb (adj Dt) Dt && offdiag Dt && meqb (mmul Dt Dt) mid.

Theorem clifford_algebra : clifford_ok = true.  Proof. vm_compute. reflexivity. Qed.
Theorem boundary_conformal_algebra : conformal_ok = true.  Proof. vm_compute. reflexivity. Qed.
Theorem conformal_generator_structure : structure_ok = true.  Proof. vm_compute. reflexivity. Qed.

(* Unpacked form of one representative relation, to show what the boolean checks mean *)
Theorem dilatation_translation : forall i, In i spat -> forall r s, comm Dt (Pc i) r s = mscale 2 (Pc i) r s.
Proof.
  intros i Hi r s. apply meqb_sound.
  destruct Hi as [<-|[<-|[<-|[]]]]; vm_compute; reflexivity.
Qed.

End Conformal.

(* ===================================================================== *)
(** * Part D. Field identities over the real numbers                       *)
(* ===================================================================== *)

Module FieldIdentities.
Open Scope R_scope.

(* [eq:scalaron_mass]: for f = alpha R + beta R^2, m_f^2 = (f'/f'' - Rbar)/3 = alpha/(6 beta) *)
Theorem scalaron_mass : forall alpha beta Rb : R, beta <> 0 ->
  ((alpha + 2 * beta * Rb) / (2 * beta) - Rb) / 3 = alpha / (6 * beta).
Proof. intros; field; auto. Qed.

(* Farmer-Larson form f = R (1 - xi Phi^2) + beta R^2 *)
Theorem scalaron_mass_FL : forall xi Phi beta Rb : R, beta <> 0 ->
  (((1 - xi * Phi ^ 2) + 2 * beta * Rb) / (2 * beta) - Rb) / 3 = (1 - xi * Phi ^ 2) / (6 * beta).
Proof. intros; field; auto. Qed.

(* [eq:stelle_masses]: Ricci basis (a = -c, b = 4c; Stelle) and Weyl basis (beta_R = c/3, alpha_W = 2c) agree *)
Theorem stelle_masses : forall c : R, c <> 0 ->
  let a := - c in let b := 4 * c in let betaR := c / 3 in let alphaW := 2 * c in
  1 / (2 * (3 * a + b)) = 1 / (2 * c) /\ 1 / (6 * betaR) = 1 / (2 * c) /\
  - 1 / b = - 1 / (4 * c) /\ - 1 / (2 * alphaW) = - 1 / (4 * c).
Proof.
  intros c Hc a b betaR alphaW; subst a b betaR alphaW.
  repeat split; field; repeat split; auto; lra.
Qed.

Theorem stelle_no_stable_c : forall c : R, ~ (0 < 1 / (2 * c) /\ 0 < - 1 / (4 * c)).
Proof.
  intros c [H1 H2]. destruct (Rlt_or_le 0 c) as [Hc|Hc].
  - assert (0 < 1 / (4 * c)) by (apply Rdiv_lt_0_compat; lra).
    assert (- 1 / (4 * c) = - (1 / (4 * c))) by (field; lra). lra.
  - destruct (Req_dec c 0) as [->|Hc0].
    + replace (2 * 0) with 0 in H1 by ring. unfold Rdiv in H1. rewrite Rinv_0, Rmult_0_r in H1. lra.
    + assert (0 < 1 / (- (2 * c))) by (apply Rdiv_lt_0_compat; lra).
      assert (1 / (2 * c) = - (1 / (- (2 * c)))) by (field; lra). lra.
Qed.

(* [eq:entropy]: 3/(64 pi G Lambda) = S_dS/(64 pi^2), with S_dS = A/(4G) = pi l^2/G and l^2 = 3/Lambda *)
Theorem desitter_entropy : forall G Lam : R, G <> 0 -> Lam <> 0 ->
  let l2 := 3 / Lam in let SdS := 4 * PI * l2 / (4 * G) in
  3 / (64 * PI * G * Lam) = SdS / (64 * PI ^ 2) /\ SdS = 3 * PI / (G * Lam).
Proof.
  intros G Lam HG HL l2 SdS; subst l2 SdS. pose proof PI_RGT_0.
  split; field; repeat split; auto; lra.
Qed.

(* [sec:inflation]: Einstein-frame potential of f = R + R^2/(6 M^2).  With F = f'(R) = 1 + R/(3M^2),
   V = (Mp^2/2) (R F - f)/F^2 = (3/4) M^2 Mp^2 (1 - 1/F)^2. *)
Theorem starobinsky_potential : forall M Mp F : R, M <> 0 -> F <> 0 ->
  let Rc := 3 * M ^ 2 * (F - 1) in let f := Rc + Rc ^ 2 / (6 * M ^ 2) in
  F = 1 + Rc / (3 * M ^ 2) /\ Mp ^ 2 / 2 * (Rc * F - f) / F ^ 2 = 3 / 4 * M ^ 2 * Mp ^ 2 * (1 - 1 / F) ^ 2.
Proof. intros M Mp F HM HF Rc f; subst Rc f. split; field; auto. Qed.

(* [eq:R2coef]: the R^2 coefficient M_P^2/(12 M^2) equals N^2/(288 pi^2 A_s) when A_s = N^2 M^2/(24 pi^2 M_P^2) *)
Theorem R2_coefficient : forall N M Mp : R, N <> 0 -> M <> 0 -> Mp <> 0 ->
  let As := N ^ 2 * M ^ 2 / (24 * PI ^ 2 * Mp ^ 2) in Mp ^ 2 / (12 * M ^ 2) = N ^ 2 / (288 * PI ^ 2 * As).
Proof. intros N M Mp HN HM HP As; subst As. pose proof PI_RGT_0. field; repeat split; auto; lra. Qed.

End FieldIdentities.

(* ===================================================================== *)
(** * Audit                                                                *)
(* ===================================================================== *)
Print Assumptions FockSpace.car_c.
Print Assumptions FockSpace.car_d.
Print Assumptions FockSpace.pauli_c.
Print Assumptions FockSpace.cdag_anticomm.
Print Assumptions FockSpace.c_cdag_anticomm.
Print Assumptions FockSpace.ddag_anticomm.
Print Assumptions FockSpace.cdag_ddag_anticomm.
Print Assumptions FockSpace.molecules_commute.
Print Assumptions FockSpace.molecule_pauli.
Print Assumptions FockSpace.molecule_hardcore.
Print Assumptions FockSpace.molecule_cross.
Print Assumptions FockSpace.molecule_atom_comm_c.
Print Assumptions FockSpace.molecule_atom_comm_d.
Print Assumptions FockSpace.H_diag.
Print Assumptions FockSpace.H_mirror.
Print Assumptions FockSpace.H_Xdag_comm.
Print Assumptions FockSpace.wdw_molecular_solution.
Print Assumptions FockSpace.molecules_at_occ.
Print Assumptions FockSpace.molecules_support.
Print Assumptions FockSpace.distinct_solutions.
Print Assumptions FockSpace.molecules_repeat.
Print Assumptions FockSpace.bcs_solution.
Print Assumptions FockSpace.pair_shift.
Print Assumptions FockSpace.balance_iff.
Print Assumptions FockSpace.zero_point_cancellation.
Print Assumptions FockSpace.no_go.
Print Assumptions FockSpace.kernel_scale.
Print Assumptions FockSpace.stelle_channel_tachyon.
Print Assumptions FockSpace.kernel_molecular.
Print Assumptions SpinorIdentities.eta_from_pairing.
Print Assumptions SpinorIdentities.single_molecule_null.
Print Assumptions SpinorIdentities.tetrad_metric_spinor.
Print Assumptions SpinorIdentities.chiral_trace_times16.
Print Assumptions SpinorIdentities.su2_pseudoreal.
Print Assumptions Conformal.boundary_conformal_algebra.
Print Assumptions Conformal.clifford_algebra.
Print Assumptions Conformal.conformal_generator_structure.
Print Assumptions FieldIdentities.scalaron_mass.
Print Assumptions FieldIdentities.stelle_masses.
Print Assumptions FieldIdentities.stelle_no_stable_c.
Print Assumptions FieldIdentities.desitter_entropy.
Print Assumptions FieldIdentities.starobinsky_potential.
Print Assumptions FieldIdentities.R2_coefficient.
'''
if not os.path.exists('MolecularWDW.v'):
    open('MolecularWDW.v', 'w').write(ROCQ_SRC)
same_v = hashlib.sha256(open('MolecularWDW.v').read().encode()).hexdigest() == hashlib.sha256(ROCQ_SRC.encode()).hexdigest()
AXFREE = ["FockSpace." + n_ for n_ in ["car_c", "car_d", "pauli_c", "cdag_anticomm", "c_cdag_anticomm", "ddag_anticomm",
          "cdag_ddag_anticomm", "molecules_commute", "molecule_pauli", "molecule_hardcore", "molecule_cross",
          "molecule_atom_comm_c", "molecule_atom_comm_d", "H_diag", "H_mirror", "H_Xdag_comm", "wdw_molecular_solution",
          "molecules_at_occ", "molecules_support", "distinct_solutions", "molecules_repeat", "bcs_solution", "pair_shift",
          "balance_iff", "zero_point_cancellation", "no_go", "kernel_scale", "stelle_channel_tachyon", "kernel_molecular"]] + \
         ["SpinorIdentities." + n_ for n_ in ["eta_from_pairing", "single_molecule_null", "tetrad_metric_spinor",
          "chiral_trace_times16", "su2_pseudoreal"]] + \
         ["Conformal." + n_ for n_ in ["boundary_conformal_algebra", "clifford_algebra", "conformal_generator_structure"]]
REALS = ["FieldIdentities." + n_ for n_ in ["scalaron_mass", "stelle_masses", "stelle_no_stable_c", "desitter_entropy",
         "starobinsky_potential", "R2_coefficient"]]
scan_ok = same_v and not re.search(r"\b(Admitted|admit|Axiom|Parameter)\b", ROCQ_SRC) and \
          all(f"Print Assumptions {n_}." in ROCQ_SRC for n_ in AXFREE + REALS)
check("Rocq source: identical to ancillary file, no Admitted/admit/Axiom, every main theorem audited", "R", scan_ok)
coqc = shutil.which('coqc') or shutil.which('rocq') or os.environ.get('COQC_BIN')
if coqc:
    cmd = [coqc, 'MolecularWDW.v'] if not coqc.endswith('rocq') else [coqc, 'compile', 'MolecularWDW.v']
    outv = subprocess.run(cmd, capture_output=True, text=True); txtv = outv.stdout + outv.stderr
    check("Rocq: file compiles", "R", outv.returncode == 0 and "Error" not in txtv)
    blocks = re.split(r"(?=Closed under the global context|Axioms:)", txtv)
    blocks = [b_ for b_ in blocks if b_.startswith(("Closed", "Axioms:"))]
    names = AXFREE + REALS
    reals_allowed = {"ClassicalDedekindReals.sig_forall_dec", "FunctionalExtensionality.functional_extensionality_dep"}
    if len(blocks) >= len(AXFREE) + len(REALS):
        # Print Assumptions output appears in source order: Fock (29), Spinor (5), Conformal (3), Field (6)
        order = re.findall(r"Print Assumptions ([\w.]+)\.", ROCQ_SRC)
        res = dict(zip(order, blocks))
        for n_ in AXFREE:
            check(f"Rocq theorem {n_}: closed under the global context (no axioms)", "R", res.get(n_, "").startswith("Closed"))
        for n_ in REALS:
            found = set(re.findall(r"(ClassicalDedekindReals\.\w+|FunctionalExtensionality\.\w+|\b\w+\.\w+)\s*\n\s*:", res.get(n_, "")))
            check(f"Rocq theorem {n_}: only the real-number library's axioms", "R", res.get(n_, "").startswith("Axioms:") and found <= reals_allowed and len(found) > 0, ", ".join(sorted(found)))
    else:
        check("Rocq: Print Assumptions output parsed", "R", False, f"{len(blocks)} blocks")
else:
    check("Rocq: compilation", "R", None, "no Rocq/Coq toolchain found")
```

**Statistics of the atoms.** We used fermionic atoms. In a pre-geometric setting the spin--statistics theorem, which presupposes a Lorentz-invariant local field theory on a background, does not apply directly, so the choice is a modeling assumption; but it is the choice that makes the whole construction consistent with spin--statistics once Lorentz invariance emerges, since spin-$\tfrac12$ atoms are fermions and the $(\tfrac12,\tfrac12)$ molecules are bosons (Proposition 8). It also gives the superfluid picture of Sec. IV E. Replacing Eq. (11) by canonical commutation relations (Schwinger-type bosonic atoms [75]) gives an equally consistent variant. It leaves the kinematics (apart from Proposition 4), the geometry, and the gravitational and observational results unchanged. In that variant the kernel of ${\hat{H}}$ is infinite dimensional already on a finite graph, the molecules are ideal bosons, and the roles in Proposition 4 are exchanged: two molecules on a single-flavor patch carry the traceless $(1,1)$, and the scale mode arises between flavors or patches. Bosonic atoms with spin $\tfrac12$, however, are at odds with spin--statistics, and in that variant the fermions of the Standard Model would have to emerge, as fermions can emerge from purely bosonic lattice systems [76]. Whether this happens is left to a follow-up study.

**Time.** Since ${\hat{H}}|\Psi\rangle =0$, there is no external time [77, 23]. The natural internal clock is the information flow itself: ${\hat{T}}$ generates transfers between configurations, and its mirror structure singles out an orientation (${/}$ flows counted positive, ${\backslash}$ flows negative). Whether this orientation can serve as an arrow of time, and how relational observables are to be defined, for example by group averaging [27], is open.

**Lorentz invariance and dimension.** A graph selects a preferred frame, and so does a superfluid: the condensate's rest frame (Corollary 2). In laboratory superfluids the low-energy excitations nevertheless see an emergent Lorentzian metric, with the speed of sound playing the role of $c$ [9, 10]; the gapless acoustic branch of the molecular network should play the same role here. Showing that preferred-frame effects are suppressed below current bounds [78], and that the effective dimension is four, are requirements any completion of the model must meet.

**Relation to other programs.** The model shares the use of spinors and ${SU(2)}$ holonomies with loop quantum gravity [5, 6], the use of a Wheeler--DeWitt constraint with quantum cosmology [79, 22], and the emergence of geometry from an underlying many-body system with analog and condensed-matter approaches [10, 9, 11]. Its gravitational sector connects to quadratic gravity [41, 46, 45] and to the MacDowell--Mansouri formulation [43, 44]. A metric and vierbein built as composites of fundamental fermions are the basis of spinor gravity [80, 81], and spacetime as a condensate of pre-geometric quanta is the basis of group-field-theory condensate cosmology [82, 83]. The mirror sector has the structure of thermofield doubling [29, 30]. Several of the following features have antecedents individually; their combination is, as far as we know, specific to this model: the chiral two-species fermionic atoms whose bosonic composite is forced to be a null vector; the mirror-antisymmetric oscillator constraint, in which both ${\hat{T}}$ and ${\hat{V}}$ are indefinite and the constraint is solved exactly with molecules as its generic solutions; the reading of the Cartan structure equations as closure and holonomy of the information-flow operator; the identification of the tetrad with the block of a de Sitter connection that mixes the two atom species; and a molecular condensate whose order parameter carries the local light cone, with its site-local part fixed by the holonomies (Corollary 2).

## X. Conclusions

We have shown that a pre-geometry of fermionic chiral spinor atoms ${/}$, ${\backslash}$ and bosonic molecules ${X}={/}{\backslash}$, governed by a Wheeler--DeWitt constraint of balanced form ${\hat{T}}|\Psi\rangle =-{\hat{V}}|\Psi\rangle$, is internally consistent and exactly solvable at the level of its minimal model. The atoms alone fix the Lorentzian light-cone structure at each patch, with only a conformal scale free. The constraint confines atoms into molecules, cancels all zero-point energy, and admits phase-coherent molecular condensates whose order parameter carries sharp null directions. Dilute molecules obey the oscillator algebra, one condensate is compatible with every connection, and in a curved region the site-local order parameter is the timelike normal, while the light rays live on the bonds. The information-flow operator ${\hat{T}}$ supplies the connection: its closure defects are torsion, its holonomies are curvature, and its spectrum detects curvature gauge invariantly. Because the constraint has no intrinsic scale, quadratic curvature functionals are the natural continuum actions. Built from holonomies and edge molecules, they give Stelle-type gravity (metric weighting), add only a Pontryagin term (chiral weighting), or give exactly general relativity with a cosmological constant (MacDowell--Mansouri weighting, in which the tetrad is the block of a de Sitter connection that mixes ${/}$ and ${\backslash}$). Solar-system tests require the scalaron to be massive, so a linear term must be present at low energies; if the cancellation of Proposition 7 extends to the dynamical theory, that term must come from the molecular condensate. In the $\alpha R+\beta R^2$ language this ties the pre-geometry to Starobinsky-plus-scalar phenomenology through $m_f^2=\alpha/6\beta$. If the continuum dynamics is one of these channels with a heavy scalaron, the model implies luminal tensor waves, no long-range scalar, a scalaron confined below $10^{-4}$ m, and exactly two tensor polarizations at detector frequencies, all consistent with present data and testable by LISA and ground-based detectors. Imposing the observed de Sitter asymptotics as a boundary condition places the model in a dS/CFT framework without changing any of these results; a boundary theory and a matching dictionary remain to be constructed. The future boundary is the limit of divergent atom pairing; the de Sitter algebra carried by the atoms is the boundary conformal algebra, with the $SU(2)$ of ${\hat{T}}$ as boundary rotations and bulk time translation as boundary dilatation; the Einstein--Hilbert coefficient is the de Sitter entropy; and Standard-Model fermions couple to the geometry through the same chiral transporters as the atoms. The $R^2$ channel supplies Starobinsky's inflaton with a pure-number coefficient fixed by the scalar amplitude. If it is the inflaton, the resulting predictions, $n_s\simeq0.960$--$0.967$, $r\simeq0.003$--$0.005$, $w=-1$, and $f_{\rm NL}^{\rm local}\simeq0.015$, will be tested by DESI, SPHEREx, and next-generation CMB polarization experiments such as LiteBIRD; two of them are already under pressure from BAO-inclusive and dark-energy analyses. What remains is to make the transporters dynamical and determine which of these channels the constraint selects. We have argued that this must involve the molecular order parameter; whether fluctuations of dynamical links also contribute is open.

**The physical picture.** Taken together, these results suggest a picture of spacetime as a molecular superfluid. Fermionic chiral spinor atoms ${/}$ and ${\backslash}$, with no space around them, are paired by the constraint into bosonic molecules ${X}={/}{\backslash}$. The molecules obey the oscillator algebra and condense, and the condensate's order parameter carries the local light cone. The atoms' static vacuum energies cancel, and the information flow between patches carries the curvature. If the observed de Sitter asymptotics are imposed, this superfluid fits a dS/CFT framework in which gravity and the Standard-Model fields appear as boundary operators. Its superfluid dynamics (phase stiffness, collective modes) and the selection of a gravitational action remain to be derived. The picture is nonetheless falsifiable. Each of the following would rule out a specific part of it:

-   a third gravitational-wave polarization at detector frequencies;

-   a scalaron range above $0.1$ mm;

-   a tensor-to-scalar ratio outside $r\simeq0.003$--$0.005$ (if the $R^2$ channel is the inflaton);

-   a confirmed departure from $w=-1$.

## Appendix A. Spinor conventions and identities

We collect the identities used in the text; each is verified numerically in the notebook.

**Pairing.** $\epsilon_{AB}$ with $\epsilon_{01}=1$ is numerically equal to $i\sigma_y$. For $L\in{SL(2,\mathbb{C})}$, $L^T\epsilon L=\epsilon$, which is the invariance of Eq. (2). For $U\in{SU(2)}$, $\sigma_y\bar U\sigma_y=U$, which is the identity behind Proposition 6.

**Soldering.** With Eq. (4), ${X}^a=\tfrac{1}{\sqrt2}\operatorname{tr}(\sigma_a{X})$, using $\operatorname{tr}(\sigma_a\sigma_b)=2\delta_{ab}$. For any $2\times2$ matrix, $\epsilon_{AB}\epsilon_{A'B'}{X}^{AA'}{X}^{BB'}=2\det{X}$, and $\det\big(\tfrac{1}{\sqrt2}{X}^a\sigma_a\big)=\tfrac12\big[({X}^0)^2-|\vec {X}|^2\big]$, which gives Eq. (5).

**Lorentz action.** $L\in{SL(2,\mathbb{C})}$ acts on molecules by ${X}\to L{X} L^\dagger$, preserving the determinant and hence $\eta$. The map $L\mapsto\Lambda(L)$, $\Lambda^a{}_b=\tfrac12\operatorname{tr}(\sigma_aL\sigma_bL^\dagger)$, is the double cover ${SL(2,\mathbb{C})}\to SO^+(1,3)$. The infinitesimal form is Eq. (10): boosts ($\theta_{0i}$) are generated by Hermitian $S$ and rotations ($\theta_{ij}$) by anti-Hermitian $S$.

**Dyad identity.** For any $\psi,\phi\in S$, $\psi^A\phi^B-\phi^A\psi^B=\omega(\psi,\phi)\,\epsilon^{AB}$ with $\epsilon^{AB}$ numerically equal to $\epsilon_{AB}$. This is the only identity needed in the proof of Proposition 3.

**Verification of the identities of Appendix A.**

```python
rng = np.random.default_rng(12)
L = rand_sl2c(rng); U = su2(rng)
check("App. A: eps = i sigma_y; L^T eps L = eps; sigma_y conj(U) sigma_y = U", "N",
      np.allclose(EPS, 1j * sy) and np.allclose(L.T @ EPS @ L, EPS) and np.allclose(sy @ U.conj() @ sy, U))
Mr = rng.normal(size=(2, 2)) + 1j * rng.normal(size=(2, 2))
Lam = np.array([[0.5 * np.trace(SIG[a_] @ L @ SIG[b_] @ L.conj().T).real for b_ in range(4)] for a_ in range(4)])
check("App. A: eps eps X X = 2 det X; Lambda(L) in SO+(1,3)", "N",
      np.isclose(np.einsum('ab,cd,ac,bd->', EPS, EPS, Mr, Mr), 2 * np.linalg.det(Mr)) and np.allclose(Lam.T @ eta @ Lam, eta)
      and np.isclose(np.linalg.det(Lam), 1) and Lam[0, 0] >= 1)
check("App. A: dyad identity psi^A phi^B - phi^A psi^B = omega(psi,phi) eps^{AB} (exact)", "S",
      sp.expand(oS * iS.T - iS * oS.T - (oS.T * EPSs * iS)[0] * EPSs) == sp.zeros(2))
```

## Appendix B. Ancillary files

**Notebook** (`molecular_spacetime_``companion.ipynb`). The notebook is an interactive version of this paper. Its markdown cells reproduce the text of each section, generated from the LaTeX source, with the same equation, proposition, table, and figure numbers; code cells placed after each result carry out its checks, labeled [S], [N], [L], or [R] as in Sec. I D. All figures and the entries of Tables I, II, and IV are generated by the notebook, and a fixed random seed makes every number reproducible. It requires Python 3 with NumPy, SciPy, SymPy, and Matplotlib (ipywidgets optional, for the interactive figures); it runs in a standard Jupyter installation and in JupyterLite. A switch at the top skips the slowest symbolic checks.

**Lean file** (`MolecularWDW.lean`). The notebook also writes this file and, if a Lean 4 toolchain ($\geq4.15$) is available, compiles it. It has no dependencies beyond Lean core; its header lists what is and is not proved, and its final section prints the axioms used by each main theorem.

**Rocq file** (`MolecularWDW.v`). An independent formalization of the results in Table VI, using only the Rocq (Coq) standard library; tested with Coq 8.18.0. The notebook writes and, if `coqc` is available, compiles it, and checks the output of `Print Assumptions` for every main theorem.

**Reproducing the formal checks.** Without Lean and Rocq toolchains the notebook reports the two compile checks as skipped, and it can then certify only the static scans. The ancillary files therefore include `verify.sh`, which compiles both files and checks their axiom reports; a `Dockerfile` that provides Coq 8.18 and Lean 4.15; and a GitHub Actions workflow (`verify.yml`) that runs the same script, so that any reader can confirm the formal results independently.
